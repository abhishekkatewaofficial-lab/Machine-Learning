# 📄 SECTION 1: Plain Text File Handling

## What is File Handling?

File handling means **reading from** and **writing to** files on your computer using Python.

Instead of losing data when a program ends, you can **save it to a file** and read it back later.

### Real-world use cases:
- Saving user settings
- Writing logs
- Reading configuration files
- Storing output of your program

## 📋 Syntax Card — The `open()` Function

```python
file = open("filename.txt", mode, encoding="utf-8")
```

### File Modes — Visual Table

| Mode | Name | File Must Exist? | Creates File? | What it Does |
|------|------|:-:|:-:|------|
| `"r"` | Read | Yes | No | Opens for reading. Default mode. |
| `"w"` | Write | No | Yes | Creates new file. **Overwrites** if exists! |
| `"a"` | Append | No | Yes | Adds to end. Never deletes existing content. |
| `"x"` | Exclusive Create | No | Yes | Creates new file. **Fails if file already exists.** |
| `"r+"` | Read + Write | Yes | No | Read AND write without erasing. |
| `"w+"` | Write + Read | No | Yes | Write AND read, but erases existing content. |
| `"rb"` | Read Binary | Yes | No | Read images, PDFs, etc. |
| `"wb"` | Write Binary | No | Yes | Write images, PDFs, etc. |

### Important — Always use `encoding="utf-8"`
```python
# Always add encoding to avoid weird character bugs
open("file.txt", "r", encoding="utf-8")
```
Without `encoding="utf-8"`, special characters (like ä, é, ñ, ₹) may cause errors or show garbage.

## 🔑 The `with` Statement — Always Use This!

There are **two ways** to open a file. One is risky, one is safe:

### Old way (risky — NOT recommended)
```python
file = open("notes.txt", "w")
file.write("Hello")
file.close()   # You MUST remember to close! If code crashes before this, file stays open.
```

### Modern way (safe — ALWAYS use this)
```python
with open("notes.txt", "w", encoding="utf-8") as file:
    file.write("Hello")
# File is AUTOMATICALLY closed here, even if an error occurs inside the block!
```

**`with` automatically closes the file** no matter what. This prevents resource leaks and data corruption.

````python
# DEMO: Difference between with and without with statement

# OLD WAY - risky
file = open("demo_without_with.txt", "w", encoding="utf-8")
file.write("Written without the with statement.")
file.close()  # Must call manually!
print("Old way - file closed:", file.closed)

# MODERN WAY - safe
with open("demo_with_statement.txt", "w", encoding="utf-8") as f:
    f.write("Written using the with statement.")
    print("Inside with block - file closed?", f.closed)  # False (still open)

print("Outside with block - file closed?", f.closed)  # True (auto-closed!)
````

## 📋 Syntax Card — Writing to a File

| Method | Syntax | What it Does |
|--------|--------|--------------|
| `write()` | `file.write("text")` | Writes a single string |
| `writelines()` | `file.writelines(["line1", "line2"])` | Writes a list of strings (no auto newlines!) |

````python
# Writing to a file using write() and writelines()

# --- write() ---
with open("my_notes.txt", "w", encoding="utf-8") as f:
    f.write("Hello, this is my first file!
")  # 
 = new line
    f.write("Python file handling is easy.
")
    f.write("Let us learn more!
")

print("File written using write()")

# --- writelines() ---
lines = [
    "Line 1: Apple
",
    "Line 2: Banana
",
    "Line 3: Cherry
"
]

with open("fruits.txt", "w", encoding="utf-8") as f:
    f.writelines(lines)   # Writes all lines from the list
    # NOTE: writelines() does NOT add 
 automatically!

print("File written using writelines()")
````

## 📋 Syntax Card — Reading from a File

| Method | Syntax | Returns | Best For |
|--------|--------|---------|----------|
| `read()` | `file.read()` | Entire file as one string | Small files |
| `read(n)` | `file.read(10)` | First n characters | Reading in chunks |
| `readline()` | `file.readline()` | One line at a time | Line-by-line reading |
| `readlines()` | `file.readlines()` | List of all lines | When you need all lines as a list |
| `for line in file` | `for line in file:` | Iterates line by line | Large files (memory efficient) |

````python
# Reading a file - all methods demonstrated

print("=" * 40)
print("1. read() - entire file as one string")
print("=" * 40)
with open("my_notes.txt", "r", encoding="utf-8") as f:
    content = f.read()
    print(content)

print("=" * 40)
print("2. read(n) - read only first 10 characters")
print("=" * 40)
with open("my_notes.txt", "r", encoding="utf-8") as f:
    chunk = f.read(10)
    print(repr(chunk))   # repr() shows 
 as literal 


print("=" * 40)
print("3. readline() - one line at a time")
print("=" * 40)
with open("my_notes.txt", "r", encoding="utf-8") as f:
    print(repr(f.readline()))  # Line 1
    print(repr(f.readline()))  # Line 2
    print(repr(f.readline()))  # Line 3

print("=" * 40)
print("4. readlines() - all lines as a list")
print("=" * 40)
with open("my_notes.txt", "r", encoding="utf-8") as f:
    lines = f.readlines()
    print(lines)

print("=" * 40)
print("5. Looping line by line (memory efficient)")
print("=" * 40)
with open("my_notes.txt", "r", encoding="utf-8") as f:
    for i, line in enumerate(f, start=1):
        print(f"Line {i}: {line.strip()}")  # strip() removes trailing 
````

## 📋 Syntax Card — Appending to a File

```python
with open("file.txt", "a", encoding="utf-8") as f:
    f.write("New content added at the end
")
```

- Mode `"a"` = append → **never deletes** existing content
- Mode `"w"` = write → **overwrites** existing content (danger!)

````python
# Appending to a file

print("Before appending:")
with open("my_notes.txt", "r", encoding="utf-8") as f:
    print(f.read())

with open("my_notes.txt", "a", encoding="utf-8") as f:
    f.write("This line was APPENDED later!
")
    f.write("Appending never deletes old content.
")

print("After appending:")
with open("my_notes.txt", "r", encoding="utf-8") as f:
    print(f.read())
````

## 📋 Syntax Card — Cursor Methods (`seek` & `tell`)

When you read a file, Python uses a **cursor** (like a blinking text cursor) that moves as you read.

| Method | Syntax | What it Does |
|--------|--------|--------------|
| `tell()` | `file.tell()` | Returns **current cursor position** (in bytes) |
| `seek(n)` | `file.seek(0)` | **Moves cursor** to position n (0 = beginning) |
| `seek(0, 2)` | `file.seek(0, 2)` | Moves cursor to the **end** of the file |

````python
# seek() and tell() demonstration

with open("my_notes.txt", "r", encoding="utf-8") as f:

    print("Cursor starts at position:", f.tell())  # 0 = beginning

    chunk = f.read(5)  # Read 5 characters - cursor moves forward
    print("Read 5 chars:", repr(chunk))
    print("Cursor is now at position:", f.tell())

    f.seek(0)  # Move cursor back to start
    print("After seek(0) - cursor back at:", f.tell())

    first_line = f.readline()
    print("First line again:", first_line.strip())

    f.seek(0, 2)  # Jump to end of file
    print("After seek(0,2) - cursor at end:", f.tell())
````

## 📋 Syntax Card — `truncate()` and `flush()`

| Method | Syntax | What it Does |
|--------|--------|--------------|
| `truncate()` | `file.truncate()` | Cuts file at current cursor position |
| `truncate(n)` | `file.truncate(20)` | Cuts file to exactly n bytes |
| `flush()` | `file.flush()` | Forces Python to write buffered data to disk immediately |

````python
# truncate() demonstration

with open("truncate_test.txt", "w", encoding="utf-8") as f:
    f.write("ABCDEFGHIJKLMNOPQRSTUVWXYZ")

print("Before truncate:")
with open("truncate_test.txt", "r", encoding="utf-8") as f:
    print(f.read())

with open("truncate_test.txt", "r+", encoding="utf-8") as f:
    f.truncate(10)  # Keep only first 10 characters

print("After truncate(10):")
with open("truncate_test.txt", "r", encoding="utf-8") as f:
    print(f.read())

# flush() demo
print("
flush() demonstration:")
with open("flush_test.txt", "w", encoding="utf-8") as f:
    f.write("Writing data...")
    f.flush()  # Ensures data is physically written to disk right now
    print("Data flushed to disk immediately!")
    f.write(" More data after flush.")
````

## 🔴 Error Handling in File Operations

File operations **fail very often** in real life. Here is what you will encounter:

| Error | When it Happens |
|-------|-----------------|
| `FileNotFoundError` | Trying to read a file that does not exist |
| `PermissionError` | File is locked or you do not have access |
| `IsADirectoryError` | You passed a folder path instead of a file |
| `FileExistsError` | Using mode `"x"` but file already exists |
| `UnicodeDecodeError` | File has characters that cannot be decoded |

## 📋 Syntax Card — `try / except / finally`

```python
try:
    # Code that might fail
    with open("file.txt", "r") as f:
        content = f.read()

except FileNotFoundError:
    # Runs if the specific error occurs
    print("File not found!")

except PermissionError:
    # Runs if a different specific error occurs
    print("No permission to access file!")

except Exception as e:
    # Catches ANY other error
    print("Something went wrong:", e)

finally:
    # ALWAYS runs, whether error happened or not
    print("Done!")
```

````python
# Error Handling - All common file errors

print("Test 1: FileNotFoundError")
print("-" * 30)
try:
    with open("this_file_does_not_exist.txt", "r") as f:
        content = f.read()
except FileNotFoundError as e:
    print(f"Error caught: {e}")

print("
Test 2: FileExistsError (mode x on existing file)")
print("-" * 30)
try:
    with open("my_notes.txt", "x") as f:  # my_notes.txt already exists!
        f.write("test")
except FileExistsError as e:
    print(f"Error caught: {e}")

print("
Test 3: Catching any error with Exception")
print("-" * 30)
try:
    with open("no_such_file.txt", "r") as f:
        print(f.read())
except Exception as e:
    print(f"Something went wrong: {type(e).__name__}: {e}")

print("
Test 4: finally block - always runs!")
print("-" * 30)
try:
    with open("my_notes.txt", "r", encoding="utf-8") as f:
        first_line = f.readline()
        print("Read successfully:", first_line.strip())
except FileNotFoundError:
    print("File not found!")
finally:
    print("finally block runs no matter what!")

print("
Test 5: Safe file read function")
print("-" * 30)

def safe_read(filename):
    """Safely read a file and return its content or None."""
    try:
        with open(filename, "r", encoding="utf-8") as f:
            return f.read()
    except FileNotFoundError:
        print(f"File "{filename}" not found!")
        return None
    except PermissionError:
        print(f"No permission to read "{filename}"!")
        return None

result = safe_read("my_notes.txt")
if result:
    print("Content read successfully!")

result2 = safe_read("missing_file.txt")
print("Result for missing file:", result2)
````

## What is a CSV file?

**CSV = Comma-Separated Values**

A CSV file stores **tabular data** (like a spreadsheet) in plain text:

```
Name,Age,City
Alice,25,Mumbai
Bob,30,Delhi
Charlie,22,Pune
```

### When do you use CSV?
- Storing and sharing structured data (tables)
- Exporting/importing spreadsheets
- Data from databases or APIs

Python has a built-in `csv` module — no installation needed!

## 📋 Syntax Card — CSV Module Overview

```python
import csv
```

| Tool | Use For | Key Methods |
|------|---------|-------------|
| `csv.writer` | Write rows as **lists** | `writerow()`, `writerows()` |
| `csv.reader` | Read rows as **lists** | iterate with `for row in reader` |
| `csv.DictWriter` | Write rows as **dicts** (with column names) | `writeheader()`, `writerow()` |
| `csv.DictReader` | Read rows as **dicts** (column name as key) | iterate with `for row in reader` |

### Important: Always use `newline=""` when opening CSV files!
```python
with open("data.csv", "w", newline="", encoding="utf-8") as f:
    writer = csv.writer(f)
```
Without `newline=""`, you get **blank lines** between every row on Windows. Always include it!

````python
import csv

# Writing a CSV file using csv.writer

# writerow() - write one row at a time
with open("students.csv", "w", newline="", encoding="utf-8") as f:
    writer = csv.writer(f)
    writer.writerow(["Name", "Age", "City", "Score"])  # Header
    writer.writerow(["Alice", 25, "Mumbai", 88])
    writer.writerow(["Bob", 30, "Delhi", 72])
    writer.writerow(["Charlie", 22, "Pune", 95])

print("students.csv written using writerow()")

# writerows() - write multiple rows at once
data = [
    ["Product", "Price", "Stock"],
    ["Apple",  50, 100],
    ["Banana", 20, 200],
    ["Mango",  80, 50],
    ["Orange", 40, 150]
]

with open("products.csv", "w", newline="", encoding="utf-8") as f:
    writer = csv.writer(f)
    writer.writerows(data)  # Writes all rows at once

print("products.csv written using writerows()")
````

````python
# Reading a CSV file using csv.reader

print("Reading students.csv:")
print("=" * 40)
with open("students.csv", "r", encoding="utf-8") as f:
    reader = csv.reader(f)
    for row in reader:
        print(row)   # Each row is a Python list

print("
With row labels:")
print("=" * 40)
with open("students.csv", "r", encoding="utf-8") as f:
    reader = csv.reader(f)
    for i, row in enumerate(reader):
        if i == 0:
            print(f"HEADER: {row}")
        else:
            print(f"Row {i}: {row}")
````

## 📋 Syntax Card — Skipping the Header Row

```python
with open("data.csv", "r") as f:
    reader = csv.reader(f)
    next(reader)   # Skip the header row
    for row in reader:
        print(row)  # Now only data rows
```

`next(reader)` moves the reader forward by one row — effectively skipping the header.

````python
# Skipping header + reading specific columns + counting rows

print("Skipping header row:")
print("=" * 40)
with open("students.csv", "r", encoding="utf-8") as f:
    reader = csv.reader(f)
    next(reader)   # Skip header
    for row in reader:
        print(row)

print("
Reading only Name and Score columns:")
print("=" * 40)
with open("students.csv", "r", encoding="utf-8") as f:
    reader = csv.reader(f)
    next(reader)   # Skip header
    for row in reader:
        name  = row[0]  # Index 0 = Name
        score = row[3]  # Index 3 = Score
        print(f"{name} scored {score}")

print("
Counting rows in students.csv:")
print("=" * 40)
with open("students.csv", "r", encoding="utf-8") as f:
    reader = csv.reader(f)
    next(reader)  # Skip header
    row_count = sum(1 for row in reader)
    print(f"Total data rows: {row_count}")
````

## 📋 Syntax Card — `DictWriter` & `DictReader`

Instead of working with row **lists**, Dict versions let you use **column names as keys** — much more readable!

### DictWriter
```python
writer = csv.DictWriter(f, fieldnames=["Name", "Age", "City"])
writer.writeheader()                                         # Writes the header row
writer.writerow({"Name": "Alice", "Age": 25, "City": "Mumbai"})
```

### DictReader
```python
reader = csv.DictReader(f)
for row in reader:
    print(row["Name"], row["Age"])  # Access by column name!
```

````python
# DictWriter - Write CSV using column names as keys

employees = [
    {"Name": "Riya",   "Department": "HR",      "Salary": 50000},
    {"Name": "Arjun",  "Department": "IT",      "Salary": 75000},
    {"Name": "Sneha",  "Department": "Finance", "Salary": 62000},
    {"Name": "Vikram", "Department": "IT",      "Salary": 80000},
]

with open("employees.csv", "w", newline="", encoding="utf-8") as f:
    fieldnames = ["Name", "Department", "Salary"]
    writer = csv.DictWriter(f, fieldnames=fieldnames)
    writer.writeheader()         # Writes: Name,Department,Salary
    writer.writerows(employees)  # Write all rows from list of dicts

print("employees.csv written using DictWriter")

# DictReader - Read CSV with column names as keys
print("
Reading employees.csv with DictReader:")
print("=" * 40)

with open("employees.csv", "r", encoding="utf-8") as f:
    reader = csv.DictReader(f)
    for row in reader:
        # Access by column name - much cleaner than row[0], row[1]!
        print(f"{row['Name']} works in {row['Department']} | Salary: Rs.{row['Salary']}")
````

````python
# Custom delimiter and appending rows

print("Custom delimiter (semicolon instead of comma):")
print("=" * 40)

with open("semicolon_data.csv", "w", newline="", encoding="utf-8") as f:
    writer = csv.writer(f, delimiter=";")   # Use ; as separator
    writer.writerow(["Name", "Age", "City"])
    writer.writerow(["Pooja", 28, "Chennai"])
    writer.writerow(["Rahul", 35, "Hyderabad"])

with open("semicolon_data.csv", "r", encoding="utf-8") as f:
    reader = csv.reader(f, delimiter=";")
    for row in reader:
        print(row)

print("
Appending new rows to existing CSV:")
print("=" * 40)

print("Before appending:")
with open("students.csv", "r", encoding="utf-8") as f:
    for row in csv.reader(f):
        print(row)

with open("students.csv", "a", newline="", encoding="utf-8") as f:
    writer = csv.writer(f)
    writer.writerow(["Diana", 27, "Kolkata", 91])
    writer.writerow(["Evan",  24, "Jaipur",  78])

print("
After appending:")
with open("students.csv", "r", encoding="utf-8") as f:
    for row in csv.reader(f):
        print(row)
````

# 🔵 SECTION 3: JSON File Handling

## What is JSON?

**JSON = JavaScript Object Notation**

JSON is a text format for storing and sharing data. It looks almost exactly like a Python dictionary!

### Python vs JSON — Type Mapping

| Python Type | JSON Equivalent |
|-------------|----------------|
| `dict` | `{}` object |
| `list` | `[]` array |
| `str` | `"string"` |
| `int` / `float` | number |
| `True` / `False` | `true` / `false` |
| `None` | `null` |

### When do you use JSON?
- Storing settings / configuration
- Communicating with APIs (most APIs use JSON)
- Storing nested/complex data

Python has a built-in `json` module — no installation needed!

## 📋 Syntax Card — JSON Module: 4 Key Functions

| Function | Syntax | Direction | File? |
|----------|--------|-----------|---------|
| `json.dump()` | `json.dump(data, file)` | Python → JSON file | Yes |
| `json.load()` | `json.load(file)` | JSON file → Python | Yes |
| `json.dumps()` | `json.dumps(data)` | Python → JSON string | No (string) |
| `json.loads()` | `json.loads(string)` | JSON string → Python | No (string) |

### Memory trick:
- **`dump`** / **`dumps`** → Python **to** JSON (**D**ump out)
- **`load`** / **`loads`** → JSON **to** Python (**L**oad in)
- With **`s`** at end = works with **s**trings (no file)
- Without `s` = works with **files**

````python
import json

# json.dump() - Write Python dict to a JSON file

student = {
    "name": "Alice",
    "age": 22,
    "city": "Mumbai",
    "is_enrolled": True,
    "score": 88.5,
    "courses": ["Python", "Data Science", "SQL"],
    "address": None
}

with open("student.json", "w", encoding="utf-8") as f:
    json.dump(student, f)

print("student.json written!")

with open("student.json", "r", encoding="utf-8") as f:
    print("
Raw file content (no indent):")
    print(f.read())

# json.load() - Read JSON file back into Python
with open("student.json", "r", encoding="utf-8") as f:
    loaded_data = json.load(f)

print("
Loaded back from JSON:")
print(loaded_data)
print("
Type:", type(loaded_data))
print("Name:", loaded_data["name"])
print("Courses:", loaded_data["courses"])
print("First course:", loaded_data["courses"][0])
````

## 📋 Syntax Card — Pretty Printing JSON

```python
json.dump(data, file, indent=4)              # Write formatted JSON to file
json.dumps(data, indent=4)                   # Formatted JSON as string
json.dumps(data, indent=4, sort_keys=True)   # Also sort keys alphabetically
```

Without `indent`, JSON is saved as one long ugly line. With `indent=4`, it is human-readable.

````python
# Pretty printing JSON with indent and sort_keys

with open("student_pretty.json", "w", encoding="utf-8") as f:
    json.dump(student, f, indent=4)

print("Pretty JSON file content (indent=4):")
with open("student_pretty.json", "r", encoding="utf-8") as f:
    print(f.read())

print("With sort_keys=True (keys in alphabetical order):")
print(json.dumps(student, indent=4, sort_keys=True))
````

````python
# json.dumps() and json.loads() - Working with strings (no file needed)

config = {
    "theme": "dark",
    "language": "en",
    "notifications": True,
    "font_size": 14
}

# json.dumps() - Python dict to JSON string
json_string = json.dumps(config, indent=2)
print("Type of json_string:", type(json_string))  # str
print("JSON string:")
print(json_string)

# json.loads() - JSON string to Python dict
python_dict = json.loads(json_string)
print("
Type of python_dict:", type(python_dict))  # dict
print("Theme:", python_dict["theme"])
print("Font size:", python_dict["font_size"])
````

````python
# Nested JSON - Reading and accessing nested keys

company = {
    "company": "TechCorp",
    "founded": 2010,
    "address": {
        "street": "42 MG Road",
        "city": "Bangalore",
        "pincode": "560001"
    },
    "employees": [
        {"id": 1, "name": "Riya",  "role": "Manager"},
        {"id": 2, "name": "Arjun", "role": "Developer"},
        {"id": 3, "name": "Sneha", "role": "Designer"}
    ],
    "is_active": True
}

with open("company.json", "w", encoding="utf-8") as f:
    json.dump(company, f, indent=4)

with open("company.json", "r", encoding="utf-8") as f:
    data = json.load(f)

print("Company:", data["company"])
print("City:", data["address"]["city"])       # Nested dict access
print("Pincode:", data["address"]["pincode"])

print("
All employees:")
for emp in data["employees"]:               # List of dicts
    print(f"  ID {emp['id']}: {emp['name']} - {emp['role']}")

print("
Second employee:", data["employees"][1]["name"])
````

## 📋 Syntax Card — Handling `JSONDecodeError`

```python
import json

try:
    data = json.loads(some_string)
except json.JSONDecodeError as e:
    print("Invalid JSON:", e)
```

### Common causes of `JSONDecodeError`:

| Mistake | Wrong | Correct |
|---------|-------|--------|
| Single quotes | `{'name': 'Alice'}` | `{"name": "Alice"}` |
| Trailing comma | `[1, 2, 3,]` | `[1, 2, 3]` |
| Unquoted key | `{name: "Alice"}` | `{"name": "Alice"}` |
| Python bool | `True` | `true` |

````python
# Handling JSONDecodeError

bad_json_strings = [
    "{{'name': 'Alice'}}",           # Single quotes - NOT valid JSON!
    "[1, 2, 3,]",                    # Trailing comma
    "{name: "Alice"}",              # Unquoted key
    "{"name": "Alice"}",           # Valid JSON
]

for s in bad_json_strings:
    try:
        result = json.loads(s)
        print(f"Valid JSON: {result}")
    except json.JSONDecodeError as e:
        print(f"Invalid JSON: "{s[:30]}..." Error: {e.msg}")

print("
Safe JSON file reader function:")

def safe_json_read(filename):
    """Safely read a JSON file and return data or None."""
    try:
        with open(filename, "r", encoding="utf-8") as f:
            return json.load(f)
    except FileNotFoundError:
        print(f"File not found: {filename}")
        return None
    except json.JSONDecodeError as e:
        print(f"Invalid JSON in {filename}: {e}")
        return None

data = safe_json_read("student.json")
if data:
    print("Loaded successfully:", data["name"])

bad = safe_json_read("nonexistent.json")
print("Result:", bad)
````

## Common Mistakes to Avoid

| Mistake | Fix |
|---------|-----|
| Forgetting `newline=""` in CSV | Always add `newline=""` in `open()` for CSV writing |
| Using `"w"` instead of `"a"` | `"w"` erases file! Use `"a"` to append |
| Forgetting `encoding="utf-8"` | Always add it to avoid character bugs |
| Single quotes in JSON string | JSON only accepts double quotes |
| Not closing file | Use `with` — it auto-closes |
| Reading non-existent file | Always use `try/except FileNotFoundError` |
| No `indent` in JSON | Use `indent=4` for human-readable JSON files |
