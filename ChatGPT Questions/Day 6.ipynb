```python
import json
sales = [
    ("Laptop", "Alice", 50000),
    ("Mouse", "Bob", 1000),
    ("Laptop", "Charlie", 55000),
    ("Keyboard", "Alice", 2000),
    ("Mouse", "Alice", 1200),
    ("Laptop", "Bob", 52000),
    ("Keyboard", "Charlie", 2500),
    ("Mouse", "Bob", 900)
]


def analyze_sales(sales):

    result = {}

    for product, buyer, price in sales:

        # Create product if it doesn't exist
        if product not in result:
            result[product] = {
                "buyers": [],
                "highest_price": price
            }

        # Add buyer only once
        if buyer not in result[product]["buyers"]:
            result[product]["buyers"].append(buyer)

        # Update highest price
        if price > result[product]["highest_price"]:
            result[product]["highest_price"] = price

    # Sort dictionary

    sorted_result = dict(
        sorted(
            result.items(),
            key=lambda item: (-item[1]["highest_price"], item[0])
        )
    )

    return sorted_result


answer = analyze_sales(sales)
print(json.dumps(answer, indent=4))
```

```python
def analyze_skills(data):

    new_dic = {}

    for item in data:

        name, skills = item.split(':')

        skills = skills.split(",")

        if name not in new_dic:

            new_dic[name] = {
                "primary_skill": skills[0],
                "other_skills": skills[1:],
                "total_skills": len(skills)
            }
  
    return new_dic

data = [
    "Alice:Python,SQL,Git",
    "Bob:Java",
    "Charlie:Python,C++,Docker,AWS",
    "David:SQL,Git",
    "Eva:Python"
]

result = analyze_skills(data)


answer = dict(
            sorted(result.items(),
                key=lambda tup: (-tup[1]["total_skills"], tup[0])
            )
)

print(json.dumps(answer, indent=4))
```

```python
#WAY 2

import json

data = [
    "Alice:Python,SQL,Git",
    "Bob:Java",
    "Charlie:Python,C++,Docker,AWS",
    "David:SQL,Git",
    "Eva:Python"
]


def analyze_skills(data):

    new_dic = {}

    for item in data:

        name, skills = item.split(":")

        skills = skills.split(",")

        first, *rest = skills

        new_dic[name] = {
            "primary_skill": first,
            "other_skills": rest,
            "total_skills": len(skills)
        }

    sorted_dict = dict(
        sorted(
            new_dic.items(),
            key=lambda item: (-item[1]["total_skills"], item[0])
        )
    )

    return sorted_dict


result = analyze_skills(data)

print(json.dumps(result, indent=4))
```

# QSN 3

```python

students = ["Alice", "Bob", "Charlie", "David"]

math = [90, 45, 78, 88]
science = [85, 55, 92, 81]
english = [88, 40, 85, 90]

def analyze_results(students, math, science, english):

    new_dic = {}

    data = zip(students,math,science,english)

    for student_name,math_marks,science_marks,english_marks in data:

        marks = [math_marks,science_marks,english_marks]


        new_dic[student_name] = {
            "marks": marks,
            "average": round(sum(marks)/len(marks), 2),
                "passed_all": all(mark >= 50 for mark in marks)
        }

    return new_dic

result = analyze_results(students, math, science, english)

answer = dict(sorted(result.items(), key= lambda tup: (-tup[1]["average"], tup[0]) ))

display(answer)
```

# QSN 4

```python
def generate_report(*scores, **details):

    result = {

        **details,               
        "scores": list(scores),
        "highest": max(scores),
        "lowest": min(scores),
        "average": round(sum(scores) / len(scores), 2),
        "passed_all": all(score >= 50 for score in scores)

    }

    return result


answer = generate_report(
    90, 85, 78, 92,
    name="Alice",
    department="AI",
    city="Pune"
)

display(answer)
```

```python
import json

emails = [
    "alice@gmail.com",
    "bob@yahoo.com",
    "charlie@gmail.com",
    "david@hotmail.com",
    "eva@gmail.com",
    "frank@yahoo.com",
    "grace@hotmail.com"
]

active = [True, False, True, True, False, True, True]


def analyze_emails(emails, active):

    result = {}

    data = zip(emails, active)

    for email, is_active in data:

        username, domain = email.split("@")

        if domain not in result:

            result[domain] = {
                "total": 0,
                "active_users": 0,
                "users": []
            }

        result[domain]["total"] += 1
        result[domain]["users"].append(username)

        if is_active:
            result[domain]["active_users"] += 1

    sorted_dict = dict(
        sorted(
            result.items(),
            key=lambda item: (-item[1]["total"], item[0])
        )
    )

    return sorted_dict


answer = analyze_emails(emails, active)

display(answer)
print(any(active))
```
