```python
import json
sales = [
    ("Laptop", "Alice", 50000),
    ("Mouse", "Bob", 1000),
    ("Laptop", "Charlie", 55000),
    ("Keyboard", "Alice", 2000),
    ("Mouse", "Alice", 1200),
    ("Laptop", "Bob", 52000),
    ("Keyboard", "Charlie", 2500),
    ("Mouse", "Bob", 900)
]


def analyze_sales(sales):

    result = {}

    for product, buyer, price in sales:

        # Create product if it doesn't exist
        if product not in result:
            result[product] = {
                "buyers": [],
                "highest_price": price
            }

        # Add buyer only once
        if buyer not in result[product]["buyers"]:
            result[product]["buyers"].append(buyer)

        # Update highest price
        if price > result[product]["highest_price"]:
            result[product]["highest_price"] = price

    # Sort dictionary

    sorted_result = dict(
        sorted(
            result.items(),
            key=lambda item: (-item[1]["highest_price"], item[0])
        )
    )

    return sorted_result


answer = analyze_sales(sales)
print(json.dumps(answer, indent=4))
```
