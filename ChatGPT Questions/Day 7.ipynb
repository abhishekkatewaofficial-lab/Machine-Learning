# QSN 1 GENERATORS

```python
lines = [
    "  Python  ",
    "",
    "Data Science",
    "   ",
    "Machine Learning",
    "AI",
    ""
]


def clean_lines(lines):

    for line in lines:

        line = line.strip()

        if line == "":
            continue

        yield line.upper()


gen = clean_lines(lines)

for line in gen:
    print(line)
```

# QSN 2

```python
logs = [
    "# Server Log",
    "INFO: Server Started",
    "",
    "ERROR: Database Connection Failed",
    "WARNING: High Memory Usage",
    "INFO: User Logged In",
    "   ",
    "# End"
]


def error_logs(logs):

    for line_no, line in enumerate(logs, start=1):

        line = line.strip()

        if line == "":
            continue

        if line.startswith("#"):
            continue

        if line.startswith("ERROR"):
            yield f"Line {line_no}: {line}"


gen = error_logs(logs)

for item in gen:
    print(item)
```

# QSN 3

```python
import itertools


learning_rates = [0.1, 0.01]
batch_sizes = [16, 32]
optimizers = ["Adam", "SGD"]


def generate_configs(learning_rates, batch_sizes, optimizers):

    result = {}

    for idx, (lr, batch_size, optimizer) in enumerate(
        itertools.product(learning_rates, batch_sizes, optimizers),
        start=1
    ):

        result[idx] = {
            "lr": lr,
            "batch_size": batch_size,
            "optimizer": optimizer
        }

    return result


answer = generate_configs(learning_rates, batch_sizes, optimizers)

print(answer)
```
