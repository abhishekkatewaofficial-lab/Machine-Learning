# 1. Take a number and print whether it’s positive, negative, or zero.

```python
num = 20

if num < 0:
    print('Negative')
elif num > 0:
    print('Positive')
else:
    print('Zero')
```

# 2. Check if a number is even or odd.

```python
num = 99

if num%2 == 0:
    print('Even')
else :
    print('Odd')
```

# 3. Check if a number is divisible by 5.

```python
num = 99

if num%5 == 0:
    print(f'Yes, {num} is divisible by 5')
else:
    print(f'No, {num} is not divisible by 5')
```

# 4. Check if a number is divisible by both 3 and 5.

```python
num = 15

if num%3 == 0 and num%5 == 0:
    print(f'yes {num} is divisible by both 3 and 5')
else:
    print(f'no {num} is not divisible by both 3 and 5')
```

# 5. Check if a given year is a leap year.

```python
year = 2100

if year%400==0:
    print(f'YEAR {year} is a leap year')

elif year%100==0:
    print(f'YEAR {year} is not a leap year')

elif year%4==0:
    print(f'YEAR {year} is a leap year')

else:
    print(f'YEAR {year} is not a leap year')
```

# 6. Take two numbers and print larger one.

```python
a = 10
b = 10

if a == b:
    print('Both Numbers are same')
elif a > b:
    print("Larger number:", a)
else:
    print("Larger number:", b)
```

# 7. Take three numbers and print largest.

```python
a = 10
b = 11
c = 12

print(f'Largest number is {max(a,b,c)}')
```

# 8. Take a temperature value and print “Cold”, “Warm”, or “Hot” using range conditions.

```python
temp = 31

if temp <= 15:
    print('Its Cold')

elif temp in range(15,31):
    print('Its Warm')

else:
    print('Its Hot')
```

# 9. Take a character and check if it’s a vowel or consonant.

```python
char = 'A'

if char.lower()  in ['a','e','i','o','u']:
    print('Vowel')
else:
    print('Consonent')
```

# 10. Take a character and check whether it’s uppercase, lowercase, a digit, or a special character.

```python
char = '@'

if char.isupper():
    print('UPPERCASE')

elif char.islower():
    print('LOWERCASE')

elif char.isnumeric():
    print('DIGIT')

else:
    print('SPECIALCHAR')
```
