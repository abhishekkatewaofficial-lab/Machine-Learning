# Qsn 2

It should return a dictionary where:

Ignore punctuation (., ,)

Ignore case (Python and python should be same)

Count frequency of each word.

Remove words that appear only once.

Return the dictionary sorted by frequency (highest first).


```python
{
    "python": 4,
    "is": 2,
    "uses": 2,
    "data": 2
}
```

````python
text = """
Python is amazing.
Python is fast.
Data Science uses Python.
AI uses Python and data.
"""

text.replace('.','').split()
````

````python
def analyze_text(text):

    # Remove punctuation, convert to lowercase, split into words
    words = text.replace(".", "").replace(",", "").lower().split()

    # Count frequency
    count_dict = {}

    for word in words:
        count_dict[word] = count_dict.get(word, 0) + 1

    # Sort by frequency (highest first)
    sorted_dict = dict(
        sorted(count_dict.items(), key=lambda item: item[1], reverse=True)
    )

    # Keep only words with frequency > 1
    answer = {}

    for key, value in sorted_dict.items():
        if value > 1:
            answer[key] = value

    return answer


result = analyze_text(text)
print(result)
````

````python
students = [
    ("Alice", "Math"),
    ("Bob", "Science"),
    ("Alice", "Physics"),
    ("Charlie", "Math"),
    ("Bob", "Math"),
    ("Alice", "Math"),
    ("David", "Science"),
    ("Charlie", "Physics"),
    ("Bob", "Science")
]

def analyze_students(students):

    new_dic = {}

    for name,subject in students:

        
        new_dic[name] = new_dic.get(name ,[])
        if subject not in new_dic[name]:
            new_dic[name].append(subject)

    return new_dic


result = analyze_students(students)

print(result.items())

sorted_dict = dict(
    sorted(
        result.items(),
        key=lambda x: (-len(x[1]), x[0])
    )
)

print(sorted_dict)
````

````python
import json
def process_records(records):

    result = {
        "valid": [],
        "invalid": []
    }

    valid_count = 1
    
    for line_no, record in enumerate(records, start=1):

        try:

            name,marks = record.split(',')
            marks = int(marks)

            if marks >= 90:
                grade = "A"
            elif marks >= 80:
                grade = "B"
            elif marks >= 70:
                grade = "C"
            else:
                grade = "D"

            result["valid"].append(
                f"{valid_count}. {name} -> Grade: {grade}"
            )

            valid_count += 1

        except:
            result["invalid"].append(
                f"Line {line_no}: {record}"
            )

    return result

records = [
    "Alice,85",
    "Bob,91",
    "Charlie,abc",
    "David,78",
    "Eva,95",
    "Frank",
    "Grace,88"
]
result = process_records(records)

print(json.dumps(result, indent=3))
````
