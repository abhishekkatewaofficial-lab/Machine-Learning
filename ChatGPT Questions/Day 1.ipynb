## Python Core #01 – Group Anagrams

```python
words = ["eat", "tea", "tan", "ate", "nat", "bat"]

expected output = 
[
    ["eat","tea","ate"],
    ["tan","nat"],
    ["bat"]
]
```

````python
words = ["eat", "tea", "tan", "ate", "nat", "bat"]

groups = {}

for word in words:

    key = ''.join(sorted(word))

    if key not in groups:
        groups[key] = []

    groups[key].append(word)

display(groups)

print(list(groups.values()))
````

## Python Core #02 – Flatten Nested List
```python
lst = [1, [2, [3, 4], 5], [6, 7], 8]

expected output = [1,2,3,4,5,6,7,8]

```

````python
# ANS 1


lst = [1, [2, [3, 4], 5], [6, 6, 7], 8]

result = []

def flattening(lst):
    for item in lst:
        if type(item) != list:

            result.append(item)

        elif type(item) == list:
            flattening(item)

flattening(lst)

print(result)
````

````python
# ANS 2

lst = [1, [2, [3, 4], 5], [6, 6, 7], 8]

result = []

def flattening(lst):
    for item in lst:
        if isinstance(item,list):
            flattening(item)
        else:
            result.append(item)

flattening(lst)

print(result)
````

## Python Core #03 – First Non-Repeating Character

Given a string, return the first character that appears only once.


```python
s = "aabbccddefg"

Output =
"e"

```



````python
# ANS 1


s = "aabbccddefg"

for char in s:

    if s.count(char) == 1:
        print(char)
        break
````

````python
#ANS 2

s = "aabbccddefg"

count = {}

for char in s:

    if char not in count:
        count[char] = 1
    else:
        count[char]+=1

for key,value in count.items():
    if value == 1:
        print(key)
        break
````

## Python Core #04 – Custom Zip Function

Implement your own version of Python's zip().

```python
a = [1,2,3]
b = ['a','b','c','d']

Output =
[(1,'a'),(2,'b'),(3,'c')]

```

````python
# ANS 1

a = [1,2,3]
b = ['a','b','c','d']

print(list(zip(a,b)))
````

````python
# ANS 2

a = [1,2,3]
b = ['a','b','c','d']


result = []
for i in range(min(len(a),len(b))):

    result.append((a[i],b[i]))
print(result)
````

## Question 5 — Rotate Matrix (90° Clockwise)

```python
matrix = [
    [1,2,3],
    [4,5,6],
    [7,8,9]
]

output = [
    [7,4,1],
    [8,5,2],
    [9,6,3]
]

```

````python
matrix = [
    [1,2,3],
    [4,5,6],
    [7,8,9]
]

result = []

n_rows = len(matrix)
n_cols = len(matrix[0])

for col in range(n_cols):
    temp = []

    for row in range(n_rows - 1, -1, -1):
        temp.append(matrix[row][col])

    result.append(temp)

print(result)
````

## Python Core #06 – Most Frequent Word

````python
text = "apple banana apple orange banana apple"

lst = text.split()
count_dict = {}

for word in lst:
    if word not in count_dict:
        count_dict[word] = 1
    else:
        count_dict[word]+=1

max_count = -1

answer = None
for key,value in count_dict.items():
    if value>max_count:
        max_count = value
        answer = key

print(answer)
````

## Python Core #07 – Merge Overlapping Intervals

```python

intervals = [
    [1,3],
    [2,6],
    [8,10],
    [15,18]
]

output = [
    [1,6],
    [8,10],
    [15,18]
]

```

````python
intervals = [
    [1,3],
    [2,6],
    [8,10],
    [15,18]
]

intervals.sort()

result = [intervals[0]]

for current in intervals[1:]:

    last = result[-1]

    if current[0] <= last[1]:
        last[1] = max(last[1], current[1])

    else:
        result.append(current)

print(result)
````

## Python Core #08 – Build Your Own Counter

````python
text = "banana"

counter = {}

for char in text:
    if char not in counter:
        counter[char] = 1
    else:
        counter[char] += 1

print(counter)
````

## Python Core #09 – Remove Consecutive Duplicates

````python
lst = [1,1,2,2,2,3,3,1,1,4]

result = [lst[0]]

for num in lst[1:]:
    if result[-1] != num:
        result.append(num)

print(result)

````
