## Python Core #01 – Group Anagrams

```python
words = ["eat", "tea", "tan", "ate", "nat", "bat"]

expected output = 
[
    ["eat","tea","ate"],
    ["tan","nat"],
    ["bat"]
]
```

````python
words = ["eat", "tea", "tan", "ate", "nat", "bat"]

groups = {}

for word in words:

    key = ''.join(sorted(word))

    if key not in groups:
        groups[key] = []

    groups[key].append(word)

display(groups)

print(list(groups.values()))
````

## Python Core #02 – Flatten Nested List
```python
lst = [1, [2, [3, 4], 5], [6, 7], 8]

expected output = [1,2,3,4,5,6,7,8]

```

````python
# ANS 1


lst = [1, [2, [3, 4], 5], [6, 6, 7], 8]

result = []

def flattening(lst):
    for item in lst:
        if type(item) != list:

            result.append(item)

        elif type(item) == list:
            flattening(item)

flattening(lst)

print(result)
````

````python
# ANS 2

lst = [1, [2, [3, 4], 5], [6, 6, 7], 8]

result = []

def flattening(lst):
    for item in lst:
        if isinstance(item,list):
            flattening(item)
        else:
            result.append(item)

flattening(lst)

print(result)
````

## Python Core #03 – First Non-Repeating Character

Given a string, return the first character that appears only once.


```python
s = "aabbccddefg"

Output =
"e"

```



````python
# ANS 1


s = "aabbccddefg"

for char in s:

    if s.count(char) == 1:
        print(char)
        break
````

````python
#ANS 2

s = "aabbccddefg"

count = {}

for char in s:

    if char not in count:
        count[char] = 1
    else:
        count[char]+=1

for key,value in count.items():
    if value == 1:
        print(key)
        break
````

## Question 5 — Rotate Matrix (90° Clockwise)

```python
matrix = [
    [1,2,3],
    [4,5,6],
    [7,8,9]
]

output = [
    [7,4,1],
    [8,5,2],
    [9,6,3]
]

```

````python
matrix = [
    [1,2,3],
    [4,5,6],
    [7,8,9]
]

result = []

n_rows = len(matrix)
n_cols = len(matrix[0])

for col in range(n_cols):
    temp = []

    for row in range(n_rows - 1, -1, -1):
        temp.append(matrix[row][col])

    result.append(temp)

print(result)
````

## Python Core #06 – Most Frequent Word

````python
text = "apple banana apple orange banana apple"

lst = text.split()
count_dict = {}

for word in lst:
    if word not in count_dict:
        count_dict[word] = 1
    else:
        count_dict[word]+=1

max_count = -1

answer = None
for key,value in count_dict.items():
    if value>max_count:
        max_count = value
        answer = key

print(answer)
````

## Python Core #07 – Merge Overlapping Intervals

```python

intervals = [
    [1,3],
    [2,6],
    [8,10],
    [15,18]
]

output = [
    [1,6],
    [8,10],
    [15,18]
]

```

````python
intervals = [
    [1,3],
    [2,6],
    [8,10],
    [15,18]
]

intervals.sort()

result = [intervals[0]]

for current in intervals[1:]:

    last = result[-1]

    if current[0] <= last[1]:
        last[1] = max(last[1], current[1])

    else:
        result.append(current)

print(result)
````

## Python Core #11 – Find the Longest Consecutive Sequence

Har number ke liye

Agar uska previous number exist karta hai


----Skip kar do

Warna

Ye sequence ki starting hai

Jab tak next number milta rahe


----Count badhao

Longest update karo

````python
nums = [100, 4, 200, 1, 3, 2]

num_set = set(nums)

longest = 0

for num in num_set:

    if num - 1 not in num_set:

        current = num
        length = 1

        while current + 1 in num_set:
            current += 1
            length += 1

        if length > longest:
            longest = length

print(longest)
````

## Python Core #13 – Find Duplicate Elements in a List

````python
def dupli(data):

    result = {}

    for num in data:
        if num not in result:
            result[num] = 1

        else:
            result[num] += 1

    answer = []

    for key,value in result.items():
        if value>1:
            answer.append(key)

    print(answer)

dupli(data = [1, 2, 3, 2, 4, 5, 1, 6, 2])
dupli(data= [5, 5, 5, 5])
dupli(data=[1, 2, 3, 4])
````

## Python Core #14 – Find the Second Largest Number

````python
nums = [10, 20, 4, 45, 99]

largest = float("-inf")
second_largest = float("-inf")

for num in nums:

    if num > largest:
        second_largest = largest
        largest = num

    elif num > second_largest and num != largest:
        second_largest = num

if second_largest == float("-inf"):
    print(None)
else:
    print(second_largest)
````

## Python Core #15 – Merge Two Dictionaries

````python
dict1 = {
    "a": 10,
    "b": 20,
    "c": 30
}

dict2 = {
    "b": 15,
    "c": 5,
    "d": 40
}

for key,value in dict2.items():

    dict1[key] = dict2.get(key,0) + value

print(dict1)
````

## Python Core #20 – Find the First Non-Repeating Character


````python
text = "aabbcdde"

count = {}

# First Pass: Count frequency
for char in text:
    count[char] = count.get(char, 0) + 1

# Second Pass: Find first non-repeating character
answer = None

for char in text:
    if count[char] == 1:
        answer = char
        break

print(answer)
````
