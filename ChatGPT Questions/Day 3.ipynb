```python
import matplotlib.pyplot as plt
```

```python
hours = [1, 2, 3, 4, 5, 6, 7]

marks = [35, 42, 50, 63, 74, 88, 95]

attendance = [60, 65, 72, 80, 84, 90, 98]
```

```python


fig,ax = plt.subplots(
    figsize = (8,5),
    dpi = 100
)



ax.scatter(hours,marks,
        label='Marks',
        c = 'red')



ax.set_xlabel('Hours')
ax.set_ylabel('Marks')
ax.set_title('Padhai')



ax.legend()
ax.grid(linestyle = '--',
        alpha = 0.5,
        axis='y')



plt.show()
```
