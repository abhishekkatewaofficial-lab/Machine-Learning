```python
def greet(name, greeting="Hello"):
    print(f"{greeting}, {name}!")

greet(name='Abhi')
greet(name='Abhi',greeting='Good Morning')


#Hello, Abhi!
#Good Morning, Abhi!
```

```python
def total(*numbers):
    total_sum = 0

    for num in numbers:
        total_sum += num

    return total_sum


result = total(5, 10, 15, 20)
print(result)
```

```python
def show_profile(**details):

    for key,value in details.items():
        print(f'{key:<5} : {value}')



show_profile(name="Abhi", age=20, city="Pune")


```
