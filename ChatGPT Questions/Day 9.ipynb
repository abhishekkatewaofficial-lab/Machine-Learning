# QSN 1


```python
def analyze_number(number):

    sign = 'positive' if number > 0 else 'negative' if number < 0 else 'zero'

    if number > 0:
        parity = 'even' if number % 2 == 0 else 'odd'
    else:
        parity = None

    return {
        'sign': sign,
        'parity': parity
    }


print(analyze_number(10))
print(analyze_number(-7))
print(analyze_number(0))
```

```python
def student_profile(name, *subjects, **details):
    return {
        "name": name,
        "subjects": subjects,
        "details": details
    }


print(student_profile(
    "Abhi",
    "Python",
    "SQL",
    "ML",
    age=20,
    city="Mumbai"
))
```

# QSN 3

```python
{
    "count": 4,
    "all_valid": True,
    "has_perfect": True,
    "average": 86.75
}
```

```python
def validate_scores(*scores):

    count = len(scores)

    # No scores provided
    if count == 0:
        return {
            "count": 0,
            "all_valid": False,
            "has_perfect": False,
            "average": None
        }

    # Check whether every score is between 0 and 100
    all_valid = all(0 <= score <= 100 for score in scores)

    # Check whether at least one score is exactly 100
    has_perfect = any(score == 100 for score in scores)

    # Calculate average only when all scores are valid
    average = sum(scores) / count if all_valid else None

    return {
        "count": count,
        "all_valid": all_valid,
        "has_perfect": has_perfect,
        "average": average
    }


print(validate_scores(80, 95, 100, 72))
print(validate_scores(80, 105, 90))
print(validate_scores())
```
