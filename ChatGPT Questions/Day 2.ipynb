## Problem 1: Longest Consecutive Sequence

```python
nums = [100, 4, 200, 1, 3, 2]

nums_set = set(nums)

longest = 0

for num in nums_set:

    if num - 1 not in nums_set:

        current = num
        length = 1

        while current + 1 in nums_set:

            current += 1
            length += 1

        longest = max(length , longest)

print(longest)

    
```

## Problem 2: Top K Frequent Elements

Difficulty: Intermediate ⭐⭐⭐☆☆

Given an integer array nums and an integer k, return the k most frequent elements.

You may return the answer in any order.

```python
import time

start1  = time.perf_counter()

nums = [1, 1, 1, 2, 2, 3]
k = 2

count = {}

# Count frequency
for num in nums:
    count[num] = count.get(num, 0) + 1

# Sort by frequency (highest first)
count = sorted(count.items(), key=lambda item: item[1], reverse=True)

answer = []

# Take first k elements
for i in range(k):
    answer.append(count[i][0])

print(answer)

end1 = time.perf_counter()-start1
print(end1)
```

```python
import time

start2  = time.perf_counter()

nums = [1, 1, 1, 2, 2, 3]
k = 2

# Step 1: Count frequency
freq = {}

for num in nums:
    freq[num] = freq.get(num, 0) + 1

# Step 2: Create buckets
buckets = [[] for _ in range(len(nums) + 1)]

for num, count in freq.items():
    buckets[count].append(num)

# Step 3: Collect top k frequent elements
result = []

for i in range(len(buckets) - 1, 0, -1):
    for num in buckets[i]:
        result.append(num)

        if len(result) == k:
            print(result)
            break

    if len(result) == k:
        break

end2 = time.perf_counter()-start2
print(end2)
```

## Problem 3: Group Anagrams

Difficulty: Intermediate ⭐⭐⭐☆☆

Given a list of strings strs, group the anagrams together.

Two strings are anagrams if they contain the same characters with the same frequencies.

```python
strs = ["eat", "tea", "tan", "ate", "nat", "bat"]

dic = {}

for word in strs:

    key = "".join(sorted(word))

    if key not in dic:

        dic[key] = []
        dic[key].append(word)

    else:

        dic[key].append(word)

print(list(dic.values()))

```

## Problem 4: Product of Array Except Self

Difficulty: Intermediate ⭐⭐⭐⭐☆

Given an integer array nums, return an array answer such that:

answer[i] is equal to the product of all elements except nums[i].

Do not use division.

Solve it in O(n) time.

```python
nums = [2, 2, 3]

result = []

for i in range(len(nums)):

    product = 1

    for j in range(len(nums)):

        if i !=  j:

            product*=nums[j]

    result.append(product)

print(result)
```

```python
nums = [2, 2, 3]

result = [1] * len(nums)

# Prefix product
prefix = 1

for i in range(len(nums)):
    result[i] = prefix
    prefix *= nums[i]
 
# Suffix product
suffix = 1

for i in range(len(nums) - 1, -1, -1):
    result[i] *= suffix
    suffix *= nums[i]

print(result)
```
