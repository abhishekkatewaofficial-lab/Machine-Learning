```python
from datetime import date
```

```python
today = date.today()
specific = date(2026, 8, 3)   # year, month, day
independence = date(1947, 8, 15) 

print('Today           :', today)
print('Specific date   :', specific)
print('Independence Day:', independence)
```

```python
print('--- Accessing Components ---')
print('Year  :', today.year)
print('Month :', today.month)
print('Day   :', today.day)
```

```python
print('--- Day of the Week ---')
days = ['Monday', 'Tuesday', 'Wednesday', 'Thursday', 'Friday', 'Saturday', 'Sunday']
print('weekday()    :', today.weekday(), '(0=Monday, 6=Sunday)')
print('isoweekday() :', today.isoweekday(), '(1=Monday, 7=Sunday)')
print('Day name     :', days[today.weekday()])
```

```python
from datetime import time

# --- Create time objects ---
t1 = time(14, 30, 0)          # 2:30 PM
t2 = time(9, 0, 0)            # 9:00 AM
t3 = time(23, 59, 59)         # One second before midnight
t4 = time(14, 30, 0, 500000)  # 2:30 PM + 500000 microseconds
midnight = time()             # All zeros = midnight

print('t1 (2:30 PM)  :', t1)
print('t2 (9:00 AM)  :', t2)
print('t3 (11:59 PM) :', t3)
print('t4 (with us)  :', t4)
print('Midnight      :', midnight)
```

```python

print()
print('--- Accessing Components ---')
print('Hour         :', t1.hour)
print('Minute       :', t1.minute)
print('Second       :', t1.second)
print('Microsecond  :', t4.microsecond)
```

```python
from datetime import datetime

# --- Getting current date and time ---
now = datetime.now()          # Local time on your machine
utc = datetime.utcnow()       # UTC time (no timezone info attached)

print('datetime.now()    :', now)
print('datetime.utcnow() :', utc)

print()
print('--- Accessing All Components ---')
print('Year        :', now.year)
print('Month       :', now.month)
print('Day         :', now.day)
print('Hour        :', now.hour)
print('Minute      :', now.minute)
print('Second      :', now.second)
print('Microsecond :', now.microsecond)
print('Weekday     :', now.weekday(), '(0=Mon, 6=Sun)')
```
