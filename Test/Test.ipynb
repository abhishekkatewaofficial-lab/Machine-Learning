```python
import requests
import time
from bs4 import BeautifulSoup
```

```python
headers = {
    'User-Agent': 'Mozilla/5.0 (Windows NT 10.0; Win64; x64) Chrome/120.0.0.0'
}

url = 'https://quotes.toscrape.com'
```

```python
try:
    response = requests.get(url=url,headers=headers,timeout=10)
    response.raise_for_status()

except requests.exceptions.RequestException as e:
    print(f'Failed: {e}')
    response = None
```

```python
if response:

    soup = BeautifulSoup(response.text , 'html.parser')

    #Find all quotes boxes

    quote_boxes = soup.find_all('div',class_ = 'quote')
    print(f'Toatl quotes in this page are: {len(quote_boxes)}')
    print()

    all_quotes = []

    for box in quote_boxes:

        quote = box.find('span',class_ = 'text').get_text(strip=True)
        author = box.find('small',class_ = 'author').get_text(strip = True)
        tags = []
        for tag_box in box.find_all('a',class_ = 'tag'):
            tags.append(tag_box.get_text(strip = True))

        all_quotes.append(
            {
                'quote':quote,
                'author':author,
                'tags':tags   
            }
        )

    for key,value in enumerate(all_quotes , 1):

        print(f'Quote {key}:')
        print(f'    Text  :{value['quote']}')
        print(f'    Author:{value['author']}')
        print(f'    Tags  :{value['tags']}')

        



```

```python
headers = {
    'User-Agent': 'Mozilla/5.0 (Windows NT 10.0; Win64; x64) Chrome/120.0.0.0'
}

url = 'https://books.toscrape.com'
```

```python
try:
    response = requests.get(url=url,headers=headers,timeout=10)
    response.raise_for_status()

except requests.exceptions.RequestException as e:
    print(f'Failure: {e}')
    response = None
```

```python
if response:

    soup = BeautifulSoup(response.text , 'html.parser')

    #Main box of Books

    book_box = soup.find_all('article',class_ = 'product_pod')
    print(f'Found {len(book_box)} books')

    all_books = []

    for box in book_box:

        book_name = box.find('h3').get_text(strip=True)
        book_price = box.find('p',class_ = 'price_color').get_text(strip=True).replace('£','').replace('Â','')
        book_rating = box.find('p','star-rating').get('class')[1]

        all_books.append(
            {
                'book':book_name,
                'price':book_price,
                'rating':book_rating
            }
        )

    print('-'*58)

    for item in all_books:
        print(f'| {item['book']:<40} | {item['price']:<3} | {item['rating']:<6}|')
      

    print('-'*58)

```
