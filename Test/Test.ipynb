```python
import seaborn as sns 
import pandas as pd
import matplotlib.pyplot as plt
```

```python
sns.set_theme(
    style='whitegrid',      # Clean background with grid
    context='notebook',     # Normal scale for Jupyter
    palette='deep',         # Default color palette
    font='DejaVu Sans',     # Cross-platform safe font
    font_scale=1.1,         # Slightly larger text for readability
    rc={
        'figure.facecolor': 'white',    # White figure background
        'axes.facecolor':   'white',    # White axes background
        'grid.alpha':        0.4,       # Subtle grid lines
        'figure.dpi':        120,       # Higher resolution
    }
)
```

```python
tips = sns.load_dataset('tips')
titanic  = sns.load_dataset('titanic')
penguins = sns.load_dataset('penguins')
flights  = sns.load_dataset('flights')
diamonds = sns.load_dataset('diamonds')
```

```python
tips['tip_pct'] = (tips['tip'] / tips['total_bill']) * 100
weekly_avg = tips.groupby('day', observed=True)['tip_pct'].mean().reset_index()
weekly_avg.columns = ['Day', 'Avg Tip %']
```

```python
fig,ax = plt.subplots(figsize = (10, 6))

sns.barplot(
    data=tips,
    x='day',
    y='total_bill',
    hue='sex',
    palette='Set2',
    legend=True,
    ax=ax
)


ax.set_title(
    "Average Total Bill by Day and Gender\n(with 95% Confidence Intervals)",
    fontsize=15,
    fontweight='bold',
    pad=15                  # Space between title and plot
)

ax.set_xlabel('Day of the Week', fontsize=12, labelpad=10)
ax.set_ylabel('Average Total Bill (USD)', fontsize=12, labelpad=10)



ax.tick_params(axis='x', rotation=45, labelsize=11)
ax.tick_params(axis='y', labelsize=11)

ax.legend(title='Customer Gender')
plt.show()
```
