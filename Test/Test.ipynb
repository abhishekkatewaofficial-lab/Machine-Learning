```python
import requests
import time
from bs4 import BeautifulSoup

headers = {
    'User-Agent': 'Mozilla/5.0 (Windows NT 10.0; Win64; x64) Chrome/120.0.0.0'
}

url = 'https://quotes.toscrape.com'

#==============================================================
from pprint import pprint
try:
    response = requests.get(url=url,headers=headers,timeout=10)
    response.raise_for_status()

except:
    print('FAILED')
    response = None

if response:

    soup = BeautifulSoup(response.text, 'html.parser')

#==============================================================

    quote_boxes = soup.find_all('div', class_ = 'quote')

    all_quotes = []

    for box in quote_boxes: 

        quote_tag  = box.find('span', class_ = 'text')
        author_tag = box.find('small', class_ = 'author')
        tags_tag   = box.find_all('a', class_ = 'tag')

        

        quote_text = quote_tag.get_text(strip = True) if quote_boxes else 'N/A'
        author_text = author_tag.get_text(strip=True) if author_tag else 'N/A'
        tags_list  = []
        for tag_text in tags_tag:

            tags_list.append(tag_text.get_text(strip=True))

        all_quotes.append({
            'quote':quote_text,
            'author':author_text,
            'tags':tags_list
        })


    for i, q in enumerate(all_quotes, 1):
            print(f'Quote {i}:')
            print(f'  Text   : {q["quote"][:60]}...')
            print(f'  Author : {q["author"]}')
            print(f'  Tags   : {q["tags"]}')
            print()

        





    
```
