```python
import seaborn as sns

# Folder where CSV files will be saved
folder = "/Users/hemlata/Python Corey Schafer/Notes/Pandas/4. Practice/Data"

# Save every dataset except tips and titanic
for dataset in sns.get_dataset_names():
    if dataset not in ["tips", "titanic"]:
        try:
            df = sns.load_dataset(dataset)
            df.to_csv(f"{folder}/{dataset}.csv", index=False)
            print(f"✓ Saved: {dataset}.csv")
        except Exception as e:
            print(f"✗ Skipped: {dataset} ({e})")

print("\nDone!")
```
