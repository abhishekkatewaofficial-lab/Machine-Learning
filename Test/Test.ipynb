```python
employees = [
    {"name": "Alice", "dept": "AI", "salary": 80000},
    {"name": "Bob", "dept": "Backend", "salary": 65000},
    {"name": "Charlie", "dept": "AI", "salary": 90000},
    {"name": "David", "dept": "Frontend", "salary": 70000},
    {"name": "Eva", "dept": "AI", "salary": 85000},
]
```

```python
def analyze(employees):

    new_dic = {}

    for employee in employees:

        if employee['dept'] not in new_dic.keys():

            new_dic[employee['dept']] = []
            new_dic[employee['dept']].append((employee['salary'],employee['name']))

        else:

            new_dic[employee['dept']].append((employee['salary'],employee['name']))

    return new_dic

result = analyze(employees)

print(result.items())

new_result = sorted(result.items() , key= lambda tup : tup[1], reverse=True)

print(new_result)

```
