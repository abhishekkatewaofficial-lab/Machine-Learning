# Section 1: Introduction — The datetime Module
---

## Why do we need the datetime module?

In real programs you often need to:
- Record **when** something happened (timestamps)
- Calculate **how many days** until an event
- Display dates in a **human-readable format** (e.g. 03 Aug 2026)
- Parse a date **from a string** (e.g. from a CSV or user input)

Python's built-in `datetime` module handles all of this.

## The 5 classes inside `datetime`

| Class | What it represents | Example |
|-------|-------------------|--------|
| `date` | Only a date (no time) | 2026-08-03 |
| `time` | Only a time (no date) | 14:30:00 |
| `datetime` | Both date AND time | 2026-08-03 14:30:00 |
| `timedelta` | A duration / difference | 7 days, 2 hours |
| `timezone` | Timezone offset from UTC | UTC+5:30 |

## How to import

```python
# OR import specific classes (cleaner, recommended)

```

# Section 2: The `date` Class
---

The `date` class stores **only a date** (year, month, day) with no time information.

## Syntax Card — `date`

| Method / Attribute | Syntax | What it Does |
|--------------------|--------|--------------|
| Create today's date | `date.today()` | Returns today's date |
| Create a specific date | `date(2026, 8, 3)` | Year, Month, Day |
| Get year | `d.year` | Returns the year as int |
| Get month | `d.month` | Returns the month as int (1–12) |
| Get day | `d.day` | Returns the day as int (1–31) |
| Day of week (0=Mon) | `d.weekday()` | 0=Monday ... 6=Sunday |
| Day of week (1=Mon) | `d.isoweekday()` | 1=Monday ... 7=Sunday |
| ISO format string | `d.isoformat()` | Returns `'2026-08-03'` |
| Format as string | `d.strftime('%d %b %Y')` | Returns `'03 Aug 2026'` |
| Replace a field | `d.replace(year=2027)` | Returns new date with year changed |

````python
from datetime import date

# --- Create dates ---
today = date.today()
specific = date(2026, 8, 3)   # year, month, day
independence = date(1947, 8, 15)  # India's Independence Day

print('Today           :', today)
print('Specific date   :', specific)
print('Independence Day:', independence)

print()
print('--- Accessing Components ---')
print('Year  :', today.year)
print('Month :', today.month)
print('Day   :', today.day)

print()
print('--- Day of the Week ---')
days = ['Monday', 'Tuesday', 'Wednesday', 'Thursday', 'Friday', 'Saturday', 'Sunday']
print('weekday()    :', today.weekday(), '(0=Monday, 6=Sunday)')
print('isoweekday() :', today.isoweekday(), '(1=Monday, 7=Sunday)')
print('Day name     :', days[today.weekday()])

print()
print('--- Other Methods ---')
print('isoformat()  :', today.isoformat())        # '2026-08-03'
print('replace()    :', today.replace(year=2027))  # Same date, next year
print('timetuple()  :', today.timetuple())          # Named tuple of date components
````

# Section 3: The `time` Class
---

The `time` class stores **only a time** (hour, minute, second, microsecond) with no date.

## Syntax Card — `time`

| Method / Attribute | Syntax | What it Does |
|--------------------|--------|--------------|
| Create a time | `time(14, 30, 0)` | Hour, Minute, Second |
| Create with microseconds | `time(14, 30, 0, 500000)` | Hour, Min, Sec, Microsecond |
| Get hour | `t.hour` | 0–23 |
| Get minute | `t.minute` | 0–59 |
| Get second | `t.second` | 0–59 |
| Get microsecond | `t.microsecond` | 0–999999 |
| ISO format string | `t.isoformat()` | Returns `'14:30:00'` |
| Format as string | `t.strftime('%I:%M %p')` | Returns `'02:30 PM'` |
| Replace a field | `t.replace(hour=9)` | Returns new time with hour changed |

> **Note:** `time(0, 0, 0)` = midnight. Arguments default to 0, so `time()` = midnight.

````python
from datetime import time

# --- Create time objects ---
t1 = time(14, 30, 0)          # 2:30 PM
t2 = time(9, 0, 0)            # 9:00 AM
t3 = time(23, 59, 59)         # One second before midnight
t4 = time(14, 30, 0, 500000)  # 2:30 PM + 500000 microseconds
midnight = time()             # All zeros = midnight

print('t1 (2:30 PM)  :', t1)
print('t2 (9:00 AM)  :', t2)
print('t3 (11:59 PM) :', t3)
print('t4 (with us)  :', t4)
print('Midnight      :', midnight)

print()
print('--- Accessing Components ---')
print('Hour         :', t1.hour)
print('Minute       :', t1.minute)
print('Second       :', t1.second)
print('Microsecond  :', t4.microsecond)

print()
print('--- Other Methods ---')
print('isoformat()  :', t1.isoformat())        # '14:30:00'
print('replace()    :', t1.replace(hour=9))     # Change hour to 9

print()
print('--- Comparing times ---')
print('t2 < t1 ?', t2 < t1)   # 9:00 AM is before 2:30 PM
print('t3 > t1 ?', t3 > t1)   # 11:59 PM is after 2:30 PM
````

# Section 4: The `datetime` Class (Most Important!)
---

The `datetime` class combines **both date and time** into one object.
This is the class you will use most often in real projects.

## Syntax Card — `datetime`

| Method / Attribute | Syntax | What it Does |
|--------------------|--------|--------------|
| Current date + time | `datetime.now()` | Local date and time right now |
| Current UTC time | `datetime.utcnow()` | UTC date and time right now |
| Specific datetime | `datetime(2026, 8, 3, 14, 30, 0)` | y, m, d, h, min, sec |
| From date + time | `datetime.combine(d, t)` | Combine a `date` and `time` object |
| From timestamp | `datetime.fromtimestamp(ts)` | Unix timestamp → datetime |
| Get year/month/day | `.year` `.month` `.day` | Access date components |
| Get hour/min/sec | `.hour` `.minute` `.second` | Access time components |
| Extract date part | `dt.date()` | Returns a `date` object |
| Extract time part | `dt.time()` | Returns a `time` object |
| Replace field | `dt.replace(year=2027)` | Returns new datetime |
| Unix timestamp | `dt.timestamp()` | datetime → Unix timestamp (float) |

````python
from datetime import datetime

# --- Getting current date and time ---
now = datetime.now()          # Local time on your machine
utc = datetime.utcnow()       # UTC time (no timezone info attached)

print('datetime.now()    :', now)
print('datetime.utcnow() :', utc)

print()
print('--- Accessing All Components ---')
print('Year        :', now.year)
print('Month       :', now.month)
print('Day         :', now.day)
print('Hour        :', now.hour)
print('Minute      :', now.minute)
print('Second      :', now.second)
print('Microsecond :', now.microsecond)
print('Weekday     :', now.weekday(), '(0=Mon, 6=Sun)')
````

````python
from datetime import datetime, date, time

# --- Creating specific datetimes ---
dt1 = datetime(2026, 8, 3)               # Date only (time defaults to 00:00:00)
dt2 = datetime(2026, 8, 3, 14, 30, 0)    # Full date + time
dt3 = datetime(2026, 12, 31, 23, 59, 59) # New Year's Eve

print('dt1:', dt1)
print('dt2:', dt2)
print('dt3:', dt3)

print()
print('--- Combining date + time objects ---')
d = date(2026, 8, 15)   # Independence Day
t = time(9, 0, 0)       # 9 AM
combined = datetime.combine(d, t)
print('Combined:', combined)

print()
print('--- Extracting date and time parts ---')
print('dt2.date() :', dt2.date())   # Returns a date object
print('dt2.time() :', dt2.time())   # Returns a time object

print()
print('--- Unix Timestamp ---')
ts = datetime.now().timestamp()     # Float: seconds since Jan 1, 1970
print('Timestamp  :', ts)
print('Back to dt :', datetime.fromtimestamp(ts))  # Convert back

print()
print('--- replace() ---')
print('Original   :', dt2)
print('Next year  :', dt2.replace(year=2027))     # Same date, next year
print('At midnight:', dt2.replace(hour=0, minute=0, second=0))
````

# Section 5: Formatting & Parsing — `strftime` and `strptime`
---

These two methods let you convert between **datetime objects** and **strings**.

```
datetime object  --strftime()-->  string    (Format: dt to string)
string           --strptime()-->  datetime  (Parse:  string to dt)
```

### Memory trick:
- **`strftime`** = string **f**ormat **t**ime → datetime → string
- **`strptime`** = string **p**arse **t**ime → string → datetime

---

## Format Codes — Visual Reference Table

| Code | Meaning | Example Output |
|------|---------|----------------|
| `%Y` | Full year (4 digits) | `2026` |
| `%y` | Short year (2 digits) | `26` |
| `%m` | Month number (zero-padded) | `08` |
| `%B` | Full month name | `August` |
| `%b` | Short month name | `Aug` |
| `%d` | Day of month (zero-padded) | `03` |
| `%A` | Full weekday name | `Monday` |
| `%a` | Short weekday name | `Mon` |
| `%H` | Hour in 24-hour format | `14` |
| `%I` | Hour in 12-hour format | `02` |
| `%M` | Minutes | `30` |
| `%S` | Seconds | `00` |
| `%f` | Microseconds | `000000` |
| `%p` | AM or PM | `PM` |
| `%j` | Day of year (1–366) | `215` |
| `%W` | Week number of year | `31` |
| `%Z` | Timezone name | `UTC` |
| `%c` | Full locale datetime | `Mon Aug  3 14:30:00 2026` |
| `%x` | Locale date | `08/03/26` |
| `%X` | Locale time | `14:30:00` |

````python
from datetime import datetime

dt = datetime(2026, 8, 3, 14, 30, 45)
print('Original datetime object:', dt)
print()

# --- strftime() = datetime to string ---
print('--- strftime() examples ---')
print(dt.strftime('%Y-%m-%d'))           # 2026-08-03
print(dt.strftime('%d/%m/%Y'))           # 03/08/2026  (Indian style)
print(dt.strftime('%d %B %Y'))           # 03 August 2026
print(dt.strftime('%d %b %Y'))           # 03 Aug 2026
print(dt.strftime('%A, %d %B %Y'))       # Monday, 03 August 2026
print(dt.strftime('%I:%M %p'))           # 02:30 PM  (12-hour)
print(dt.strftime('%H:%M:%S'))           # 14:30:45  (24-hour)
print(dt.strftime('%d %b %Y, %I:%M %p')) # 03 Aug 2026, 02:30 PM
print(dt.strftime('%Y/%m/%d %H:%M'))     # 2026/08/03 14:30
print(dt.strftime('Day %j of %Y'))       # Day 215 of 2026
print(dt.strftime('Week %W of %Y'))      # Week 31 of 2026
````

````python
from datetime import datetime

# --- strptime() = string to datetime ---
# Syntax: datetime.strptime(date_string, format_code)
# The format code MUST match the string exactly!

print('--- strptime() examples ---')

s1 = '2026-08-03'
dt1 = datetime.strptime(s1, '%Y-%m-%d')
print(f'String: {s1!r:25} -> datetime: {dt1}')

s2 = '03/08/2026'
dt2 = datetime.strptime(s2, '%d/%m/%Y')
print(f'String: {s2!r:25} -> datetime: {dt2}')

s3 = '03 August 2026'
dt3 = datetime.strptime(s3, '%d %B %Y')
print(f'String: {s3!r:25} -> datetime: {dt3}')

s4 = '03 Aug 2026, 02:30 PM'
dt4 = datetime.strptime(s4, '%d %b %Y, %I:%M %p')
print(f'String: {s4!r:25} -> datetime: {dt4}')

s5 = '2026/08/03 14:30:45'
dt5 = datetime.strptime(s5, '%Y/%m/%d %H:%M:%S')
print(f'String: {s5!r:25} -> datetime: {dt5}')

print()
print('--- After parsing, you can access components ---')
print('Year of dt3  :', dt3.year)
print('Month of dt3 :', dt3.month)
print('Hour of dt4  :', dt4.hour)
````

## Extra: `fromisoformat()` — The Modern & Easier Way to Parse

Python 3.7+ has a much simpler method for parsing ISO format strings:

```python
datetime.fromisoformat('2026-08-03')               # date part only
datetime.fromisoformat('2026-08-03T14:30:00')      # date + time
datetime.fromisoformat('2026-08-03T14:30:00+05:30') # with timezone
```

| Method | When to use |
|--------|-------------|
| `fromisoformat()` | When your string is in standard ISO format (YYYY-MM-DD or YYYY-MM-DDTHH:MM:SS) |
| `strptime()` | When your string is in any other custom format |

> ISO format is the format most APIs, databases, and JSON files use. Learn this — you will see it everywhere!

````python
from datetime import date, datetime

# fromisoformat() is cleaner than strptime for standard formats

print('--- date.fromisoformat() ---')
d1 = date.fromisoformat('2026-08-03')
print('Date only :', d1, '| type:', type(d1))

print()
print('--- datetime.fromisoformat() ---')
dt1 = datetime.fromisoformat('2026-08-03')
dt2 = datetime.fromisoformat('2026-08-03T14:30:00')
dt3 = datetime.fromisoformat('2026-08-03T14:30:00.500000')
print('Date part only   :', dt1)
print('Date + time      :', dt2)
print('With microseconds:', dt3)

print()
print('--- Comparing fromisoformat() vs strptime() ---')
s = '2026-08-03T14:30:00'

# Old way
dt_old = datetime.strptime(s, '%Y-%m-%dT%H:%M:%S')  # Must specify format!
print('strptime()      :', dt_old)

# New way
dt_new = datetime.fromisoformat(s)  # No format needed!
print('fromisoformat() :', dt_new)
print('Same result?    :', dt_old == dt_new)

print()
print('--- Real-world use: parsing data from APIs/JSON ---')
# APIs commonly return timestamps like this:
api_timestamps = [
    '2026-01-15T09:00:00',
    '2026-03-22T14:45:30',
    '2026-07-04T00:00:00',
]
for ts in api_timestamps:
    dt = datetime.fromisoformat(ts)
    print(f'  {ts} -> {dt.strftime("%d %B %Y, %I:%M %p")}')
````

# Section 6: `timedelta` — Date Arithmetic
---

A `timedelta` object represents a **duration** (a difference between two dates/times).

You can use it to:
- **Add** or **subtract** time from a date
- Find the **difference** between two dates

## Syntax Card — `timedelta`

```python
from datetime import timedelta

timedelta(days=0, seconds=0, microseconds=0,
          milliseconds=0, minutes=0, hours=0, weeks=0)
```

| Operation | Example | Result |
|-----------|---------|--------|
| Add days | `date.today() + timedelta(days=7)` | 7 days from now |
| Subtract days | `date.today() - timedelta(days=30)` | 30 days ago |
| Add hours | `datetime.now() + timedelta(hours=5)` | 5 hours from now |
| Difference | `d2 - d1` | Returns a `timedelta` object |
| Total seconds | `td.total_seconds()` | Duration in seconds |
| Days component | `td.days` | Whole days in the duration |
| Seconds component | `td.seconds` | Remaining seconds (after days) |

````python
from datetime import datetime, date, timedelta

today = date.today()
now   = datetime.now()

print('Today:', today)
print()

# --- Adding time ---
print('--- Adding time ---')
print('7 days later         :', today + timedelta(days=7))
print('30 days later        :', today + timedelta(days=30))
print('1 week later         :', today + timedelta(weeks=1))
print('5 hours from now     :', now   + timedelta(hours=5))
print('90 minutes from now  :', now   + timedelta(minutes=90))

print()

# --- Subtracting time ---
print('--- Subtracting time ---')
print('7 days ago   :', today - timedelta(days=7))
print('30 days ago  :', today - timedelta(days=30))
print('1 year ago   :', today - timedelta(days=365))

print()

# --- timedelta attributes ---
td = timedelta(days=2, hours=5, minutes=30)
print('--- timedelta attributes ---')
print('timedelta     :', td)
print('days          :', td.days)
print('seconds       :', td.seconds)   # Only the remaining seconds, not counting days!
print('total_seconds :', td.total_seconds())
````

````python
from datetime import datetime, date, timedelta

# --- Difference between two dates ---
print('--- Difference between two dates ---')

today       = date.today()
independence = date(1947, 8, 15)   # India's Independence Day

diff = today - independence
print('Today              :', today)
print('Independence Day   :', independence)
print('Difference         :', diff)
print('In days            :', diff.days)
print('In years (approx)  :', diff.days // 365)

print()

# --- Days until a future event ---
print('--- Days until a future event ---')

new_year = date(today.year + 1, 1, 1)   # Next New Year
days_left = (new_year - today).days
print(f'Days until New Year {today.year + 1}: {days_left} days')

print()

# --- Difference between two datetimes ---
print('--- Difference between two datetimes ---')
start = datetime(2026, 1, 1, 9, 0, 0)   # 9 AM on Jan 1
end   = datetime(2026, 1, 1, 17, 30, 0) # 5:30 PM on Jan 1

duration = end - start
print('Start            :', start)
print('End              :', end)
print('Duration         :', duration)
print('In hours         :', duration.total_seconds() / 3600)
print('In minutes       :', duration.total_seconds() / 60)
````

## Comparing and Sorting Dates

You can compare `date`, `time`, and `datetime` objects using standard Python operators:

| Operator | Meaning |
|----------|---------|
| `d1 < d2` | d1 is earlier than d2 |
| `d1 > d2` | d1 is later than d2 |
| `d1 == d2` | same date |
| `d1 <= d2` | earlier or same |

You can also use `min()`, `max()`, and `sorted()` directly on a list of dates!

````python
from datetime import date, datetime

d1 = date(2026, 1, 1)
d2 = date(2026, 8, 3)
d3 = date(2025, 12, 31)

print('--- Comparing dates ---')
print('d1 < d2  :', d1 < d2)   # True  (Jan is before Aug)
print('d1 > d3  :', d1 > d3)   # True  (2026 is after 2025)
print('d1 == d2 :', d1 == d2)  # False

print()
print('--- Finding min / max from a list ---')
events = [
    date(2026, 12, 25),   # Christmas
    date(2026, 1, 26),    # Republic Day
    date(2026, 8, 15),    # Independence Day
    date(2026, 10, 2),    # Gandhi Jayanti
    date(2026, 11, 14),   # Children's Day
]

print('Earliest event :', min(events))
print('Latest event   :', max(events))

print()
print('--- Sorting dates chronologically ---')
sorted_events = sorted(events)
for d in sorted_events:
    print(' ', d.strftime('%d %B %Y'))

print()
print('--- Sorting in reverse (latest first) ---')
for d in sorted(events, reverse=True):
    print(' ', d.strftime('%d %B %Y'))

print()
print('--- Finding date closest to today ---')
today = date.today()
future_events = [e for e in events if e >= today]
if future_events:
    next_event = min(future_events)
    print('Next upcoming event :', next_event.strftime('%d %B %Y'))
    print('Days away           :', (next_event - today).days)
````

## Generating a Date Range

Python has no built-in `date_range()` (unlike pandas), but you can easily make one
using a `while` loop and `timedelta`:

```python
current = start_date
while current <= end_date:
    print(current)
    current += timedelta(days=1)   # Move to next day
```

You can also wrap it in a **generator function** so it is reusable.

````python
from datetime import date, timedelta

# --- Basic date range with while loop ---
print('--- All dates in first week of August 2026 ---')
start = date(2026, 8, 1)
end   = date(2026, 8, 7)
current = start
while current <= end:
    print(f'  {current.strftime("%d %B %Y")} ({current.strftime("%A")})')
    current += timedelta(days=1)

print()

# --- Reusable date range generator function ---
def date_range(start, end, step_days=1):
    """Generate all dates from start to end (inclusive)."""
    current = start
    while current <= end:
        yield current
        current += timedelta(days=step_days)

print('--- Every 7 days in August 2026 ---')
for d in date_range(date(2026, 8, 1), date(2026, 8, 31), step_days=7):
    print(f'  {d.strftime("%d %B %Y")}')

print()

# --- Practical: list all weekdays (Mon-Fri) in a month ---
print('--- All weekdays in August 2026 ---')
weekdays = [
    d for d in date_range(date(2026, 8, 1), date(2026, 8, 31))
    if d.weekday() < 5   # 0=Mon ... 4=Fri, 5=Sat, 6=Sun
]
print(f'Total working days in Aug 2026: {len(weekdays)}')
print('First 5:', [d.strftime('%d %b') for d in weekdays[:5]])
````

# Section 7: Timezone Basics
---

## What is a timezone?

The world is divided into **time zones**. When it is 2:30 PM in India (IST),
it is 9:00 AM in the UK (UTC+0), and 4:00 AM in New York (UTC-5).

- **UTC** = Coordinated Universal Time (the global standard baseline)
- **IST** = UTC + 5:30 (India is 5 hours 30 minutes ahead of UTC)

## Naive vs Aware datetimes

| Type | Description | Example |
|------|-------------|--------|
| **Naive** | No timezone info attached | `datetime.now()` |
| **Aware** | Has timezone info attached | `datetime.now(timezone.utc)` |

In real-world apps (APIs, databases), always use **aware** datetimes to avoid confusion.

## Syntax Card — `timezone`

| Method | Syntax | What it Does |
|--------|--------|--------------|
| UTC time | `datetime.now(timezone.utc)` | Current time in UTC |
| Custom offset | `timezone(timedelta(hours=5, minutes=30))` | Create IST timezone |
| Make aware | `dt.replace(tzinfo=timezone.utc)` | Attach timezone to existing datetime |
| Timezone name | `dt.tzname()` | Returns timezone name string |

````python
from datetime import datetime, timezone, timedelta

# --- Naive datetime (no timezone) ---
naive = datetime.now()
print('Naive datetime (no timezone):', naive)
print('tzinfo:', naive.tzinfo)   # None

print()

# --- Aware datetime (with UTC) ---
utc_now = datetime.now(timezone.utc)
print('Aware datetime (UTC):', utc_now)
print('tzinfo:', utc_now.tzinfo)
print('tzname:', utc_now.tzname())

print()

# --- Create custom timezone (IST = UTC+5:30) ---
IST = timezone(timedelta(hours=5, minutes=30))
ist_now = datetime.now(IST)
print('IST datetime:', ist_now)
print('tzname:', ist_now.tzname())

print()

# --- Convert from UTC to IST ---
print('--- Converting UTC to IST ---')
utc_time = datetime.now(timezone.utc)
ist_time  = utc_time.astimezone(IST)
print('UTC time :', utc_time.strftime('%Y-%m-%d %H:%M:%S %Z'))
print('IST time :', ist_time.strftime('%Y-%m-%d %H:%M:%S %Z'))

print()

# --- Make a naive datetime aware ---
naive_dt = datetime(2026, 8, 3, 14, 30, 0)
aware_dt  = naive_dt.replace(tzinfo=timezone.utc)
print('Naive :', naive_dt, '| tzinfo:', naive_dt.tzinfo)
print('Aware :', aware_dt, '| tzinfo:', aware_dt.tzinfo)
````

# Section 8: The `calendar` Module
---

The `calendar` module provides useful utilities for working with **calendars and months**.

## Syntax Card — `calendar`

| Function | Syntax | What it Does |
|----------|--------|--------------|
| Leap year check | `calendar.isleap(2024)` | Returns `True` if leap year |
| Days in month | `calendar.monthrange(2026, 2)` | Returns (weekday of 1st, total days) |
| Month name | `calendar.month_name[8]` | `'August'` |
| Short month | `calendar.month_abbr[8]` | `'Aug'` |
| Day name | `calendar.day_name[0]` | `'Monday'` |
| Short day | `calendar.day_abbr[0]` | `'Mon'` |
| Print month | `calendar.prmonth(2026, 8)` | Prints calendar for Aug 2026 |
| Month as string | `calendar.month(2026, 8)` | Returns calendar as string |

# Section 9: Error Handling
---

Datetime operations can fail when:
- A date value is **invalid** (e.g. month 13, day 32)
- A string **does not match** the format code in `strptime()`
- You try to **subtract aware and naive** datetimes

| Error | When it Happens |
|-------|-----------------|
| `ValueError` | Invalid date values or format mismatch in `strptime()` |
| `TypeError` | Wrong type passed (e.g. string instead of int) |
| `TypeError` | Mixing aware and naive datetimes in arithmetic |

````python
from datetime import datetime, date, timezone

# --- ValueError: Invalid date values ---
print('Test 1: Invalid month')
try:
    d = date(2026, 13, 1)   # Month 13 does not exist!
except ValueError as e:
    print(f'  ValueError: {e}')

print('Test 2: Invalid day')
try:
    d = date(2026, 2, 30)   # Feb 30 does not exist!
except ValueError as e:
    print(f'  ValueError: {e}')

print('Test 3: strptime format mismatch')
try:
    dt = datetime.strptime('03-08-2026', '%Y-%m-%d')  # Format does not match!
except ValueError as e:
    print(f'  ValueError: {e}')

print('Test 4: Mixing aware and naive datetimes')
try:
    aware  = datetime.now(timezone.utc)
    naive  = datetime.now()
    result = aware - naive     # Cannot subtract naive from aware!
except TypeError as e:
    print(f'  TypeError: {e}')

print()
print('--- Safe date parser function ---')

def safe_parse(date_string, fmt):
    """Safely parse a date string. Returns datetime or None."""
    try:
        return datetime.strptime(date_string, fmt)
    except ValueError as e:
        print(f'Could not parse "{date_string}" with format "{fmt}": {e}')
        return None

print(safe_parse('03-08-2026', '%d-%m-%Y'))   # OK
print(safe_parse('99-99-9999', '%d-%m-%Y'))   # Bad values
print(safe_parse('August 3',   '%d-%m-%Y'))   # Format mismatch
````

# Bonus: Humanizing Time — '2 hours ago', 'yesterday'
---

In real apps, instead of showing `2026-08-03 14:30:00`, you want to show
**human-friendly text** like `'2 hours ago'` or `'in 3 days'`.

This is a very common feature in social media, notification systems, and dashboards.

We can build this from scratch using `timedelta`!

````python
from datetime import datetime, timedelta

def time_ago(dt):
    """
    Convert a past datetime to a human-readable 'X ago' string.
    """
    now  = datetime.now()
    diff = now - dt          # timedelta (always positive for past events)
    total_seconds = diff.total_seconds()

    if total_seconds < 60:
        return 'just now'
    elif total_seconds < 3600:
        minutes = int(total_seconds // 60)
        return f'{minutes} minute{"s" if minutes > 1 else ""} ago'
    elif total_seconds < 86400:
        hours = int(total_seconds // 3600)
        return f'{hours} hour{"s" if hours > 1 else ""} ago'
    elif diff.days == 1:
        return 'yesterday'
    elif diff.days < 30:
        return f'{diff.days} days ago'
    elif diff.days < 365:
        months = diff.days // 30
        return f'{months} month{"s" if months > 1 else ""} ago'
    else:
        years = diff.days // 365
        return f'{years} year{"s" if years > 1 else ""} ago'


def time_until(dt):
    """
    Convert a future datetime to a human-readable 'in X' string.
    """
    now  = datetime.now()
    diff = dt - now          # timedelta (positive for future)
    total_seconds = diff.total_seconds()

    if total_seconds < 0:
        return 'already passed'
    elif total_seconds < 60:
        return 'in a few seconds'
    elif total_seconds < 3600:
        minutes = int(total_seconds // 60)
        return f'in {minutes} minute{"s" if minutes > 1 else ""}'
    elif total_seconds < 86400:
        hours = int(total_seconds // 3600)
        return f'in {hours} hour{"s" if hours > 1 else ""}'
    elif diff.days == 1:
        return 'tomorrow'
    elif diff.days < 30:
        return f'in {diff.days} days'
    elif diff.days < 365:
        months = diff.days // 30
        return f'in {months} month{"s" if months > 1 else ""}'
    else:
        years = diff.days // 365
        return f'in {years} year{"s" if years > 1 else ""}'


# --- Test time_ago() ---
now = datetime.now()

past_events = [
    ('Posted 30 seconds ago',  now - timedelta(seconds=30)),
    ('Posted 45 minutes ago',  now - timedelta(minutes=45)),
    ('Posted 3 hours ago',     now - timedelta(hours=3)),
    ('Posted yesterday',       now - timedelta(days=1)),
    ('Posted 10 days ago',     now - timedelta(days=10)),
    ('Posted 3 months ago',    now - timedelta(days=90)),
    ('Posted 2 years ago',     now - timedelta(days=730)),
]

print('--- time_ago() output ---')
for label, dt in past_events:
    print(f'  {label:<30} -> {time_ago(dt)}')

print()

# --- Test time_until() ---
future_events = [
    ('In 2 minutes',  now + timedelta(minutes=2)),
    ('In 5 hours',    now + timedelta(hours=5)),
    ('Tomorrow',      now + timedelta(days=1)),
    ('In 3 weeks',    now + timedelta(weeks=3)),
    ('In 4 months',   now + timedelta(days=120)),
    ('In 2 years',    now + timedelta(days=730)),
]

print('--- time_until() output ---')
for label, dt in future_events:
    print(f'  {label:<20} -> {time_until(dt)}')
````

# Bonus: Saving & Reading Datetimes in Files (CSV + JSON)
---

This connects everything from your **File Handling notebook** with datetime!

## The Problem: datetime is NOT JSON serializable!

```python
import json
from datetime import datetime

data = {'timestamp': datetime.now()}   # datetime object
json.dumps(data)                       # ERROR! TypeError
```

## The Fixes

| Format | Write (datetime → file) | Read (file → datetime) |
|--------|--------------------------|-------------------------|
| **JSON** | Convert to string with `.isoformat()` | Parse back with `fromisoformat()` |
| **CSV** | Convert to string with `.strftime()` | Parse back with `strptime()` |

````python
import json, csv
from datetime import datetime

# ==============================================================
# PART 1: JSON Serialization Problem and Fix
# ==============================================================

print('--- The Problem ---')
data = {'name': 'Alice', 'joined': datetime.now()}
try:
    json.dumps(data)
except TypeError as e:
    print(f'TypeError: {e}')
    print('datetime objects cannot be saved to JSON directly!')

print()
print('--- Fix 1: Convert to isoformat string before saving ---')

# Convert datetime to string
data_safe = {
    'name':      'Alice',
    'joined':    datetime.now().isoformat(),          # datetime -> string
    'last_login': datetime(2026, 7, 31, 9, 0).isoformat()
}

with open('datetime_test.json', 'w', encoding='utf-8') as f:
    json.dump(data_safe, f, indent=4)

print('Saved to datetime_test.json:')
with open('datetime_test.json', 'r', encoding='utf-8') as f:
    print(f.read())

# Read back and convert string to datetime
with open('datetime_test.json', 'r', encoding='utf-8') as f:
    loaded = json.load(f)

# Convert strings back to datetime objects
loaded['joined']     = datetime.fromisoformat(loaded['joined'])
loaded['last_login'] = datetime.fromisoformat(loaded['last_login'])

print('Read back:')
print(f"  name      : {loaded['name']}")
print(f"  joined    : {loaded['joined']} | type: {type(loaded['joined']).__name__}")
print(f"  last_login: {loaded['last_login']} | type: {type(loaded['last_login']).__name__}")

print()

# ==============================================================
# PART 2: CSV with Datetime
# ==============================================================

print('--- Saving datetimes to CSV ---')

events = [
    {'event': 'Meeting with team',  'datetime': datetime(2026, 8, 3,  9, 30)},
    {'event': 'Project deadline',   'datetime': datetime(2026, 8, 10, 17, 0)},
    {'event': 'Product launch',     'datetime': datetime(2026, 9, 1,  10, 0)},
]

# Save to CSV - convert datetime to string
with open('events.csv', 'w', newline='', encoding='utf-8') as f:
    writer = csv.writer(f)
    writer.writerow(['event', 'datetime'])         # Header
    for e in events:
        writer.writerow([
            e['event'],
            e['datetime'].strftime('%Y-%m-%d %H:%M:%S')  # datetime -> string
        ])

print('Saved events.csv')

# Read back and convert string to datetime
print()
print('--- Reading datetimes from CSV ---')
with open('events.csv', 'r', encoding='utf-8') as f:
    reader = csv.DictReader(f)
    for row in reader:
        # Convert string back to datetime
        dt = datetime.strptime(row['datetime'], '%Y-%m-%d %H:%M:%S')
        days_away = (dt.date() - datetime.now().date()).days
        print(f"  {row['event']:<25} | {dt.strftime('%d %b %Y, %I:%M %p')} | {days_away} days away")
````

## Mini Project 1: Age Calculator

Enter a date of birth and get the **exact age** in years, months, and days.

**Concepts used:** `date`, `date.today()`, subtraction, `timedelta`

````python
from datetime import date

def calculate_age(birth_date):
    """
    Calculate exact age from a date of birth.
    Returns a dict with years, months, days.
    """
    today = date.today()

    # Calculate years
    years = today.year - birth_date.year

    # Check if birthday has happened yet this year
    # (month and day of birthday vs today)
    birthday_this_year = (today.month, today.day)
    birthday_month_day = (birth_date.month, birth_date.day)

    if birthday_this_year < birthday_month_day:
        years -= 1   # Birthday not yet reached this year

    # Calculate months
    months = today.month - birth_date.month
    if today.day < birth_date.day:
        months -= 1   # Current day is before birth day in the month
    if months < 0:
        months += 12  # Wrap around (e.g. -3 months -> 9 months)

    # Calculate remaining days
    if today.day >= birth_date.day:
        days = today.day - birth_date.day
    else:
        # Days remaining from last month
        import calendar
        last_month = today.month - 1 if today.month > 1 else 12
        last_month_year = today.year if today.month > 1 else today.year - 1
        days_in_last_month = calendar.monthrange(last_month_year, last_month)[1]
        days = days_in_last_month - birth_date.day + today.day

    # Total days alive
    total_days = (today - birth_date).days

    return {'years': years, 'months': months, 'days': days, 'total_days': total_days}


# --- Test the calculator ---
test_dobs = [
    date(2000, 1, 1),    # Jan 1, 2000
    date(1995, 8, 15),   # Aug 15, 1995
    date(2003, 12, 25),  # Christmas 2003
]

print('Age Calculator')
print('=' * 50)
for dob in test_dobs:
    age = calculate_age(dob)
    print(f'DOB: {dob.strftime("%d %B %Y")}')
    print(f'  Age       : {age["years"]} years, {age["months"]} months, {age["days"]} days')
    print(f'  Total days: {age["total_days"]:,} days alive')
    print()

# Try with YOUR date of birth!
print('--- Try it yourself ---')
my_dob = date(2000, 6, 15)  # Change this to your DOB!
result = calculate_age(my_dob)
print(f'Your age: {result["years"]} years, {result["months"]} months, {result["days"]} days')
````

## Mini Project 2: Birthday Countdown

Enter a birthday and find out **how many days until the next one**.
Also shows which day of the week the birthday will fall on!

**Concepts used:** `date`, `date.replace()`, `timedelta`, `strftime()`

````python
from datetime import date

def birthday_countdown(birth_date):
    """
    Calculate days until the next birthday.
    """
    today = date.today()

    # Set birthday to current year
    next_birthday = birth_date.replace(year=today.year)

    # If birthday already passed this year, use next year
    if next_birthday < today:
        next_birthday = birth_date.replace(year=today.year + 1)

    days_left = (next_birthday - today).days

    return {
        'next_birthday': next_birthday,
        'days_left':     days_left,
        'day_of_week':   next_birthday.strftime('%A'),
        'formatted':     next_birthday.strftime('%d %B %Y')
    }


# --- Test the countdown ---
test_birthdays = [
    date(1999, 8, 15),
    date(2001, 1, 1),
    date(1998, 12, 31),
    date(2002, 3, 10),
]

print('Birthday Countdown')
print('=' * 50)
for bday in test_birthdays:
    info = birthday_countdown(bday)

    if info['days_left'] == 0:
        msg = 'TODAY! Happy Birthday!'
    elif info['days_left'] == 1:
        msg = 'TOMORROW!'
    else:
        msg = f'{info["days_left"]} days away'

    print(f'Birthday: {bday.strftime("%d %b")}')
    print(f'  Next birthday : {info["formatted"]} ({info["day_of_week"]})')
    print(f'  Countdown     : {msg}')
    print()

# Try with your birthday!
print('--- Your Birthday ---')
my_bday = date(2000, 6, 15)  # Change this!
info = birthday_countdown(my_bday)
print(f'Next birthday: {info["formatted"]} ({info["day_of_week"]})')
print(f'Days left    : {info["days_left"]}')
````

## Mini Project 3: Event Timer

Enter any event date and the program tells you:
- If it is **in the past** — how long ago it was
- If it is **today** — happy event day!
- If it is **in the future** — how long until it happens

**Concepts used:** `datetime`, `timedelta`, `strftime()`, comparison operators

````python
from datetime import datetime

def event_timer(event_name, event_datetime):
    """
    Tell you how far in the past or future an event is.
    """
    now  = datetime.now()
    diff = event_datetime - now   # positive = future, negative = past
    total_seconds = diff.total_seconds()

    print(f'Event   : {event_name}')
    print(f'Date    : {event_datetime.strftime("%d %B %Y, %I:%M %p")}')

    if abs(total_seconds) < 60:      # Within 1 minute
        print('Status  : Happening RIGHT NOW!')

    elif total_seconds > 0:          # Future event
        days    = diff.days
        hours   = int((total_seconds % 86400) // 3600)
        minutes = int((total_seconds % 3600) // 60)

        print(f'Status  : UPCOMING')

        if days >= 365:
            years  = days // 365
            months = (days % 365) // 30
            print(f'Countdown : {years} year(s), {months} month(s) to go')
        elif days >= 30:
            months = days // 30
            print(f'Countdown : {months} month(s), {days % 30} day(s) to go')
        elif days >= 1:
            print(f'Countdown : {days} day(s), {hours} hour(s) to go')
        else:
            print(f'Countdown : {hours} hour(s), {minutes} minute(s) to go')

    else:                            # Past event
        days  = abs(diff.days)
        print(f'Status  : PAST')

        if days >= 365:
            years = days // 365
            print(f'Happened : {years} year(s) ago')
        elif days >= 30:
            months = days // 30
            print(f'Happened : {months} month(s) ago')
        elif days >= 1:
            print(f'Happened : {days} day(s) ago')
        else:
            hours = int(abs(total_seconds) // 3600)
            print(f'Happened : {hours} hour(s) ago')

    print()


# --- Test with various events ---
print('=' * 50)
print('          EVENT TIMER')
print('=' * 50)
print()

events = [
    ('Moon Landing',          datetime(1969, 7, 20, 20, 17, 0)),
    ('India Independence',    datetime(1947, 8, 15, 0, 0, 0)),
    ('New Year 2027',         datetime(2027, 1, 1, 0, 0, 0)),
    ('Christmas 2026',        datetime(2026, 12, 25, 0, 0, 0)),
]

for name, dt in events:
    event_timer(name, dt)

# Add your own event!
my_event = datetime(2027, 8, 3, 9, 0, 0)  # Change this!
event_timer('My Custom Event', my_event)
````

## What you learned

| Topic | Key Takeaway |
|-------|--------------|
| `date` | Works with just dates; use `date.today()` and `date(y, m, d)` |
| `time` | Works with just times; use `time(h, m, s)` |
| `datetime` | Combines both; use `datetime.now()` most often |
| `strftime` | Converts datetime **to** a string (f = format) |
| `strptime` | Converts a string **to** datetime (p = parse) |
| `timedelta` | Represents duration; use for date arithmetic |
| `timezone` | Always use aware datetimes in real apps |
| `calendar` | Useful for leap year checks, month/day names |
