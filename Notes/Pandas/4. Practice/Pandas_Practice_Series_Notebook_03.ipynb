### 🐧 Business Scenario

The wildlife research team wants a species-level summary of body mass and flipper length to compare the three penguin species in the Palmer Archipelago.

Work with the `penguins` dataset. Remove rows where `body_mass_g` or `flipper_length_mm` is missing. Compute `body_mass_kg` = `body_mass_g` / 1000, rounded to 2 dp. Group by `species` and compute: mean `body_mass_kg` → `avg_body_mass_kg` (rounded to 2 dp), mean `flipper_length_mm` → `avg_flipper_mm` (rounded to 1 dp), count → `penguin_count`. Sort by `avg_body_mass_kg` descending. Reset the index.

### 📝 Problem Statement

1. Drop rows where `body_mass_g` or `flipper_length_mm` is null
2. Compute `body_mass_kg` = `body_mass_g` / 1000, rounded to 2 dp
3. Group by `species`, compute:
   - `avg_body_mass_kg` — mean of `body_mass_kg`, rounded to 2 dp
   - `avg_flipper_mm` — mean of `flipper_length_mm`, rounded to 1 dp
   - `penguin_count` — count of rows
4. Sort by `avg_body_mass_kg` descending
5. Reset index

### 📊 Expected Output Format

| species | avg_body_mass_kg | avg_flipper_mm | penguin_count |
|---|---|---|---|
| Gentoo | ... | ... | ... |
| Chinstrap | ... | ... | ... |
| Adelie | ... | ... | ... |

### 🧠 Concepts Tested

```
dropna (multi-col)  +  Computed Column (unit conversion)  +  GroupBy (1-key)  +  Named Aggregation (mean×2 + count)  +  round() ×2  +  sort_values  +  reset_index
```

<details>
<summary>💡 Hint — Click to expand</summary>

<br>

```python
df = penguins.dropna(subset=["body_mass_g","flipper_length_mm"]).copy()
df["body_mass_kg"] = (df["body_mass_g"] / 1000).round(2)
result = df.groupby("species").agg(
    avg_body_mass_kg = ("body_mass_kg",      "mean"),
    avg_flipper_mm   = ("flipper_length_mm",  "mean"),
    penguin_count    = ("body_mass_kg",        "count")
).reset_index()
result["avg_body_mass_kg"] = result["avg_body_mass_kg"].round(2)
result["avg_flipper_mm"]   = result["avg_flipper_mm"].round(1)
result = result.sort_values("avg_body_mass_kg", ascending=False).reset_index(drop=True)
```

```python
# Only removes a row if another row has the exact same values
# in ALL four of these columns.

    subset=["species","island","sex","body_mass_g"],
    keep="first"
)

# Build reference table from the deduplicated data:
species_ref = (
    df.groupby("species")
      .agg(
          species_avg_mass = ("body_mass_g",   "mean"),
          species_avg_bill = ("bill_length_mm", "mean"),
          species_total    = ("species",        "count")
      )
      .round(2)
      .reset_index()
)

merged["is_above_species_avg"] = merged["body_mass_g"] > merged["species_avg_mass"]

result["above_avg_pct"]= (result["above_avg_count"]/result["n"]*100).round(1)
result = result.sort_values(["species","n"],
```

```python
combined["peak_quarter"]  = combined[q_cols].idxmax(axis=1)
```

```python
# pct_change() = (current - previous) / previous per group:
df["mom_rate"] = df.groupby("year")["passengers"].pct_change()
# Jan → NaN (no prior month in this year)

df["mom_pct"]  = (df["mom_rate"] * 100).round(1)
df = df[df["mom_rate"].notna()]   # remove Jan rows

# vs shift(1) — same formula, one step:
# pct_change() = (s - s.shift(1)) / s.shift(1)
```

## ⚡ D — Quick-Reference Formula Table

| Goal | Formula | Notes |
|---|---|---|
| Fixed-width bins | `pd.cut(col, bins=[edges], labels=[...])` | Specify edges or n bins |
| Equal-freq bins | `pd.qcut(col, q=4, labels=[...])` | Each bin ≈ same # rows |
| Long → wide | `groupby([k1,k2]).sum().unstack(fill_value=0)` | Last key → col headers |
| Wide → long | `pd.melt(df, id_vars=[...], value_vars=[...])` | Collapses cols to rows |
| Frequency table | `pd.crosstab(row, col)` | Cleaner than pivot for counts |
| Row % table | `pd.crosstab(row, col, normalize="index") × 100` | Each row sums to 100 |
| Group broadcast | `groupby(k)["col"].transform("mean")` | Returns same-length Series |
| Lag within group | `groupby(k)["col"].shift(1)` | Sort FIRST! |
| MoM growth rate | `groupby(k)["col"].pct_change() × 100` | Sort FIRST! NaN for 1st row |
| Running total | `groupby(k)["col"].cumsum()` | Sort FIRST! |
| Rank within group | `groupby(k)["col"].rank(method="dense", ascending=False)` | 1 = top |
| All rows for group | `groupby(k).size()` | Includes NaN rows |
| Non-null count | `groupby(k)["col"].count()` | Skips NaN in that col |
| Boolean group flag | `groupby(k).agg(flag=("bool_col","any"))` | any / all |
| Group-level filter | `groupby(k).filter(lambda g: len(g) >= 10)` | Keeps/drops WHOLE groups |
| Top-N full rows | `groupby(k).apply(lambda g: g.nlargest(n,"col"))` | All columns |
| Max row per group | `groupby(k)["col"].idxmax()` → `df.loc[idx]` | Returns index label |
| Peak column | `df[cols].idxmax(axis=1)` | Per-row column name of max |
| datetime parse | `pd.to_datetime(s, format="%Y-%b")` | "%Y-%b" → 1949-Jan |
| Period string | `date.dt.to_period("M").astype(str)` | "1949-01" |
| MultiIndex group | `multi_df.groupby(level=0)["col"].transform("sum")` | No reset needed |
| Row sum | `df[cols].sum(axis=1)` | axis=1 → across columns |
| Row mean | `df[cols].mean(axis=1)` | axis=1 → across columns |
| Label column | `np.where(condition, "Yes", "No")` | Vectorised if-else |
| Stack 3 tables | `pd.concat([a,b,c], ignore_index=True)` | Same columns required |
| Dedup subset | `drop_duplicates(subset=[cols], keep="first")` | Or keep="last" |
| Range filter | `.between(lo, hi)` | inclusive both ends |
| Cap values | `.clip(lower=lo, upper=hi)` | Replaces out-of-range values |

## 🔑 E — The 6 Golden Rules (Always Apply)

| Rule | When |
|---|---|
| **Sort before shift/cumsum/pct_change** | Any lag or running-total operation |
| **Sort before unstack** | Ensures correct long→wide order |
| **fillna(False) before any/all** | Boolean columns with NaN |
| **reset_index(drop=True) after apply()** | Flatten MultiIndex from apply |
| **dropna(subset=[...]) on mass_tertile** | After pd.qcut — edge rows may be NaN |
| **columns.name = None after unstack** | Removes axis label from column index |
