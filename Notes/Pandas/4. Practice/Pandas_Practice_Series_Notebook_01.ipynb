### 📝 Problem Statement

From the `employees` dataset, filter only **Active** employees, select the columns `emp_id`, `employee_name`, `dept_id`, `salary_inr`, and `city`, then sort the result by `salary_inr` in **descending** order.

### ✅ Requirements

1. Filter rows where `status == 'Active'`
2. Select only the 5 specified columns (in that order)
3. Sort by `salary_inr` descending
4. Reset the index

### 📊 Expected Output Format

| index | emp_id | employee_name | dept_id | salary_inr | city |

Before any analysis, the Data Engineering team performs a **data quality audit**. They need a summary report showing, for each column in the `employees` dataset, exactly how many values are missing and what percentage of the total rows that represents — sorted from the most missing to the least.

### 📝 Problem Statement

Using the `employees` DataFrame, produce a **missing value audit report** showing:

- `missing_count` — number of missing values in each column
- `missing_pct` — percentage of missing values, **rounded to 2 decimal places**

Include **only columns that have at least 1 missing value**. Sort the result by `missing_pct` in **descending** order.

### ✅ Requirements

1. Compute missing count per column
2. Compute missing percentage per column (relative to total rows)
3. Round percentage to 2 decimal places
4. Filter out columns with zero missing values
5. Sort by percentage descending

### 📊 Expected Output Format

| column | missing_count | missing_pct |
|---|---|---|
| salary_inr | ... | ...% |
| dept_id | ... | ...% |
| ... | ... | ... |

### 🧠 Concepts Tested

```
isnull() + sum() + Arithmetic on Series + round() + Boolean Filtering + sort_values()
```

````python
missing_count = employees.isna().sum().to_frame(name='missing_count')
missing_pct = (employees.isna().mean() * 100).round(2).to_frame('missing_pct')

missing_report = (
    missing_count
    .join(missing_pct)
    .sort_values(by='missing_count', 
                 ascending=False)
)

missing_report
````

In the `customers` DataFrame:

1. **Clean** the `customer_name` column: strip leading/trailing spaces and convert to **Title Case**
2. **Drop exact duplicate rows** (same `customer_id` appearing more than once)
3. Count the number of **unique customers per city** (using the deduplicated data)

### 🏢 Business Scenario

The Finance team is auditing **high-value orders**. They want to see only the orders that were:
- Successfully **Delivered**
- Paid via **Credit Card** or **UPI**
- With an order amount greater than **₹50,000**

Results should be sorted by `amount_inr` descending, and only specific columns should be shown.

### 📝 Problem Statement

From the `orders` DataFrame, apply all three filters simultaneously and return only the columns:
`order_id`, `customer_id`, `product_id`, `order_date`, `amount_inr`, `payment_method`

Sort by `amount_inr` descending. Reset the index.

### ✅ Requirements

1. Filter: `order_status == 'Delivered'`
2. Filter: `payment_method` is `'Credit Card'` OR `'UPI'`
3. Filter: `amount_inr > 50000`
4. Select only the 6 specified columns
5. Sort by `amount_inr` descending, reset index

### 📊 Expected Output Format

| index | order_id | customer_id | product_id | order_date | amount_inr | payment_method |

| 0 | O... | C... | P... | 20xx-xx-xx | 899990.0 | Credit Card |
| ... | | | | | | |

### 🧠 Concepts Tested

```
Multi-condition Filtering  +  isin()  +  Column Selection  +  Sorting  +  Index Reset
```

````python
high_value_orders = orders[
    (orders['order_status'] == 'Delivered') & 
    (orders['payment_method'].isin(['Credit Card','UPI'])) &
    (orders['amount_inr'] > 50_000)
].sort_values(by='amount_inr',
              ascending=False,
              ignore_index=True)[['order_id','customer_id','product_id','order_date','amount_inr','payment_method']]

high_value_orders
````

<details>

- For checking membership in a list of values, `.isin([...])` is cleaner than chaining `|` conditions.
- Missing `amount_inr` values will **automatically be excluded** by the `> 50000` filter — think about why.
- The order of operations matters: filter all conditions first, then select columns, then sort.

### 🏢 Business Scenario

The HR Head wants a **department-level salary report** to understand how salaries are distributed across departments. The report should show, per department: the **average salary**, the **highest salary**, the **lowest salary**, and the **total headcount** — for **Active employees only**, sorted by average salary descending.

### 📝 Problem Statement

From the `employees` DataFrame:

1. Filter to **Active** employees only
2. Group by `dept_id`
3. Aggregate `salary_inr` to compute:
   - `avg_salary` — mean, rounded to 0 decimal places
   - `max_salary` — maximum
   - `min_salary` — minimum
   - `headcount` — count of employees (use `count` on salary; missing salaries excluded from count)
4. Sort by `avg_salary` descending
5. Reset index

### ✅ Requirements

- Use named aggregation (`.agg()` with keyword arguments)
- Missing `salary_inr` values should not be included in count
- Round `avg_salary` to 0 decimal places after aggregation

### 📊 Expected Output Format

| dept_id | avg_salary | max_salary | min_salary | headcount |
|---|---|---|---|---|
| D07 | 108000.0 | 114000.0 | ... | ... |
| D05 | ... | ... | ... | ... |
| ... | | | | |

### 🧠 Concepts Tested

```
Boolean Filtering  +  GroupBy  +  Named Aggregation  +  round()  +  sort_values()  +  reset_index()
```

````python
employees[employees['status'] == 'Active'] \
    .groupby('dept_id') \
    .agg(
        avg_salary=('salary_inr', 'mean'),
        highest_salary=('salary_inr', 'max'),
        lowest_salary=('salary_inr', 'min'),
        headcount=('emp_id', 'count')
    ) \
    .round({'avg_salary': 2}) \
    .sort_values(by='avg_salary', ascending=False)
````

<details>
<summary>💡 Hint — Click to expand (try on your own first!)</summary>

<br>

- Filter **before** grouping — it is both semantically correct and more efficient.
- Named aggregation syntax: `.agg(new_col_name=('column', 'function'))`
- `mean()` by default **skips NaN values** — so your average will be computed only over employees who have a salary recorded.
- `count()` also skips NaN — which is exactly what you want for headcount of employees with known salaries.
- Apply `round()` **after** the aggregation, not inside it.

</details>

### 🏢 Business Scenario

The HR team wants to identify **above-average earners within each department**. Rather than computing each department's average manually, they want a single Pandas operation that adds the department average as a new column, and then filters to employees who earn **more than their own department's average salary**.

### 📝 Problem Statement

From the `employees` DataFrame:

1. Consider only **Active** employees with a **non-null salary**
2. Add a new column `dept_avg_salary` containing the **mean salary of the employee's department** (using `groupby` + `transform`)
3. Filter to keep only employees whose `salary_inr` is **strictly greater than** their `dept_avg_salary`
4. Select: `emp_id`, `employee_name`, `dept_id`, `salary_inr`, `dept_avg_salary`
5. Round `dept_avg_salary` to 2 decimal places
6. Sort by `dept_id` ASC, then `salary_inr` DESC
7. Reset index

### ✅ Requirements

- Use `groupby().transform('mean')` — **not** `groupby().agg()`
- The `transform` should broadcast the department average back to every row
- No loops, no apply

### 📊 Expected Output Format

| index | emp_id | employee_name | dept_id | salary_inr | dept_avg_salary |
|---|---|---|---|---|---|
| 0 | E... | ... | D01 | 105000.0 | 89xxx.xx |
| 1 | E... | ... | D01 | 101000.0 | 89xxx.xx |
| ... | | | | | |

### 🧠 Concepts Tested

```
Boolean Filtering  +  GroupBy + Transform  +  Comparison Filtering  +  Column Selection  +  Sorting
```

````python
result = employees[(employees['status'] == 'Active') & (employees['salary_inr'].notna())].copy()

result['dept_avg_salary'] = result.groupby('dept_id')['salary_inr'].transform('mean').round(2)

result = result[
    result["salary_inr"] > result["dept_avg_salary"]
][["emp_id", "employee_name", "dept_id", "salary_inr", "dept_avg_salary"]].sort_values(by=['dept_id','salary_inr'],
                                                                                       ascending=[True,False],
                                                                                       ignore_index=True)

display(result.head())
````

<details>
<summary>💡 Hint — Click to expand (try on your own first!)</summary>

<br>

- `groupby().transform('mean')` returns a Series of the **same length** as the original DataFrame — each value is the mean of its group. This is what makes it different from `agg()`.
- Filter for Active + non-null salary **before** computing the transform — this ensures the department average is computed correctly.
- Assign the result of `transform` as a new column on the filtered DataFrame.
- Then use a simple boolean comparison `salary_inr > dept_avg_salary` to filter.
- Missing salaries may propagate into `transform` — filtering them out first avoids this.

</details>

### 🏢 Business Scenario

The Finance team needs a **year-wise order revenue summary** to understand business growth. They want to see total revenue from **Delivered** orders only, grouped by the year the order was placed — sorted chronologically. Only years with **more than 20** delivered orders should be included.

### 📝 Problem Statement

From the `orders` DataFrame:

1. Parse `order_date` as a datetime column
2. Extract the **year** from `order_date` into a new column `order_year`
3. Filter to orders where `order_status == 'Delivered'` **and** `amount_inr` is **not null**
4. Group by `order_year` and compute:
   - `total_orders` — count of orders
   - `total_revenue` — sum of `amount_inr`, rounded to 2 decimal places
   - `avg_order_value` — mean of `amount_inr`, rounded to 2 decimal places
5. Filter to include only years where `total_orders > 20`
6. Sort by `order_year` ascending
7. Reset index

### ✅ Requirements

- Use `pd.to_datetime()` for parsing
- Use `.dt.year` to extract the year
- Use named aggregation inside `.agg()`
- Filter the grouped result (not the raw data) for `total_orders > 20`

### 📊 Expected Output Format

| order_year | total_orders | total_revenue | avg_order_value |
|---|---|---|---|
| 2022 | ... | ... | ... |
| 2023 | ... | ... | ... |
| 2024 | ... | ... | ... |

### 🧠 Concepts Tested

```
Datetime Parsing  +  .dt accessor  +  Boolean Filtering  +  GroupBy  +  Named Aggregation  +  Post-Aggregation Filtering
```

````python
result = orders[orders["order_status"] == "Delivered"].copy()

result["order_year"] = result["order_date"].dt.year

result = (
    result.groupby("order_year")
    .agg(
        total_orders=("order_id", "count"),
        total_revenue=("amount_inr", "sum"),
        avg_order_value=("amount_inr", "mean")
    )
    .query("total_orders > 20")
    .sort_index()
    .reset_index()
    .round({"avg_order_value" : 2})
)

display(result)
````

<details>
<summary>💡 Hint — Click to expand (try on your own first!)</summary>

<br>

- Use `pd.to_datetime(df['col'])` to convert a string column to datetime. You can assign it back to the same column.
- After conversion, `.dt.year` extracts the integer year from each date.
- Filter for Delivered orders **and** non-null amounts before grouping — this keeps your revenue calculations clean.
- The filter `total_orders > 20` is applied **after** aggregation — on the resulting grouped DataFrame.
- Think about whether you need to sort before or after the final filter.

</details>

````python
# Step1: Copy Datasets

result = orders.copy()
prod = products.copy()
vendor = vendors.copy()

# Step2: Filter Delivered orders with non-null amount

result = result[
    (result['order_status'] == 'Delivered') &
    (result['amount_inr'].notna())
]

# Step3: Merge orders with products

result = pd.merge(
    result,
    prod,
    on='product_id',
    how='inner'
)

# Step4: Clean vendor names

vendor['vendor_name'] = vendor['vendor_name'].str.strip().str.title()

# Step5: Merge with vendors

result = pd.merge(
    result,
    vendor,
    on='vendor_id',
    how='inner'
)

# Step6: Compute vendor summary

result = result.groupby(
    ['vendor_id', 'vendor_name']
).agg(
    total_orders=('order_id', 'count'),
    total_revenue=('amount_inr', 'sum'),
    unique_products_sold=('product_id', 'nunique'),
    avg_order_value=('amount_inr', 'mean')
).round({
    'avg_order_value': 2
})

# Step7: Sort and reset index

result = result.sort_values(
    by='total_revenue',
    ascending=False
).reset_index()

display(result)
````

### 🏢 Business Scenario

The CEO wants a single **Top Employee Per Department** report based on a multi-factor performance score. The score combines three normalized metrics:
- **Salary score** (40%): Employee's salary as a percentile within their department
- **Attendance score** (40%): Employee's avg H1 2024 attendance rate normalized to 0–100
- **Transaction score** (20%): Employee's total Successful H1 2024 transaction amount, ranked as a percentile across all employees

Find the **single highest-scoring employee per department** and produce a leaderboard enriched with department names.

### 📝 Problem Statement

**Step 1 — Salary Percentile:**
1. `employees` — Active, deduplicated, non-null salary
2. `salary_pct` = `groupby("dept_id")["salary_inr"].rank(pct=True) * 100`

**Step 2 — Attendance Score:**
3. Filter `attendance` to H1 2024 (year=2024, month≤6); drop null `present_days`
4. Compute `attendance_rate` per row; group by `emp_id`, compute `avg_att_rate` = mean (rounded 2dp)
5. `attendance_score` = `avg_att_rate` (already 0–100 scale — no transformation needed)

**Step 3 — Transaction Percentile:**
6. Filter `transactions` to `txn_status=="Success"`, non-null `amount_inr`, H1 2024
7. Group by `emp_id`, compute `total_txn_amount` = sum of `amount_inr`
8. `txn_pct` = `total_txn_amount.rank(pct=True) * 100` (global percentile across all employees)

**Step 4 — Merge & Score:**
9. Start from Active employees (from Step 1); left join attendance summary; left join transaction summary
10. Fill missing `avg_att_rate` with 0; fill missing `total_txn_amount` with 0; fill missing `txn_pct` with 0
11. Compute `composite_score` = (`salary_pct` × 0.40) + (`attendance_score` × 0.40) + (`txn_pct` × 0.20), rounded to 2 dp

**Step 5 — Top Employee Per Dept:**
12. Within each `dept_id`, find the employee with the **highest `composite_score`** — use `groupby("dept_id")["composite_score"].transform("max")` and filter where `composite_score == dept_max`
13. Merge with `departments` (inner join) to get `department_name`
14. Select: `department_name`, `emp_id`, `employee_name`, `salary_inr`, `salary_pct`, `avg_att_rate`, `txn_pct`, `composite_score`
15. Round `salary_pct` and `txn_pct` to 2 dp
16. Sort by `composite_score` descending
17. Reset index

### ✅ Requirements

- Three independent pipelines, then left-joined onto employees
- `txn_pct` is computed on the aggregated transaction DataFrame, then brought in via merge
- `transform("max")` for top-per-group selection — not `nlargest()` with `groupby`
- If two employees tie for the top score in a dept, both should appear

## 💡 Pandas Interview Rules You Should Never Forget

```
1. Always filter FIRST — shrink data before merging or grouping
2. Clean strings BEFORE merging — dirty keys break joins silently
3. fillna() BEFORE groupby sum — NaN propagates and ruins totals
4. Sort BEFORE shift() or diff() — order matters for time-series ops
5. transform() keeps shape; agg() reduces shape — know which you need
6. rank(pct=True) gives percentiles; rank(method="dense") handles ties
7. unstack() reshapes from long to wide; stack() goes wide to long
8. pd.cut() = fixed bins; pd.qcut() = equal-frequency bins
9. isin() + ~ is your set-difference tool — never use Python set()
10. Build 3 clean pipelines and merge — never try to do everything in one query
```
