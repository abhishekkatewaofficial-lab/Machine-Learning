### 🍽️ Business Scenario

A restaurant manager wants a **day-wise tipping report for smoking customers only**. She suspects that smokers tip differently across the days of the week.

Work with the `tips` dataset. Consider **only rows where `smoker == "Yes"`** and remove any rows where `total_bill` is missing. Compute the tip percentage per transaction as `(tip / total_bill) × 100`. Then group by `day` and compute:
- The **mean tip percentage**, rounded to 2 decimal places → name this column `avg_tip_pct`
- The **count of transactions** → name this column `transaction_count`

Sort the result by `avg_tip_pct` descending so the highest-tipping day appears first. Reset the index before returning the final DataFrame.

### 📝 Problem Statement

1. Filter rows where `smoker == "Yes"`
2. Drop rows where `total_bill` is null
3. Compute `tip_pct` = (`tip` / `total_bill`) × 100
4. Group by `day`, compute:
   - `avg_tip_pct` — mean of `tip_pct`, rounded to 2 dp
   - `transaction_count` — count of rows
5. Sort by `avg_tip_pct` descending
6. Reset index

### 📊 Expected Output Format

| day | avg_tip_pct | transaction_count |
|---|---|---|
| Fri | 17.xx | ... |
| Sun | 15.xx | ... |
| ... | | |

### 🧠 Concepts Tested

```
Boolean Filter  +  dropna  +  Computed Column  +  GroupBy  +  Named Aggregation  +  round()  +  sort_values  +  reset_index
```

````python
# Step1: Copy Dataset

result = tips.copy()

# Step2: Keep only smokers

result = result[result['smoker'] == 'Yes']

# Step3: Remove rows with missing total_bill

result = result[result['total_bill'].notna()]

# Step4: Compute tip percentage

result['tip_pct'] = (
    result['tip'] / result['total_bill'] * 100
)

# Step5: Day-wise summary

result = result.groupby('day',observed=True).agg(
    avg_tip_pct=('tip_pct', 'mean'),
    transaction_count=('day', 'count')
).round({
    'avg_tip_pct': 2
})

# Step6: Sort and reset index

result = result.sort_values(
    by='avg_tip_pct',
    ascending=False
).reset_index()

display(result)
````

````python
fig,ax = plt.subplots(
    nrows=1,
    ncols=2
)

sns.barplot(
    data=result,
    x='day',
    y='avg_tip_pct',
    palette='deep',
    hue='day',
    legend=False,
    ax=ax[0]
)

ax[0].set_xlabel('Day')
ax[0].set_ylabel('Average Tip Percent')
ax[0].set_label('Day vs avg tip')

sns.barplot(
    data=result,
    x='day',
    y='transaction_count',
    palette='deep',
    hue='day',
    legend=False,
    ax=ax[1]
)


plt.tight_layout()
plt.show()
````

<details>
<summary>💡 Hint — Click to expand (try on your own first!)</summary>

<br>

- `df[df["smoker"] == "Yes"]` — boolean filter. Assign result before proceeding.
- `dropna(subset=["total_bill"])` — drops only rows where `total_bill` is null.
- Compute `tip_pct` as a new column **before** `groupby`.
- Named aggregation syntax: `groupby("day").agg(avg_tip_pct=("tip_pct","mean"), transaction_count=("tip_pct","count"))`
- Round after agg: `result["avg_tip_pct"] = result["avg_tip_pct"].round(2)`

</details>

### 🚢 Business Scenario

A maritime historian wants to study **how much surviving passengers paid** across each passenger class on the Titanic.

Work with the `titanic` dataset. Consider **only passengers who survived** (`survived == 1`) and remove any rows where `fare` is missing. Group the remaining data by `pclass` and compute three metrics:
- The **average fare** per class, rounded to 2 decimal places → name this `avg_fare`
- The **total fare collected** per class, rounded to 2 decimal places → name this `total_fare`
- The **count of surviving passengers** per class → name this `passenger_count`

Sort the result by `pclass` ascending (1st class first) and reset the index.

### 📝 Problem Statement

1. Filter rows where `survived == 1`
2. Drop rows where `fare` is null
3. Group by `pclass`, compute:
   - `avg_fare` — mean of `fare`, rounded to 2 dp
   - `total_fare` — sum of `fare`, rounded to 2 dp
   - `passenger_count` — count of rows
4. Sort by `pclass` ascending
5. Reset index

### 📊 Expected Output Format

| pclass | avg_fare | total_fare | passenger_count |
|---|---|---|---|
| 1 | ... | ... | ... |
| 2 | ... | ... | ... |
| 3 | ... | ... | ... |

### 🧠 Concepts Tested

```
Boolean Filter  +  dropna  +  GroupBy  +  Named Aggregation (mean + sum + count)  +  round() ×2  +  sort_values  +  reset_index
```

````python
# Step1: Copy dataset

result = titanic.copy()

# Step2: Keep only surviving passengers

result = result[result['survived'] == 1]

# Step3: Remove rows with missing fare

result = result[result['fare'].notna()]

# Step4: Passenger class summary

result = result.groupby('pclass', observed=True).agg(
    avg_fare=('fare', 'mean'),
    total_fare=('fare', 'sum'),
    passenger_count=('pclass', 'count')
).round({
    'avg_fare': 2,
    'total_fare': 2
})

# Step5: Sort and reset index

result = result.sort_values(
    by='pclass',
    ascending=True
).reset_index()

display(result)
````

<details>
<summary>💡 Hint — Click to expand (try on your own first!)</summary>

<br>

- `titanic[titanic["survived"] == 1]` — filter surviving passengers first.
- `dropna(subset=["fare"])` — removes only rows with null fare.
- Named agg: `agg(avg_fare=("fare","mean"), total_fare=("fare","sum"), passenger_count=("fare","count"))`
- Apply `.round(2)` on `avg_fare` and `total_fare` columns after aggregation.
- `sort_values("pclass")` sorts ascending by default — no need to pass `ascending=True`.

</details>

### 🍽️ Business Scenario

The restaurant's operations team wants to compare **per-person billing and tipping** between Lunch and Dinner service, but only for tables with 2 or more guests (solo diners are excluded from this report).

Work with the `tips` dataset. First, rename three columns to make the report cleaner: `total_bill` → `bill_amount`, `size` → `party_size`, `tip` → `tip_amount`. Then keep only rows where `party_size >= 2` and remove rows where `bill_amount` is missing.

Compute two new columns:
- `bill_per_person` = `bill_amount` / `party_size`, rounded to 2 decimal places
- `tip_per_person` = `tip_amount` / `party_size`, rounded to 2 decimal places

Group by `time` (Lunch / Dinner) and compute:
- Mean `bill_per_person` → `avg_bill_per_person` (rounded to 2 dp)
- Mean `tip_per_person` → `avg_tip_per_person` (rounded to 2 dp)
- Count of tables → `total_tables`

Sort by `avg_bill_per_person` descending and reset the index.

### 📝 Problem Statement

1. Rename: `total_bill` → `bill_amount`, `size` → `party_size`, `tip` → `tip_amount`
2. Filter rows where `party_size >= 2`
3. Drop rows where `bill_amount` is null
4. Compute `bill_per_person` = `bill_amount` / `party_size`, rounded to 2 dp
5. Compute `tip_per_person` = `tip_amount` / `party_size`, rounded to 2 dp
6. Group by `time`, compute:
   - `avg_bill_per_person` — mean of `bill_per_person`, rounded to 2 dp
   - `avg_tip_per_person` — mean of `tip_per_person`, rounded to 2 dp
   - `total_tables` — count of rows
7. Sort by `avg_bill_per_person` descending
8. Reset index

### 📊 Expected Output Format

| time | avg_bill_per_person | avg_tip_per_person | total_tables |
|---|---|---|---|
| Dinner | ... | ... | ... |
| Lunch | ... | ... | ... |

### 🧠 Concepts Tested

```
rename()  +  Boolean Filter  +  dropna  +  Computed Column ×2  +  GroupBy  +  Named Aggregation  +  round() ×2  +  sort_values  +  reset_index
```

````python
# Step1: Copy dataset

tip = tips.copy()

# Step2: Keep only tables with 2 or more guests
# Remove rows where total_bill is missing

tip = tip[
    (tip['size'] >= 2) &
    (tip['total_bill'].notna())
]

# Step3: Compute per-person bill and tip

tip['bill_per_person'] = tip['total_bill'] / tip['size']
tip['tip_per_person'] = tip['tip'] / tip['size']

# Step4: Lunch vs Dinner summary

tip = tip.groupby('time', observed=True).agg(
    avg_bill_per_person=('bill_per_person', 'mean'),
    avg_tip_per_person=('tip_per_person', 'mean'),
    total_tables=('time', 'count')
).round({
    'avg_bill_per_person': 2,
    'avg_tip_per_person': 2
}).reset_index()

display(tip)
````

<summary>💡 Hint — Click to expand (try on your own first!)</summary>

- Rename **first**, then filter using the new column name `party_size`.
- Compute both `bill_per_person` and `tip_per_person` as new columns **before** `groupby`.
- Named agg: `agg(avg_bill_per_person=("bill_per_person","mean"), ...)`
- Apply `.round(2)` on both result columns after aggregation.

### 🚢 Business Scenario

A port authority analyst needs to understand **how many passengers boarded from each embarkation port, broken down by passenger class**.

Work with the `titanic` dataset. Some rows have missing embarkation port values — **fill them with `"Unknown"`** (do not drop these rows; they must still appear in the report). Then remove any rows where `fare` is missing. Group the data by `embarked` and `pclass` and count the number of passengers in each group — name this count column `passenger_count`. Sort the result by `passenger_count` descending and reset the index.

### 📝 Problem Statement

1. Fill null values in `embarked` with `"Unknown"`
2. Drop rows where `fare` is null
3. Group by `embarked` and `pclass`, count passengers → name this column `passenger_count`
4. Sort by `passenger_count` descending
5. Reset index

### 📊 Expected Output Format

| embarked | pclass | passenger_count |
|---|---|---|
| S | 3 | ... |
| S | 1 | ... |
| ... | | |

### 🧠 Concepts Tested

```
fillna (column-specific)  +  dropna (subset)  +  GroupBy (2-key)  +  size / named count  +  sort_values  +  reset_index
```

````python
# Step1: Copy dataset

titan = titanic.copy()

# Step2: Fill missing embarkation ports

titan['embarked'] = titan['embarked'].fillna('Unknown')

# Step3: Remove rows with missing fare

titan = titan[titan['fare'].notna()]

# Step4: Passenger count by embarkation port and passenger class

titan = titan.groupby(
    ['embarked', 'pclass'],
    observed=True
).agg(
    passenger_count=('pclass', 'count')
).sort_values(
    by='passenger_count',
    ascending=False
).reset_index()



display(titan)
````

<details>
<summary>💡 Hint — Click to expand (try on your own first!)</summary>

<br>

- `df["embarked"].fillna("Unknown")` — fill only the `embarked` column. Assign back or use `inplace=True`.
- Fill nulls **before** dropping — otherwise the fill is wasted.
- Two approaches to count per group:
  - `.groupby(["embarked","pclass"]).size().reset_index(name="passenger_count")`
  - `.groupby(["embarked","pclass"]).agg(passenger_count=("fare","count")).reset_index()`
- Both give the same result here since fare nulls were already dropped.

</details>

````python
sns.barplot(
    data=titan,
    x='embarked',
    y='passenger_count',
    hue='pclass',
    legend=True
)

plt.title('Passenger Count by Embarkation Port and Passenger Class')
plt.xlabel('Embarkation Port')
plt.ylabel('Passenger Count')
plt.legend(title = 'Passenger Class')
plt.show()
````

### 🍽️ Business Scenario

The restaurant manager is reviewing **Dinner service performance by day of the week**. She wants the average tip percentage and total revenue per day — but only for days that had **at least 10 dinner transactions** (days with fewer records are too small a sample to draw conclusions from).

Work with the `tips` dataset. Filter rows where `time == "Dinner"` and remove any rows where `total_bill` is missing. Compute the tip percentage per transaction as `(tip / total_bill) × 100`. Group by `day` and compute:
- Mean tip percentage, rounded to 2 dp → `avg_tip_pct`
- Sum of `total_bill`, rounded to 2 dp → `total_revenue`
- Count of transactions → `transaction_count`

Keep **only days where `transaction_count >= 10`**. Sort the remaining rows by `avg_tip_pct` descending and reset the index.

### 📝 Problem Statement

1. Filter rows where `time == "Dinner"`
2. Drop rows where `total_bill` is null
3. Compute `tip_pct` = (`tip` / `total_bill`) × 100
4. Group by `day`, compute:
   - `avg_tip_pct` — mean of `tip_pct`, rounded to 2 dp
   - `total_revenue` — sum of `total_bill`, rounded to 2 dp
   - `transaction_count` — count of rows
5. Keep only rows where `transaction_count >= 10`
6. Sort by `avg_tip_pct` descending
7. Reset index

### 📊 Expected Output Format

| day | avg_tip_pct | total_revenue | transaction_count |
|---|---|---|---|
| Sun | 16.xx | ... | ... |
| Sat | 15.xx | ... | ... |
| ... | | | |

### 🧠 Concepts Tested

```
Boolean Filter  +  dropna  +  Computed Column  +  GroupBy  +  Named Aggregation (mean+sum+count)  +  Post-Aggregation Filter  +  sort_values  +  reset_index
```

````python
# Step1: Copy dataset

tip = tips.copy()

# Step2: Keep only Dinner transactions
# Remove rows with missing total_bill

tip = tip[
    (tip['time'] == 'Dinner') &
    (tip['total_bill'].notna())
]

# Step3: Compute tip percentage

tip['tip_pct'] = (
    tip['tip'] / tip['total_bill']
) * 100

# Step4: Summary by day

tip = tip.groupby(
    'day',
    observed=True
).agg(
    avg_tip_pct=('tip_pct', 'mean'),
    total_revenue=('total_bill', 'sum'),
    transaction_count=('day', 'count')
).query(
    "transaction_count >= 10"
).round({
    'avg_tip_pct': 2,
    'total_revenue': 2
}).sort_values(
    by='avg_tip_pct',
    ascending=False
).reset_index()

display(tip)
````

````python
plt.style.use('default')
fig, ax = plt.subplots(
    nrows=1,
    ncols=2,
    figsize=(12, 5)
)

sns.barplot(
    data=tip,
    x='day',
    y='avg_tip_pct',
    hue='day',
    dodge=False,
    ax=ax[0]
)

ax[0].set_title('Average Tip Percentage')
ax[0].set_xlabel('Day')
ax[0].set_ylabel('Tip Percentage (%)')


sns.barplot(
    data=tip,
    x='day',
    y='total_revenue',
    hue='day',
    dodge=False,
    ax=ax[1]
)

ax[1].set_title('Total Dinner Revenue')
ax[1].set_xlabel('Day')
ax[1].set_ylabel('Revenue ($)')

plt.tight_layout()
plt.show()
````

<details>
<summary>💡 Hint — Click to expand (try on your own first!)</summary>

<br>

- Filter `time == "Dinner"` first, then `dropna` — avoids nulls skewing the tip percentage.
- Compute `tip_pct` as a new column before `groupby` so you can aggregate on it.
- After `groupby().agg(...)`, the result is a regular DataFrame — you can filter it like any other:
  `result = result[result["transaction_count"] >= 10]`
- This is the **post-aggregation filter** pattern — filtering on aggregated values, not individual rows.
- Sort and reset index **after** the post-agg filter.

</details>

result = df.groupby(["col1","col2"]).agg(

result = df.groupby("col").agg({

## 8. 🔗 pd.merge — Joining DataFrames

```python
# Basic left merge (most common):
pd.merge(df_left, df_right, on="key_col", how="left")

# Multi-key merge (self-join on 2 keys):
pd.merge(df, ref, on=["col1","col2"], how="left")

# Two-summary merge (two groupby outputs combined):
summary_a = df.groupby(["day","smoker"]).agg(...).reset_index()
summary_b = df.groupby("day").agg(...).reset_index()
merged    = pd.merge(summary_a, summary_b, on="day", how="left")

# Three-table merge (step by step):
step1 = pd.merge(a, b, on="key", how="left")
final = pd.merge(step1, c, on="key", how="left")

# how options:
# "left"  → keep all rows from left, match from right (NaN if no match)
# "inner" → keep only rows with matches in BOTH
# "outer" → keep all rows from both (NaN where no match)
```

## 9. 🔀 groupby().transform() — Broadcast Group Aggregate

```python
# transform() returns a Series with the SAME shape as df
# — each row gets its group's aggregate (no merge needed)

# 1-key transform:
df["day_avg"] = df.groupby("day")["tip_pct"].transform("mean").round(2)

# 2-key transform:
df["cls_sex_avg"] = df.groupby(["pclass","sex"])["fare"].transform("mean").round(2)
df["cls_sex_std"] = df.groupby(["pclass","sex"])["fare"].transform("std").round(2)

# Available aggfuncs: "mean", "sum", "std", "min", "max", "count", "median"

# After transform, use the broadcast value for row-level comparison:
df["is_premium"]  = df["fare"]    > df["cls_sex_avg"]
df["abs_dev"]     = (df["fare"]   - df["cls_sex_avg"]).abs().round(2)

# transform() vs agg():
# agg()       → reduces: 244 rows → 4 rows (one per group)
# transform() → broadcasts: 244 rows → still 244 rows
```

## 10. 🏅 groupby().rank() — Within-Group Ranking

```python
# Within-group ranking — returns same shape as df (like transform)

# 1-key rank:
df["fare_rank"] = df.groupby("pclass")["fare"].rank(
    method="dense",    # no gaps between ranks for ties
    ascending=False    # highest fare = rank 1
)

# 2-key rank:
df["within_rank"] = df.groupby(["pclass","sex"])["fare"].rank(
    method="dense", ascending=False
)

# method options:
# "dense"   → ties share same rank, no gap: [1,1,2]
# "min"     → ties get lowest rank, gap after: [1,1,3]
# "average" → ties get average rank (default): [1.5,1.5,3]

# Flag top-N within group:
df["is_top3"]  = df["fare_rank"] <= 3
df["is_elite"] = df["within_rank"] <= 5
```

## 11. ⏩ shift() and cumsum() — Lag & Running Total

### shift() — Lag/Lead
```python
# MUST sort first before using shift on time-ordered data!
result = result.sort_values("day_order")

result["prev_rev"] = result["daily_revenue"].shift(1)   # lag (previous row)
result["next_rev"] = result["daily_revenue"].shift(-1)  # lead (next row)

# shift(1): Row 0 → NaN (no prior row), Row 1 → Row 0's value
# shift(-1): Last row → NaN (no next row), others shift up

result["revenue_change"] = (result["daily_revenue"] - result["prev_rev"]).round(2)
```

### cumsum() — Running Total
```python
# MUST sort first before cumsum!
result = result.sort_values("day_order")

result["cum_rev"]  = result["daily_revenue"].cumsum().round(2)
result["cum_tips"] = result["daily_tips"].cumsum().round(2)

# cumsum: [100, 200, 300, 400] → [100, 300, 600, 1000]
# Each row = sum of all rows up to and including itself
```

## 12. 🔁 groupby().unstack() — Wide Matrix

```python
# Reshape a 2-level groupby into a wide-format DataFrame
# rows = first key, columns = second key (unstacked)

matrix = (
    df.groupby(["day","smoker"])["tip_pct"]
      .mean()
      .round(2)
      .unstack(level="smoker")  # smoker values → columns
)
# → DataFrame: rows=day, columns=[No, Yes]

matrix.columns.name = None   # remove "smoker" axis label
matrix = matrix.fillna(0.0)  # no NaN cells
matrix = matrix.reset_index()

# unstack() vs pivot_table():
# Both produce the same wide result.
# unstack() chains from any groupby; pivot_table() is a standalone call.

# Row-wise operations on the wide matrix:
time_cols = ["Dinner","Lunch"]
matrix["best_time"]    = matrix[time_cols].idxmax(axis=1)  # col name of max
matrix["avg_tip_pct"]  = matrix[time_cols].mean(axis=1).round(2)  # row mean
```

## 14. ✅ groupby().agg with "any" and "all"

```python
# "any" → True if AT LEAST ONE row in the group is True (OR logic)
# "all" → True if EVERY row in the group is True (AND logic)

df["is_survivor"] = df["survived"] == 1
df["has_deck"]    = df["deck"].notna()
df["is_child"]    = df["age"] < 18

result = df.groupby(["pclass","sex"]).agg(
    any_survivor  = ("is_survivor", "any"),  # ≥1 survivor?
    all_have_deck = ("has_deck",    "all"),  # all have cabin info?
    any_child     = ("is_child",    "any"),  # ≥1 child?
).reset_index()

# Intuition:
# Group = [False, False, True, False]  → any() = True, all() = False
# Group = [True,  True,  True, True]   → any() = True, all() = True
# Group = [False, False, False]        → any() = False, all() = False
```

## 15. ⚡ Multi-Pipeline Pattern

```python
# Two completely independent analysis chains — each starts from raw data
# Combined at the end via pd.merge on a shared key

# Pipeline A — own working copy
df_a = tips.dropna(subset=["total_bill","tip"]).copy()
df_a["tip_pct"] = df_a["tip"] / df_a["total_bill"] * 100
df_a = df_a[df_a["day"].isin(["Sat","Sun"])]
pipeline_a = (
    df_a.groupby(["day","smoker"])
        .agg(avg_tip_pct=("tip_pct","mean"), visit_count=("tip_pct","count"))
        .round(2).reset_index()
)

# Pipeline B — completely independent fresh copy
df_b = tips.dropna(subset=["total_bill"]).copy()
pipeline_b = (
    df_b.groupby("day")
        .agg(day_revenue=("total_bill","sum"), txn_count=("total_bill","count"))
        .round(2).reset_index()
)

# Combine
final = pd.merge(pipeline_a, pipeline_b, on="day", how="left")

# Key rule: Do NOT share intermediate variables between pipelines.
# Each pipeline is independent from start to finish.
```

## 16. 🔄 Self-Join Pattern

```python
# Self-join: merge df with an aggregated version of itself
# Used to attach group-level context to individual rows

# 1-key self-join (single key):
ref = (
    df.groupby("pclass")
      .agg(class_avg=("fare","mean"), class_size=("fare","count"))
      .round(2).reset_index()
)
merged = pd.merge(df, ref, on="pclass", how="left")

# 2-key self-join (two-key — more specific):
ref = (
    df.groupby(["pclass","embarked"])
      .agg(group_avg=("fare","mean"), group_size=("fare","count"))
      .round(2).reset_index()
)
ref    = ref[ref["group_size"] >= 5]          # filter small groups
merged = pd.merge(df, ref, on=["pclass","embarked"], how="left")

# After self-join:
merged["is_premium"] = merged["fare"] > merged["group_avg"]
merged = merged.dropna(subset=["group_avg"])  # drop unmatched groups

# Equivalent to transform() but more explicit:
# transform() is cleaner for simple cases;
# self-join is better when you need group_size filtering before merging.
```
