### 💎 Business Scenario

A gemstone retailer wants to understand the **premium end of the market**. They're only interested in diamonds with `price` above $5 000 and `carat` of at least 1.0. For this segment, they want to know the average price, highest carat, and count by `cut` quality.

Work with the `diamonds` dataset. Filter: `price > 5000` and `carat >= 1.0`. Drop rows where `price` or `carat` is null (confirm there are none, but include the step). Compute `price_per_carat` = `price / carat`, rounded to 2 dp. Group by `cut` and compute: count → `n`, mean `price` → `avg_price` (2dp), max `carat` → `max_carat`, mean `price_per_carat` → `avg_ppc` (2dp). Sort by `avg_price` descending. Reset the index.

### 📝 Problem Statement

1. Filter: `price > 5000` and `carat >= 1.0`
2. Drop rows where `price` or `carat` is null
3. Compute `price_per_carat` = `price / carat`, round 2dp
4. Group by `cut`, compute: `n` (count), `avg_price` (mean price, 2dp), `max_carat` (max), `avg_ppc` (mean price_per_carat, 2dp)
5. Sort by `avg_price` descending; reset index

### 📊 Expected Output Format

| cut | n | avg_price | max_carat | avg_ppc |
|---|---|---|---|---|
| Ideal | ... | ... | ... | ... |
| Premium | ... | ... | ... | ... |
| ... | | | | |

### 🧠 Concepts Tested

```
Boolean filter (2 conditions with &)  +  dropna (2-col)
Computed column: price / carat  +  round(2)
GroupBy (1-key: cut)  +  Named Agg (4 funcs)
sort_values (desc)  +  reset_index
```

<details>
<summary>💡 Hint — Click to expand</summary>

<br>

```python
df = diamonds.copy()
df = df[(df["price"] > 5000) & (df["carat"] >= 1.0)]
df = df.dropna(subset=["price","carat"])
df["price_per_carat"] = (df["price"] / df["carat"]).round(2)
result = df.groupby("cut").agg(
    n        = ("price",         "count"),
    avg_price= ("price",         "mean"),
    max_carat= ("carat",         "max"),
    avg_ppc  = ("price_per_carat","mean")
).reset_index()
result["avg_price"] = result["avg_price"].round(2)
result["avg_ppc"]   = result["avg_ppc"].round(2)
result = result.sort_values("avg_price", ascending=False).reset_index(drop=True)
```

```python
).round(2)
cnt_wide = pd.pivot_table(
    df, index="cut", columns="clarity_tier",
    values="price_per_carat", aggfunc="count"
)

# Combine with MultiIndex columns (metric × clarity_tier):
combined = pd.concat([ppc_wide, cnt_wide], axis=1, keys=["avg_ppc","count"])
print("Combined MultiIndex column shape:", combined.shape)
print("Columns:", list(combined.columns[:4]), "...")

# stack(level=1): fold the clarity_tier level (level=1) into row index:
# Before: rows=cut, cols=(metric, clarity_tier)
# After:  rows=(cut, clarity_tier), cols=metric
long = combined.stack(level=1)
long.columns.name = None
long = long.reset_index()
long.columns = ["cut","clarity_tier","avg_ppc","count"]

long = long.dropna(subset=["avg_ppc","count"])
long["count"]            = long["count"].astype(int)
long["ppc_rank_in_cut"]  = (
    long.groupby("cut")["avg_ppc"]
    .rank(method="dense", ascending=False)
    .astype(int)
)
long = long.sort_values(["cut","ppc_rank_in_cut"]).reset_index(drop=True)
display(long)
```

```python
import numpy as np
df = mpg.copy()
df["horsepower"] = df["horsepower"].fillna(df["horsepower"].median())
df = df.dropna(subset=["mpg","origin","weight"])
df["decade"]       = (df["model_year"]//10)*10
df["origin_upper"] = df["origin"].str.upper()

# Step 1 — apply() returning DataFrame (per-origin decade profile):
def origin_decade_stats(g):
    summary = g.groupby("decade").agg(
        n          = ("mpg","count"),
        avg_mpg    = ("mpg","mean"),
        avg_hp     = ("horsepower","mean"),
        avg_weight = ("weight","mean")
    ).reset_index()
    summary["avg_mpg"]    = summary["avg_mpg"].round(2)
    summary["avg_hp"]     = summary["avg_hp"].round(1)
    summary["avg_weight"] = summary["avg_weight"].round(0)
    summary["origin_upper"] = g["origin_upper"].iloc[0]
    return summary.sort_values("decade")

profiles = (
    df.groupby("origin_upper", group_keys=False)
    .apply(origin_decade_stats)
    .reset_index(drop=True)
)

# Step 2 — shift() for decade-over-decade change:
profiles = profiles.sort_values(["origin_upper","decade"]).reset_index(drop=True)
profiles["prev_avg_mpg"]  = profiles.groupby("origin_upper")["avg_mpg"].shift(1)
profiles["mpg_improvement"] = (profiles["avg_mpg"]-profiles["prev_avg_mpg"]).round(2)
profiles = profiles.dropna(subset=["prev_avg_mpg"])

# Step 3 — where() to cap extreme values:
profiles["mpg_improvement_capped"] = profiles["mpg_improvement"].where(
    profiles["mpg_improvement"].between(-5, 10), other=np.nan
)

# Step 4 — rank() improvement within each origin:
profiles["improvement_rank"] = (
    profiles.groupby("origin_upper")["mpg_improvement"]
    .rank(method="dense", ascending=False)
    .astype(int)
)

# Step 5 — idxmax() best decade per origin via apply:
best_decade = (
    profiles.groupby("origin_upper")
    .apply(lambda g: g.loc[g["avg_mpg"].idxmax(), "decade"])
    .reset_index()
)
best_decade.columns = ["origin_upper","best_decade"]
profiles = pd.merge(profiles, best_decade, on="origin_upper", how="left")

result = profiles.sort_values(
    ["origin_upper","improvement_rank"]).reset_index(drop=True)
display(result)
```
