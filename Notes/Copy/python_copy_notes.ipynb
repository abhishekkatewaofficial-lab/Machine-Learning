## Section 1: The Assignment Problem

Variables assigned to each other point to the SAME object.

```
a = [1,2,3]
b = a
print('a:', a, 'b:', b)
```

Modifying b modifies a.

```
b[0] = 99
print('a is also changed:', a)
```

Prove they're same object using id().

```
print('Same object?', id(a) == id(b))
```

Same trap with nested lists.

```
nested = [[1, 2], [3, 4]]
other = nested
other[0][0] = 99
print(nested)
```

## Section 2: Shallow Copy — 4 ways

Method 1: list.copy()

```
a = [1, 2, 3]
b = a.copy()
```

Method 2: slicing [:]

```
c = a[:]
```

Method 3: list() constructor

```
d = list(a)
```

Method 4: copy.copy()

```
import copy
e = copy.copy(a)
```

Verify all 4 give shallow copies (new top-level objects).

```
print(id(a) != id(b) and id(a) != id(c) and id(a) != id(d) and id(a) != id(e))
```

Shallow copy trap: nested lists are still shared!

```
a = [[1,2],[3,4]]
b = a.copy()
b[0][0] = 999
print('a is also changed!', a)
```

Visualize shallow vs deep copy with diagram.

```
print('Shallow Copy:\n a -> [ ref1, ref2 ] \n       \\   / \n b -> [ ref1, ref2 ]\n')
print('Deep Copy:\n a -> [ ref1, ref2 ] \n b -> [ ref3, ref4 ]')
```

## Section 3: Deep Copy

Use deepcopy for nested structures.

```
import copy
a = [[1,2],[3,4]]
b = copy.deepcopy(a)
print('Original:', a)
```

Deep copy on nested list — nested lists are NOT shared.

```
b[0][0] = 999
print('a remains:', a, 'b changes:', b)
```

Deep copy on dict with nested values.

```
d1 = {'k': [1,2]}
d2 = copy.deepcopy(d1)
d2['k'][0] = 99
print('d1:', d1, 'd2:', d2)
```

Deep copy on complex object (object with list attributes).

```
class Node:
    def __init__(self, val):
        self.val = val
        self.children = []

n1 = Node(1)
n1.children.append(Node(2))
n2 = copy.deepcopy(n1)
n2.children[0].val = 99
print(n1.children[0].val)
```

Performance: deepcopy is slower.

```
import timeit
setup = 'import copy; a = [[1,2]*100]*100'
print('Shallow:', timeit.timeit('a.copy()', setup=setup, number=1000))
print('Deep:', timeit.timeit('copy.deepcopy(a)', setup=setup, number=1000))
```

## Section 4: copy vs deepcopy — When to Use What

Decision table:
| Situation | Use |
| 1D list/dict | .copy() or [:] |
| Nested list/dict | deepcopy() |
| List of immutables | any copy method |
| List of objects | deepcopy() |

Immutable objects (int, str, tuple) — copying is pointless.

```
import copy
a = (1, 2, [3,4])
b = copy.deepcopy(a)
print(id(a) == id(b)) # False because of the mutable list inside!
```

When shallow copy is ENOUGH — list of numbers.

```
nums = [1, 2, 3]
nums_copy = nums[:]
nums_copy[0] = 9
print(nums)
```

When shallow copy FAILS — list of lists.

```
matrix = [[1,2], [3,4]]
matrix_copy = matrix[:]
matrix_copy[0][0] = 9
print(matrix)
```

## Section 6: DSA — Backtracking with Deep Copy

Backtracking = explore all paths, undo choices. Deep copy saves current state.

Generate all subsets using backtracking + deepcopy (naive way).

```
import copy
def subsets(nums):
    res = []
    def dfs(i, path):
        if i == len(nums):
            res.append(copy.deepcopy(path))
            return
        path.append(nums[i])
        dfs(i+1, path)
        path.pop()
        dfs(i+1, path)
    dfs(0, [])
    return res
print(subsets([1,2]))
```

N-Queens problem — save board state with deepcopy.

```
import copy
def solveNQueens(n):
    res = []
    board = [['.']*n for _ in range(n)]
    def backtrack(r):
        if r == n:
            res.append(copy.deepcopy(board))
            return
        # simplified backtracking for brevity
        board[r][0] = 'Q'
        backtrack(r+1)
        board[r][0] = '.'
    backtrack(0)
    return len(res)
print(solveNQueens(4))
```

All permutations using backtracking + deepcopy.

```
def permute(nums):
    res = []
    def dfs(path):
        if len(path) == len(nums):
            res.append(path[:]) # slice is faster shallow copy
            return
        for n in nums:
            if n not in path:
                path.append(n)
                dfs(path)
                path.pop()
    dfs([])
    return res
print(permute([1,2,3]))
```

Generate all valid parentheses combinations.

```
def generateParenthesis(n):
    res = []
    def dfs(op, cl, path):
        if len(path) == 2*n:
            res.append(''.join(path))
            return
        if op < n:
            path.append('(')
            dfs(op+1, cl, path)
            path.pop()
        if cl < op:
            path.append(')')
            dfs(op, cl+1, path)
            path.pop()
    dfs(0, 0, [])
    return res
print(generateParenthesis(2))
```

## Section 7: copy.__copy__ and copy.__deepcopy__ (Advanced)

Implement __copy__ in a custom class.

```
import copy
class MyObj:
    def __init__(self, val):
        self.val = val
    def __copy__(self):
        print('Custom shallow copy')
        return type(self)(self.val)

obj = MyObj(1)
obj2 = copy.copy(obj)
```

Implement __deepcopy__ in a custom class.

```
import copy
class MyObj2:
    def __init__(self, val):
        self.val = val
    def __deepcopy__(self, memo):
        print('Custom deep copy')
        return type(self)(copy.deepcopy(self.val, memo))

obj = MyObj2([1,2])
obj2 = copy.deepcopy(obj)
```

## Section 8: Common Mistakes and Traps

Trap 1 — using = instead of .copy() in a loop.

```
res = []
curr = []
for i in range(3):
    curr.append(i)
    res.append(curr) # WRONG: appends reference
print(res)
```

Trap 2 — shallow copy of 2D matrix in graph problems.

```
visited = [[False]*2]*2
visited[0][0] = True
print(visited) # Changes both rows!
```

Trap 3 — function default argument mutation.

```
def add_item(item, lst=[]):
    lst.append(item)
    return lst
print(add_item(1))
print(add_item(2)) # Returns [1, 2]
```

## Section 9: Cheat Sheet

- **Assignment (`=`)**: Same object.
- **Shallow Copy (`copy()`, `[:]`)**: New top-level object, inner objects shared.
- **Deep Copy (`copy.deepcopy()`)**: Completely independent clone.
- **Backtracking**: Use `[:]` to save 1D paths, `deepcopy` for grids. Avoid deepcopy if you can backtrack (undo).

```
# Solution
def ex4_permute(s):
    res = []
    def dfs(path, used):
        if len(path) == len(s):
            res.append(''.join(path))
            return
        for i, c in enumerate(s):
            if not used[i]:
                used[i] = True
                path.append(c)
                dfs(path, used)
                path.pop()
                used[i] = False
    dfs([], [False]*len(s))
    return res
```
