DSA: Caesar cipher by shifting each character by k positions.

```python
def caesar_cipher(text, k):
    res = []
    for c in text:
        if c.islower():
            res.append(chr((ord(c) - ord('a') + k) % 26 + ord('a')))
        else:
            res.append(c)
    return ''.join(res)
print(caesar_cipher('abc', 2))
```

DSA: Check if two strings are anagrams using frequency array with ord().

```python
def is_anagram(s, t):
    if len(s) != len(t): return False
    freq = [0]*26
    for c in s:
        freq[ord(c)-ord('a')] += 1
    for c in t:
        freq[ord(c)-ord('a')] -= 1
    return all(x == 0 for x in freq)
print(is_anagram('listen', 'silent'))
```

DSA: Find the missing letter in an alphabet sequence using ord().

```python
def find_missing_letter(chars):
    for i in range(len(chars) - 1):
        if ord(chars[i+1]) - ord(chars[i]) > 1:
            return chr(ord(chars[i]) + 1)
print(find_missing_letter(['a','b','d']))
```

DSA: Find the character that appears an odd number of times using XOR with ord().

```python
def find_odd_char(s):
    res = 0
    for c in s:
        res ^= ord(c)
    return chr(res)
print(find_odd_char('aabbc'))
```

## Section 6: String DSA Problems

Check if a string is a palindrome.

```python
def is_palindrome(s):
    return s == s[::-1]
print(is_palindrome('racecar'))
```

Count vowels and consonants using string.ascii_lowercase.

```python
import string
def count_v_c(s):
    vowels = set('aeiou')
    v, c = 0, 0
    for char in s.lower():
        if char in string.ascii_lowercase:
            if char in vowels:
                v += 1
            else:
                c += 1
    return v, c
print(count_v_c('hello world'))
```

Find all unique characters in a string.

```python
def unique_chars(s):
    return ''.join(set(s))
print(unique_chars('abracadabra'))
```

Group anagrams by using sorted word as a key.

```python
from collections import defaultdict
def group_anagrams(words):
    groups = defaultdict(list)
    for w in words:
        groups[''.join(sorted(w))].append(w)
    return list(groups.values())
print(group_anagrams(['eat', 'tea', 'tan', 'ate', 'nat', 'bat']))
```

Find the longest common prefix of a list of strings.

```python
def longest_common_prefix(strs):
    if not strs: return ''
    shortest = min(strs, key=len)
    for i, ch in enumerate(shortest):
        for other in strs:
            if other[i] != ch:
                return shortest[:i]
    return shortest
print(longest_common_prefix(['flower', 'flow', 'flight']))
```

Run-length encoding and decoding.

```python
def rle_encode(s):
    if not s: return ''
    res = []
    count = 1
    for i in range(1, len(s)):
        if s[i] == s[i-1]:
            count += 1
        else:
            res.append(s[i-1] + str(count))
            count = 1
    res.append(s[-1] + str(count))
    return ''.join(res)
print(rle_encode('aaabbc'))
```

Valid parentheses check.

```python
def is_valid(s):
    stack = []
    mapping = {')': '(', '}': '{', ']': '['}
    for char in s:
        if char in mapping:
            top = stack.pop() if stack else '#'
            if mapping[char] != top:
                return False
        else:
            stack.append(char)
    return not stack
print(is_valid('()[]{}'))
```

Reverse words in a sentence.

```python
def reverse_words(s):
    return ' '.join(s.split()[::-1])
print(reverse_words('the sky is blue'))
```

```python
# Solution
def ex4_longest_substring(s):
    char_set = set()
    l = res = 0
    for r in range(len(s)):
        while s[r] in char_set:
            char_set.remove(s[l])
            l += 1
        char_set.add(s[r])
        res = max(res, r - l + 1)
    return res
```
