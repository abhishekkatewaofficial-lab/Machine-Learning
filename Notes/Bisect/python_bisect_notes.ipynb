## Section 1: Why bisect?
Binary search theory — O(log n) vs O(n) linear search
```text
Array: [2, 4, 6, 8, 10, 12, 14, 16]
Search for 10

Linear Search (O(n)):
[2] -> [4] -> [6] -> [8] -> [10] (Found! 5 steps)

Binary Search (O(log n)):
Step 1: Mid is 8, 10 > 8 (discard left half)
Step 2: Mid is 12, 10 < 12 (discard right half)
Step 3: Mid is 10 (Found! 3 steps)
```


````python
# Manual binary search implementation vs bisect
def binary_search(arr, target):
    left, right = 0, len(arr) - 1
    while left <= right:
        mid = (left + right) // 2
        if arr[mid] == target:
            return mid
        elif arr[mid] < target:
            left = mid + 1
        else:
            right = mid - 1
    return -1

arr = [2, 4, 6, 8, 10, 12, 14]
print("Manual binary search index:", binary_search(arr, 10))
````

## Section 2: bisect_right() — Upper Bound
`bisect_right(arr, x)` returns the rightmost index where `x` can be inserted to keep `arr` sorted. All elements to the LEFT are `<= x`.
Also called 'upper_bound' in C++.


````python
# Basic bisect_right example — sorted list, find insertion point
arr = [10, 20, 30, 40, 50]
x = 35
idx = bisect.bisect_right(arr, x)
print(f"Insert {x} at index {idx}")
````

````python
# bisect_right when element exists — returns index AFTER the element
arr = [10, 20, 30, 40, 50]
x = 30
idx = bisect.bisect_right(arr, x)
print(f"bisect_right for {x} is {idx} (points to 40)")
````

````python
# bisect_right when element doesn't exist
arr = [10, 20, 30, 40, 50]
x = 25
idx = bisect.bisect_right(arr, x)
print(f"bisect_right for {x} is {idx}")
````

````python
# bisect_right with lo and hi parameters — search in subarray
arr = [10, 20, 30, 40, 50, 60, 70]
x = 45
# Search only between indices 2 and 5 (i.e. [30, 40, 50])
idx = bisect.bisect_right(arr, x, lo=2, hi=5)
print(f"bisect_right for {x} in subarray is {idx}")
````

````python
# Visual print showing where the pointer lands
arr = [1, 3, 5, 7, 9]
x = 6
idx = bisect.bisect_right(arr, x)
print("Array:", arr)
print(" " * 8 + "  " * idx + "^")
print(" " * 8 + f"Insert {x} here")
````

## Section 3: bisect_left() — Lower Bound
`bisect_left(arr, x)` returns the leftmost index where `x` can be inserted. All elements to the LEFT are `< x` (strictly less).
Also called 'lower_bound' in C++.


````python
# Basic bisect_left example
arr = [10, 20, 30, 40, 50]
x = 35
idx = bisect.bisect_left(arr, x)
print(f"Insert {x} at index {idx}")
````

````python
# bisect_left when element exists — returns index OF the element
arr = [10, 20, 30, 40, 50]
x = 30
idx = bisect.bisect_left(arr, x)
print(f"bisect_left for {x} is {idx} (points to {arr[idx]})")
````

````python
# Difference between bisect_left and bisect_right with duplicate elements
arr = [1, 2, 2, 2, 3]
x = 2
left_idx = bisect.bisect_left(arr, x)
right_idx = bisect.bisect_right(arr, x)
print(f"Array: {arr}")
print(f"bisect_left({x}): {left_idx}")
print(f"bisect_right({x}): {right_idx}")
````

````python
# Visual comparison: bisect_left vs bisect_right on [1,2,2,2,3]
arr = [1, 2, 2, 2, 3]
x = 2
left_idx = bisect.bisect_left(arr, x)
right_idx = bisect.bisect_right(arr, x)

print("Array:", arr)
print(" " * 8 + "  " * left_idx + "^ (left)")
print(" " * 8 + "  " * right_idx + "^ (right)")
````

## Section 4: bisect() — alias


````python
# bisect.bisect() is same as bisect.bisect_right()
arr = [1, 2, 2, 2, 3]
print("bisect:", bisect.bisect(arr, 2))
print("bisect_right:", bisect.bisect_right(arr, 2))
````

````python
# When to use which — general rule
# Use bisect_left when you want to FIND an element or get elements strictly smaller.
# Use bisect_right when you want to COUNT occurrences or get elements <= x.
````

## Section 5: insort_right() and insort_left()

````python
# insort_right — insert and maintain order
arr = [1, 3, 5, 7]
bisect.insort_right(arr, 4)
print("After insort_right(4):", arr)
````

````python
# insort_left — insert at leftmost valid position
arr = [1, 2, 2, 4]
bisect.insort_left(arr, 2)
print("After insort_left(2):", arr)
````

## Section 6: Core DSA Patterns with bisect


````python
# Pattern 1: Search (Is element present?)
# Check if x exists in sorted array using bisect_left
def contains(arr, x):
    i = bisect.bisect_left(arr, x)
    return i < len(arr) and arr[i] == x

arr = [10, 20, 30, 40]
print("Contains 30?", contains(arr, 30))
print("Contains 25?", contains(arr, 25))
````

````python
# Pattern 2: Count occurrences of x
def count_x(arr, x):
    return bisect.bisect_right(arr, x) - bisect.bisect_left(arr, x)
````

````python
# Example with [1,2,2,2,3,4] — count 2s
arr = [1, 2, 2, 2, 3, 4]
print("Count of 2:", count_x(arr, 2))
````

````python
# Pattern 3: Count elements < x
# bisect_left(arr, x) gives count of elements strictly less than x
arr = [10, 20, 30, 40]
print("Elements < 25:", bisect.bisect_left(arr, 25))
````

````python
# Pattern 4: Count elements <= x
# bisect_right(arr, x) gives count of elements <= x
print("Elements <= 20:", bisect.bisect_right(arr, 20))
````

````python
# Pattern 5: Count elements in range [lo, hi]
def count_range(arr, lo, hi):
    return bisect.bisect_right(arr, hi) - bisect.bisect_left(arr, lo)
````

````python
# DSA — count elements in range query
arr = [1, 3, 5, 7, 9, 11]
print("Count in [4, 10]:", count_range(arr, 4, 10))
````

````python
# Pattern 6: Find floor (largest element <= x)
def find_floor(arr, x):
    i = bisect.bisect_right(arr, x) - 1
    return arr[i] if i >= 0 else None
````

````python
# DSA — floor in sorted array
arr = [2, 4, 6, 8]
print("Floor of 5:", find_floor(arr, 5))
print("Floor of 1:", find_floor(arr, 1))
````

````python
# Pattern 7: Find ceiling (smallest element >= x)
def find_ceiling(arr, x):
    i = bisect.bisect_left(arr, x)
    return arr[i] if i < len(arr) else None
````

````python
# DSA — ceiling in sorted array
arr = [2, 4, 6, 8]
print("Ceiling of 5:", find_ceiling(arr, 5))
print("Ceiling of 9:", find_ceiling(arr, 9))
````

````python
# Pattern 8: K-th position
# Find rank of x in sorted array (1-indexed position)
arr = [10, 20, 30, 40]
print("Rank of 30 (1-indexed):", bisect.bisect_left(arr, 30) + 1)
````

````python
# Find k-th smallest using binary search + bisect
# Given multiple arrays, find k-th smallest element overall
arrs = [[1, 4, 7], [2, 5, 8], [3, 6, 9]]
# (A common technique involves binary searching the answer space and counting elements <= mid)
def count_less_equal(arrs, val):
    return sum(bisect.bisect_right(a, val) for a in arrs)

low, high = 1, 9
k = 5
ans = -1
while low <= high:
    mid = (low + high) // 2
    if count_less_equal(arrs, mid) >= k:
        ans = mid
        high = mid - 1
    else:
        low = mid + 1
print("5th smallest element:", ans)
````

## Section 7: DSA Problems using bisect


````python
# DSA — Longest Increasing Subsequence (LIS) using bisect_left
# Classic O(n log n) LIS algorithm using patience sorting
# Explain the tails array approach
def lengthOfLIS(nums):
    tails = []
    for num in nums:
        idx = bisect.bisect_left(tails, num)
        if idx == len(tails):
            tails.append(num)
        else:
            tails[idx] = num
    return len(tails)

print("LIS of [10,9,2,5,3,7,101,18]:", lengthOfLIS([10,9,2,5,3,7,101,18]))
````

````python
# DSA — Count smaller elements to the right (using bisect + sorted insert)
def countSmaller(nums):
    res = []
    sorted_list = []
    for num in reversed(nums):
        idx = bisect.bisect_left(sorted_list, num)
        res.append(idx)
        bisect.insort(sorted_list, num)
    return res[::-1]

print("Smaller to right for [5,2,6,1]:", countSmaller([5,2,6,1]))
````

````python
# DSA — Median of a stream (use two sorted lists + bisect)
# Note: For stream median, heaps are better, but bisect + insort is O(N) per insert
class MedianFinder:
    def __init__(self):
        self.arr = []
    def addNum(self, num: int) -> None:
        bisect.insort(self.arr, num)
    def findMedian(self) -> float:
        n = len(self.arr)
        if n % 2 == 1: return float(self.arr[n//2])
        return (self.arr[n//2 - 1] + self.arr[n//2]) / 2.0
        
mf = MedianFinder()
mf.addNum(1); mf.addNum(2); 
print("Median:", mf.findMedian())
````

````python
# DSA — Search in rotated sorted array concept explained
# Bisect requires a sorted array. If rotated, find the pivot first (with manual binary search),
# then use bisect on the appropriate sorted half.
def search_rotated(nums, target):
    if not nums: return -1
    lo, hi = 0, len(nums)-1
    while lo < hi:
        mid = (lo + hi) // 2
        if nums[mid] > nums[hi]: lo = mid + 1
        else: hi = mid
    
    pivot = lo
    # Now we have two sorted halves: nums[:pivot] and nums[pivot:]
    if target >= nums[pivot] and target <= nums[-1]:
        idx = bisect.bisect_left(nums, target, lo=pivot, hi=len(nums))
    else:
        idx = bisect.bisect_left(nums, target, lo=0, hi=pivot)
        
    return idx if idx < len(nums) and nums[idx] == target else -1

print("Search 0 in [4,5,6,7,0,1,2]:", search_rotated([4,5,6,7,0,1,2], 0))
````

````python
# DSA — Time Based Key-Value Store — bisect on timestamps
class TimeMap:
    import collections
    def __init__(self):
        self.store = collections.defaultdict(list)
    def set(self, key: str, value: str, timestamp: int) -> None:
        self.store[key].append((timestamp, value))
    def get(self, key: str, timestamp: int) -> str:
        arr = self.store[key]
        idx = bisect.bisect_right(arr, (timestamp, chr(127)))
        return arr[idx-1][1] if idx > 0 else ""
````

````python
# DSA — H-Index calculation using bisect
def hIndex(citations):
    citations.sort()
    n = len(citations)
    lo, hi = 0, n
    while lo < hi:
        mid = (lo + hi + 1) // 2
        # Count elements >= mid
        idx = bisect.bisect_left(citations, mid)
        count = n - idx
        if count >= mid:
            lo = mid
        else:
            hi = mid - 1
    return lo

print("H-index [3,0,6,1,5]:", hIndex([3,0,6,1,5]))
````

````python
# DSA — Find First and Last Position of element in sorted array
def searchRange(nums, target):
    left = bisect.bisect_left(nums, target)
    if left == len(nums) or nums[left] != target:
        return [-1, -1]
    right = bisect.bisect_right(nums, target) - 1
    return [left, right]

print("Range of 8 in [5,7,7,8,8,10]:", searchRange([5,7,7,8,8,10], 8))
````

````python
# DSA — Minimum number of meetings rooms needed (event scheduling)
def minMeetingRooms(intervals):
    starts = sorted([i[0] for i in intervals])
    ends = sorted([i[1] for i in intervals])
    rooms = 0
    end_ptr = 0
    for start in starts:
        if start < ends[end_ptr]:
            rooms += 1
        else:
            end_ptr += 1
    return rooms
````

## Section 8: bisect on Custom Objects
bisect works on any comparable objects, not just integers


````python
# bisect on list of strings (alphabetical order)
words = ["apple", "banana", "cherry", "date"]
print("Insert 'cat':", bisect.bisect_left(words, "cat"))
````

````python
# bisect on list of tuples (sorted by first element)
events = [(1, "wakeup"), (5, "lunch"), (9, "dinner")]
print("Insert event at time 3:", bisect.bisect_left(events, (3, "snack")))
````

````python
# Using key parameter (Python 3.10+) — bisect with key function
students = [{"name": "Alice", "score": 85}, {"name": "Bob", "score": 92}, {"name": "Charlie", "score": 98}]
idx = bisect.bisect_left(students, 90, key=lambda x: x["score"])
print("Insert student with score 90 at:", idx)
````

````python
# Workaround for older Python — use a key-projected list
scores = [s["score"] for s in students]
idx = bisect.bisect_left(scores, 90)
print("Using projected list, idx:", idx)
````

````python
# Count words alphabetically before a given word
vocab = ["ant", "bat", "cat", "dog"]
print("Words before 'cat':", bisect.bisect_left(vocab, "cat"))
````

````python
# Find all words starting with a prefix using bisect_left and bisect_right
vocab = ["apple", "applet", "application", "banana", "bat"]
prefix = "app"
# Start of prefix
start = bisect.bisect_left(vocab, prefix)
# End of prefix (prefix + ~ which is a high char)
end = bisect.bisect_right(vocab, prefix + chr(127))
print("Words starting with 'app':", vocab[start:end])
````

````python
# Autocomplete simulation — find words with common prefix using bisect
def autocomplete(vocab, prefix):
    start = bisect.bisect_left(vocab, prefix)
    end = bisect.bisect_right(vocab, prefix + chr(127))
    return vocab[start:end]

print("Autocomplete 'ba':", autocomplete(vocab, "ba"))
````

## Section 11: Performance Comparison


````python
# Matplotlib bar chart — linear search O(n) vs bisect O(log n) on large arrays
import timeit
import bisect

arr = list(range(1000000))

def linear_search():
    return 999999 in arr

def binary_search():
    return bisect.bisect_left(arr, 999999)

lin_time = timeit.timeit(linear_search, number=100)
bin_time = timeit.timeit(binary_search, number=100)

print(f"Linear search time: {lin_time:.6f}s")
print(f"Binary search time: {bin_time:.6f}s")

# In a real notebook, you'd do:
# plt.bar(['Linear', 'Binary'], [lin_time, bin_time])
# plt.ylabel("Time (seconds)")
# plt.show()
````

````python
# Insort vs sort every time — performance comparison
def using_insort():
    res = []
    for i in range(1000, 0, -1):
        bisect.insort(res, i)

def using_sort():
    res = []
    for i in range(1000, 0, -1):
        res.append(i)
        res.sort()

ins_time = timeit.timeit(using_insort, number=10)
srt_time = timeit.timeit(using_sort, number=10)
print(f"Insort time: {ins_time:.6f}s")
print(f"Sort every time: {srt_time:.6f}s")
````

## Section 12: Cheat Sheet
| Pattern | Code |
|---|---|
| Is element present | `idx = bisect_left(a, x); return idx < len(a) and a[idx] == x` |
| Count elements < x | `bisect_left(a, x)` |
| Count elements <= x | `bisect_right(a, x)` |
| Count occurrences of x | `bisect_right(a, x) - bisect_left(a, x)` |
| Count in range [lo, hi] | `bisect_right(a, hi) - bisect_left(a, lo)` |
| Find floor (<=x) | `idx = bisect_right(a, x) - 1; return a[idx] if idx >= 0 else None` |
| Find ceiling (>=x) | `idx = bisect_left(a, x); return a[idx] if idx < len(a) else None` |
| LIS (insert/replace) | `idx = bisect_left(tails, x); if idx == len(tails): tails.append(x) else: tails[idx] = x` |
