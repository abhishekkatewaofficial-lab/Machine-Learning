# Section 1: Why `collections`? 🤔
---

## The Problem with Plain `dict`, `list`, `tuple`

| Situation | Plain Python | `collections` solution |
|-----------|-------------|------------------------|
| Count frequency of items | Manual loop + `if key in dict` | `Counter` — one line |
| Group items into lists | KeyError if key missing | `defaultdict(list)` |
| Fast remove from front of list | `list.pop(0)` is O(n) SLOW | `deque.popleft()` is O(1) FAST |
| Access tuple fields by name | `point[0]`, `point[1]` — confusing | `namedtuple` — `point.x`, `point.y` |
| Move dict key to front/back | Complex manual code | `OrderedDict.move_to_end()` |
| Merge dicts with priority | Manual update loops | `ChainMap` — one line |

**`collections` doesn't replace list/dict — it gives you the RIGHT tool for specific jobs.**

````python
from collections import Counter, defaultdict, deque, namedtuple, OrderedDict, ChainMap
print('All collections tools imported successfully!')
print('Tools available:', ['Counter', 'defaultdict', 'deque', 'namedtuple', 'OrderedDict', 'ChainMap'])
````

# Section 2: `Counter` ⭐ Most Used
---

## What is `Counter`?

A `Counter` is a special dictionary where:
- **Keys** = the items you're counting
- **Values** = how many times each item appears

```python
from collections import Counter
c = Counter(['a', 'b', 'a', 'c', 'a', 'b'])
# Result: Counter({'a': 3, 'b': 2, 'c': 1})
```

No need to write: `if key in dict: dict[key] += 1 else: dict[key] = 1`
Counter does it **automatically**!

### Basic Usage — Create a Counter

````python
from collections import Counter

# Count characters in a string
text    = 'hello world'
counter = Counter(text)

print('Text   :', text)
print('Counter:', counter)
print('Count of l:', counter['l'])
print('Count of z (not present):', counter['z'])  # Returns 0, not KeyError!
````

````python
from collections import Counter

# Count elements in a list
fruits  = ['apple', 'banana', 'apple', 'cherry', 'banana', 'apple']
counter = Counter(fruits)

print('Fruits :', fruits)
print('Counter:', counter)
````

````python
from collections import Counter

# Count words in a sentence
sentence = 'the quick brown fox jumps over the lazy dog the fox'
words    = sentence.split()
counter  = Counter(words)

print('Word counts:')
for word, count in counter.items():
    print(f'  {word!r:12} → {count}')
````

### `most_common(n)` — Get Top N Most Frequent Items

````python
from collections import Counter

# Top 3 most common words
sentence = 'I love python python is great I love python coding'
counter  = Counter(sentence.split())

print('Top 3 words:')
for word, count in counter.most_common(3):
    print(f'  {word!r:12}: {"█" * count} ({count})')
````

````python
from collections import Counter

# Find the most common character in a password (security check)
password = 'aabbccaaddeeff'
counter  = Counter(password)

print(f'Password: {password!r}')
print('Most repeated chars:', counter.most_common(3))
print('\nSecurity tip: avoid passwords with repeated characters!')
````

### Visualize Counter with Bar Chart

````python
import matplotlib.pyplot as plt
from collections import Counter

# Visualize word frequency in a paragraph
text    = 'python is great python is easy python is powerful java is also popular java is verbose'
counter = Counter(text.split())

words, counts = zip(*counter.most_common(6))

plt.figure(figsize=(9, 4))
bars = plt.bar(words, counts, color=['steelblue','tomato','seagreen','orange','purple','brown'])
plt.title('Word Frequency', fontsize=14, fontweight='bold')
plt.xlabel('Words')
plt.ylabel('Count')
for bar, count in zip(bars, counts):
    plt.text(bar.get_x() + bar.get_width()/2, bar.get_height() + 0.05, str(count), ha='center', fontweight='bold')
plt.tight_layout()
plt.show()
````

### Counter Arithmetic — Add, Subtract, Intersect, Union

````python
from collections import Counter

# Add two counters (combine stock from two warehouses)
warehouse1 = Counter({'apple': 50, 'banana': 30, 'cherry': 20})
warehouse2 = Counter({'apple': 10, 'banana': 40, 'mango': 15})

total_stock = warehouse1 + warehouse2
print('Warehouse 1:', dict(warehouse1))
print('Warehouse 2:', dict(warehouse2))
print('Total stock:', dict(total_stock))
````

````python
from collections import Counter

# Subtract (how many items were sold from the store?)
initial_stock = Counter({'apple': 50, 'banana': 30, 'cherry': 20})
sold          = Counter({'apple': 15, 'banana': 30, 'cherry': 5})

remaining = initial_stock - sold
print('Initial stock:', dict(initial_stock))
print('Sold         :', dict(sold))
print('Remaining    :', dict(remaining))
````

````python
from collections import Counter

# & (intersection) — common items with minimum count
# | (union)        — all items with maximum count

c1 = Counter({'a': 3, 'b': 2, 'c': 1})
c2 = Counter({'a': 1, 'b': 5, 'd': 3})

print('c1          :', dict(c1))
print('c2          :', dict(c2))
print('c1 & c2 (min):', dict(c1 & c2))  # min of shared keys
print('c1 | c2 (max):', dict(c1 | c2))  # max of all keys
````

# Section 3: `defaultdict` — Dict Without KeyError
---

## The Problem `defaultdict` Solves

With a regular dict, accessing a missing key crashes your program:
```python
d = {}
d['new_key'] += 1   # ❌ KeyError: 'new_key'
```

With `defaultdict`, it automatically creates a default value:
```python
from collections import defaultdict
d = defaultdict(int)
d['new_key'] += 1   # ✅ Works! Starts from 0 automatically
```

You pass a **factory function** that creates the default value:
| `defaultdict(int)` | Default = `0` |
| `defaultdict(list)` | Default = `[]` |
| `defaultdict(set)` | Default = `set()` |
| `defaultdict(str)` | Default = `''` |

### `defaultdict(int)` — Counting without KeyError

````python
from collections import defaultdict

# Count word frequency — no need for 'if word in dict' check!
sentence = 'the cat sat on the mat the cat is fat'
word_count = defaultdict(int)

for word in sentence.split():
    word_count[word] += 1   # No KeyError even on first access!

print('Word counts:')
for word, count in sorted(word_count.items()):
    print(f'  {word:8}: {count}')
````

````python
from collections import defaultdict

# Count votes in an election
votes = ['Alice', 'Bob', 'Alice', 'Charlie', 'Bob', 'Alice', 'Bob', 'Charlie', 'Alice']
vote_count = defaultdict(int)

for candidate in votes:
    vote_count[candidate] += 1

print('Election Results:')
for candidate, count in sorted(vote_count.items(), key=lambda x: -x[1]):
    print(f'  {candidate:10}: {"█" * count} ({count} votes)')
winner = max(vote_count, key=vote_count.get)
print(f'\n🏆 Winner: {winner}')
````

### `defaultdict(list)` — Grouping items

````python
from collections import defaultdict

# Group students by their grade
students = [
    ('Alice', 'A'), ('Bob', 'B'), ('Charlie', 'A'),
    ('Diana', 'C'), ('Eve', 'B'), ('Frank', 'A')
]

grade_groups = defaultdict(list)
for name, grade in students:
    grade_groups[grade].append(name)

print('Students by Grade:')
for grade in sorted(grade_groups):
    print(f'  Grade {grade}: {grade_groups[grade]}')
````

````python
from collections import defaultdict

# Group products by category
products = [
    ('Laptop',   'Electronics'),
    ('Apple',    'Fruits'),
    ('Phone',    'Electronics'),
    ('Banana',   'Fruits'),
    ('Shirt',    'Clothing'),
    ('Tablet',   'Electronics'),
    ('Jeans',    'Clothing'),
]

by_category = defaultdict(list)
for product, category in products:
    by_category[category].append(product)

print('Products by Category:')
for category, items in sorted(by_category.items()):
    print(f'  {category}: {items}')
````

````python
from collections import defaultdict

# Build an adjacency list for a graph (friends network)
friendships = [('Alice','Bob'), ('Alice','Charlie'), ('Bob','Diana'), ('Charlie','Eve')]

graph = defaultdict(list)
for a, b in friendships:
    graph[a].append(b)
    graph[b].append(a)   # undirected: both directions

print('Friend Network (Adjacency List):')
for person, friends in sorted(graph.items()):
    print(f'  {person}: {friends}')
````

### `defaultdict(set)` — Grouping unique items

````python
from collections import defaultdict

# Track which cities each user visited (no duplicate cities)
visits = [
    ('Alice', 'Delhi'), ('Bob', 'Mumbai'), ('Alice', 'Mumbai'),
    ('Alice', 'Delhi'),  # Alice visited Delhi again — won't be duplicated!
    ('Bob', 'Delhi'), ('Charlie', 'Mumbai'), ('Bob', 'Mumbai')  # Bob Mumbai again
]

cities_visited = defaultdict(set)
for user, city in visits:
    cities_visited[user].add(city)  # set auto-removes duplicates

print('Unique cities visited:')
for user, cities in sorted(cities_visited.items()):
    print(f'  {user}: {sorted(cities)}')
````

### Nested `defaultdict` — Dict of dicts

````python
from collections import defaultdict

# Nested defaultdict: track sales[city][product] = amount
sales = defaultdict(lambda: defaultdict(int))

data = [
    ('Delhi', 'Laptop', 5000),
    ('Delhi', 'Phone',  3000),
    ('Mumbai', 'Laptop', 7000),
    ('Delhi', 'Laptop', 2000),   # Second Delhi Laptop sale
    ('Mumbai', 'Phone',  4000),
]

for city, product, amount in data:
    sales[city][product] += amount

print('Sales by City and Product:')
for city, products in sorted(sales.items()):
    print(f'  {city}:')
    for product, amount in sorted(products.items()):
        print(f'    {product:8}: ₹{amount:,}')
````

# Section 4: `deque` — Fast Queue & Stack
---

## What is `deque`? (pronounced 'deck')

A **double-ended queue** — you can efficiently add/remove from BOTH ends.

### Why not just use a `list`?
| Operation | `list` | `deque` |
|-----------|--------|--------|
| Add to end | `append()` — O(1) ✅ | `append()` — O(1) ✅ |
| Remove from end | `pop()` — O(1) ✅ | `pop()` — O(1) ✅ |
| Add to front | `insert(0, x)` — **O(n) SLOW** ❌ | `appendleft()` — **O(1) FAST** ✅ |
| Remove from front | `pop(0)` — **O(n) SLOW** ❌ | `popleft()` — **O(1) FAST** ✅ |

**Use `deque` whenever you need to add/remove from the front of a sequence!**

### Basic Operations — append, appendleft, pop, popleft

````python
from collections import deque

# Create a deque
dq = deque([1, 2, 3, 4, 5])
print('Initial deque:', dq)

dq.append(6)          # Add to right end
print('After append(6)     :', dq)

dq.appendleft(0)      # Add to left end
print('After appendleft(0) :', dq)
````

````python
from collections import deque

dq = deque([0, 1, 2, 3, 4, 5, 6])
print('Initial deque:', dq)

right = dq.pop()       # Remove from right end
print(f'After pop()       : {dq}  (removed: {right})')

left = dq.popleft()    # Remove from left end
print(f'After popleft()   : {dq}  (removed: {left})')
````

### `rotate(n)` — Rotate elements left or right

````python
from collections import deque

# Rotate right (positive n)
dq = deque([1, 2, 3, 4, 5])
print('Original     :', dq)

dq.rotate(2)    # move 2 elements from right to left
print('rotate(2)    :', dq)

dq.rotate(-2)   # undo — move 2 elements from left to right
print('rotate(-2)   :', dq)
````

````python
from collections import deque

# Real use: Round-robin scheduling using rotate
tasks = deque(['Task_A', 'Task_B', 'Task_C', 'Task_D'])
workers = ['Worker1', 'Worker2']

print('Assigning tasks with round-robin:')
assigned = []
while tasks:
    for worker in workers:
        if tasks:
            task = tasks.popleft()
            assigned.append((worker, task))
            print(f'  {worker} gets {task}')
````

### `maxlen` — Fixed-size sliding window

````python
from collections import deque

# Keep only the last 3 readings (sliding window)
# When maxlen is reached, oldest item is auto-removed!

last_3 = deque(maxlen=3)
sensor_data = [10, 20, 30, 40, 50, 60, 70]

print('Processing sensor data with window size 3:')
for value in sensor_data:
    last_3.append(value)
    avg = sum(last_3) / len(last_3)
    print(f'  Added {value:2d} | Window: {list(last_3)} | Avg: {avg:.1f}')
````

````python
from collections import deque

# Browser history — keep last 5 visited pages
history = deque(maxlen=5)
pages   = ['google.com', 'github.com', 'stackoverflow.com',
           'youtube.com', 'python.org', 'docs.python.org', 'reddit.com']

for page in pages:
    history.append(page)
    print(f'Visited: {page:25} | History: {list(history)}')
````

### `deque` as a Queue (FIFO) and as a Stack (LIFO)

````python
from collections import deque

# Queue (FIFO — First In, First Out) — like a ticket queue
ticket_queue = deque()

# People join the queue
for person in ['Alice', 'Bob', 'Charlie', 'Diana']:
    ticket_queue.append(person)
    print(f'  {person} joined the queue: {list(ticket_queue)}')

print()
# People are served from the front
while ticket_queue:
    served = ticket_queue.popleft()  # Remove from FRONT
    print(f'  Serving: {served:10} | Remaining: {list(ticket_queue)}')
````

````python
from collections import deque

# Stack (LIFO — Last In, First Out) — like undo history in a text editor
undo_stack = deque()

actions = ['type "Hello"', 'type " World"', 'bold text', 'change font']
for action in actions:
    undo_stack.append(action)
    print(f'  Did: {action:20} | Stack: {list(undo_stack)}')

print()
# Undo — remove from the RIGHT (last action)
print('Undoing actions:')
while undo_stack:
    undone = undo_stack.pop()  # Remove from END
    print(f'  Undid: {undone:20} | Stack: {list(undo_stack)}')
````

# Section 5: `namedtuple` — Readable Lightweight Class
---

## What is `namedtuple`?

A `namedtuple` is a **tuple** where each position has a **name**.

```python
# Regular tuple — confusing!
point = (10, 20)
x = point[0]    # What is index 0? x or y? Hard to remember!

# namedtuple — crystal clear!
Point = namedtuple('Point', ['x', 'y'])
point = Point(10, 20)
x = point.x     # Obviously x! Much more readable.
```

**Benefits:**
- ✅ More readable than tuples (access by name)
- ✅ Less memory than a full class or dict
- ✅ Immutable (cannot accidentally change values)
- ✅ Supports tuple unpacking

### Creating and Using namedtuple

````python
from collections import namedtuple

# Create a Point namedtuple
Point = namedtuple('Point', ['x', 'y'])

p = Point(10, 20)
print('Point        :', p)
print('Access by name:', p.x, p.y)       # recommended
print('Access by index:', p[0], p[1])    # still works like a tuple
print('Is a tuple?   :', isinstance(p, tuple))
````

````python
from collections import namedtuple

# Student record
Student = namedtuple('Student', ['name', 'age', 'grade', 'score'])

alice = Student(name='Alice', age=20, grade='A', score=95)
bob   = Student('Bob', 21, 'B', 78)

print('Alice:', alice)
print('Bob  :', bob)
print()
print(f'Alice grade: {alice.grade}, score: {alice.score}')
print(f'Bob grade  : {bob.grade},   score: {bob.score}')
````

### Comparison: tuple vs dict vs namedtuple

````python
from collections import namedtuple

# Storing a product — 3 ways

# Way 1: Regular tuple (confusing, no names)
product_tuple = ('Laptop', 'Electronics', 75000, 4.5)
print('Tuple access (confusing):', product_tuple[2])   # What is index 2?

# Way 2: Dictionary (clear but uses more memory)
product_dict = {'name': 'Laptop', 'category': 'Electronics', 'price': 75000, 'rating': 4.5}
print('Dict access (clear)     :', product_dict['price'])

# Way 3: namedtuple (clear + lightweight!)
Product = namedtuple('Product', ['name', 'category', 'price', 'rating'])
product = Product('Laptop', 'Electronics', 75000, 4.5)
print('namedtuple access (best):', product.price)
````

# Section 6: `OrderedDict` — Dict with Order Control
---

## What is `OrderedDict`?

In **Python 3.7+**, regular dicts already remember insertion order.
So `OrderedDict` is mainly used when you need **extra order operations** that regular dict doesn't have:

1. `move_to_end(key)` — move a key to the front or back
2. Equality considers order (two OrderedDicts with same items in different order are NOT equal)
3. Great for implementing **LRU Cache** pattern

````python
from collections import OrderedDict

# Basic OrderedDict
od = OrderedDict()
od['first']  = 1
od['second'] = 2
od['third']  = 3

print('OrderedDict:', od)
print('Keys in order:', list(od.keys()))
````

````python
from collections import OrderedDict

# move_to_end — move a key to the back (last=True) or front (last=False)
od = OrderedDict([('a', 1), ('b', 2), ('c', 3), ('d', 4)])
print('Original        :', list(od.keys()))

od.move_to_end('a')           # move 'a' to the END
print('After move a→end:', list(od.keys()))

od.move_to_end('d', last=False)  # move 'd' to the FRONT
print('After move d→front:', list(od.keys()))
````

````python
from collections import OrderedDict

# LRU Cache (Least Recently Used) — a common interview pattern!
# The idea: when cache is full, remove the LEAST recently used item

class LRUCache:
    def __init__(self, capacity):
        self.cache    = OrderedDict()
        self.capacity = capacity

    def get(self, key):
        if key not in self.cache:
            return -1
        self.cache.move_to_end(key)  # Mark as recently used
        return self.cache[key]

    def put(self, key, value):
        if key in self.cache:
            self.cache.move_to_end(key)
        self.cache[key] = value
        if len(self.cache) > self.capacity:
            self.cache.popitem(last=False)  # Remove LEAST recently used (front)

cache = LRUCache(3)
cache.put('page1', 'data1')
cache.put('page2', 'data2')
cache.put('page3', 'data3')
print('Cache after 3 puts:', list(cache.cache.keys()))
cache.get('page1')           # page1 is now most recent
cache.put('page4', 'data4') # cache full → page2 evicted (least recent)
print('Cache after page4   :', list(cache.cache.keys()))
````

````python
from collections import OrderedDict

# OrderedDict considers ORDER in equality (regular dict does not!)
od1 = OrderedDict([('a', 1), ('b', 2)])
od2 = OrderedDict([('b', 2), ('a', 1)])
d1  = {'a': 1, 'b': 2}
d2  = {'b': 2, 'a': 1}

print('od1 == od2 (OrderedDict):', od1 == od2)  # False — order differs
print('d1  == d2  (regular dict):', d1  == d2)   # True  — order ignored
````

# Section 7: `ChainMap` — Multiple Dicts as One View
---

## What is `ChainMap`?

`ChainMap` **combines multiple dictionaries** into a single view without copying the data.

**Key behaviour:** When you look up a key, it **searches dicts in order** (first dict wins).

```python
cm = ChainMap(user_settings, app_defaults, system_defaults)
# Lookup order: user_settings → app_defaults → system_defaults
# First match wins!
```

**Perfect for:** configuration systems with layers (user > app > default)

````python
from collections import ChainMap

# Combine two dicts — first dict has priority
dict1 = {'a': 1, 'b': 2}
dict2 = {'b': 20, 'c': 3}  # 'b' exists in both!

cm = ChainMap(dict1, dict2)
print('dict1:', dict1)
print('dict2:', dict2)
print('ChainMap:', dict(cm))
print('\nKey b =', cm['b'])  # dict1 wins! (it comes first)
````

````python
from collections import ChainMap

# Real use: layered configuration (user > app > system defaults)
system_defaults = {'theme': 'light', 'font_size': 12, 'lang': 'en', 'timeout': 30}
app_defaults    = {'theme': 'dark',  'font_size': 14}
user_settings   = {'font_size': 18}   # user only changed font size

# Priority: user → app → system
config = ChainMap(user_settings, app_defaults, system_defaults)

print('Final Config:')
for key in ['theme', 'font_size', 'lang', 'timeout']:
    print(f'  {key:12}: {config[key]}')
````

````python
from collections import ChainMap

# Updates always go to the FIRST map
defaults = {'color': 'blue', 'size': 'medium'}
user     = {}

config = ChainMap(user, defaults)
print('Before update:', dict(config))

config['color'] = 'red'   # goes into 'user' dict, not 'defaults'
print('After  update:', dict(config))
print('User dict now:', user)      # user dict updated
print('Defaults still:', defaults)  # defaults unchanged
````

````python
from collections import ChainMap

# new_child() — create a new scope (like local vs global variables)
global_vars = ChainMap({'x': 10, 'y': 20})
print('Global scope:', dict(global_vars))

# Create a new local scope that 'inherits' from global
local_vars = global_vars.new_child({'x': 999, 'z': 30})  # x is overridden locally
print('Local scope :', dict(local_vars))
print('\nLocal x  =', local_vars['x'])   # local value
print('Local y  =', local_vars['y'])   # falls through to global
print('Global x =', global_vars['x'])  # global unchanged
````

# Section 8: Deep Dive — Strings with `collections` 🟢🟡

````python
from collections import deque

# Use deque as an efficient queue for BFS (Breadth-First Search)
# BFS visits nodes level by level in a graph/tree

def bfs(graph, start):
    visited = []
    queue   = deque([start])
    seen    = {start}

    while queue:
        node = queue.popleft()  # O(1) — efficient!
        visited.append(node)
        for neighbor in graph.get(node, []):
            if neighbor not in seen:
                seen.add(neighbor)
                queue.append(neighbor)
    return visited

graph = {'A': ['B','C'], 'B': ['D','E'], 'C': ['F'], 'D':[], 'E':[], 'F':[]}
result = bfs(graph, 'A')
print('BFS traversal order:', result)
````

````python
from collections import ChainMap

# Merge multiple dicts with priority — first dict wins
user_prefs   = {'dark_mode': True, 'font_size': 16}
team_prefs   = {'dark_mode': False, 'font_size': 14, 'notifications': True}
system_prefs = {'dark_mode': False, 'font_size': 12, 'notifications': False, 'timeout': 30}

final_config = ChainMap(user_prefs, team_prefs, system_prefs)
print('Final configuration (user overrides team overrides system):')
for key in ['dark_mode', 'font_size', 'notifications', 'timeout']:
    print(f'  {key:20}: {final_config[key]}')
````

````python
from collections import Counter
import itertools

# Counter + chain.from_iterable — count all words across many lists
documents = [
    ['python', 'is', 'great', 'python'],
    ['java', 'is', 'verbose', 'python', 'is', 'better'],
    ['I', 'love', 'python', 'coding'],
]

all_words  = itertools.chain.from_iterable(documents)
word_count = Counter(all_words)

print('Word frequency across all documents:')
for word, count in word_count.most_common():
    print(f'  {word:10}: {"█" * count} ({count})')
````

````python
from collections import defaultdict
import itertools

# defaultdict + groupby — proper grouping with clean storage
students = [
    {'name': 'Alice',   'grade': 'A', 'score': 92},
    {'name': 'Bob',     'grade': 'B', 'score': 78},
    {'name': 'Charlie', 'grade': 'A', 'score': 88},
    {'name': 'Diana',   'grade': 'C', 'score': 65},
    {'name': 'Eve',     'grade': 'B', 'score': 82},
]

sorted_students = sorted(students, key=lambda s: s['grade'])
grade_map = defaultdict(list)

for grade, group in itertools.groupby(sorted_students, key=lambda s: s['grade']):
    for student in group:
        grade_map[grade].append((student['name'], student['score']))

print('Grades with student details:')
for grade, members in sorted(grade_map.items()):
    avg = sum(s for _, s in members) / len(members)
    print(f'  Grade {grade}: {members} | Avg: {avg:.1f}')
````

````python
from collections import Counter, deque
import itertools

# deque + islice — efficient rolling window with Counter
# Find the most common element in every 3-element window of a list

data   = [1, 2, 1, 3, 2, 1, 4, 2, 1, 3]
window = deque(maxlen=3)
results = []

for item in data:
    window.append(item)
    if len(window) == 3:
        most_common = Counter(window).most_common(1)[0][0]
        results.append((list(window).copy(), most_common))

print('Rolling window (size 3) — most common element:')
for win, most in results:
    print(f'  Window {win} → Most common: {most}')
````

````python
from collections import namedtuple, Counter
import itertools

# namedtuple + itertools.combinations — generate all team matchups
Player = namedtuple('Player', ['name', 'rating'])

players = [
    Player('Alice',   95),
    Player('Bob',     82),
    Player('Charlie', 78),
    Player('Diana',   90),
]

print('All 1v1 matchups:')
for p1, p2 in itertools.combinations(players, 2):
    diff     = abs(p1.rating - p2.rating)
    balanced = '✅ Balanced' if diff < 15 else '⚠️  Unbalanced'
    print(f'  {p1.name:10} ({p1.rating}) vs {p2.name:10} ({p2.rating}) | Diff: {diff:2d} | {balanced}')
````

# Section 13: ⚡ `collections` Cheat Sheet
---

## `Counter`
```python
from collections import Counter
c = Counter(iterable)          # count all items
c['key']                       # get count (returns 0 if missing)
c.most_common(n)               # top n items as [(item, count), ...]
c.update(more_items)           # add more items
list(c.elements())             # expand back to list
c1 + c2  # combine counts
c1 - c2  # subtract counts
c1 & c2  # minimum of shared keys
c1 | c2  # maximum of all keys
```

## `defaultdict`
```python
from collections import defaultdict
d = defaultdict(int)     # missing key → 0
d = defaultdict(list)    # missing key → []
d = defaultdict(set)     # missing key → set()
d = defaultdict(lambda: defaultdict(int))  # nested
d['key'] += 1            # no KeyError!
d['key'].append(val)     # no KeyError!
```

## `deque`
```python
from collections import deque
dq = deque(iterable)           # create
dq = deque(iterable, maxlen=n) # fixed-size window
dq.append(x)                   # add to right
dq.appendleft(x)               # add to left
dq.pop()                       # remove from right
dq.popleft()                   # remove from left O(1)!
dq.rotate(n)                   # rotate right (n>0) or left (n<0)
```

## `namedtuple`
```python
from collections import namedtuple
Point = namedtuple('Point', ['x', 'y'])  # define
p = Point(10, 20)            # create
p.x, p.y                     # access by name
p._asdict()                  # → {'x': 10, 'y': 20}
p._replace(x=99)             # new tuple with updated field
p._fields                    # → ('x', 'y')
```

## `OrderedDict`
```python
from collections import OrderedDict
od = OrderedDict()
od.move_to_end('key')          # move to end
od.move_to_end('key', last=False)  # move to front
od.popitem(last=False)         # remove first item
```

## `ChainMap`
```python
from collections import ChainMap
cm = ChainMap(dict1, dict2, dict3)  # first dict has priority
cm['key']                            # searches dict1 → dict2 → dict3
cm['new'] = val                      # always updates dict1
cm.new_child({'x': 1})              # create child scope
cm.maps                              # list of underlying dicts
```

## When to Use What
| Situation | Use |
|-----------|----|
| Count frequency of items | `Counter` |
| Top N most common items | `Counter.most_common(n)` |
| Dict without KeyError | `defaultdict` |
| Group items into lists | `defaultdict(list)` |
| Group unique items | `defaultdict(set)` |
| Fast queue / stack | `deque` |
| Sliding window (last N items) | `deque(maxlen=n)` |
| Readable lightweight record | `namedtuple` |
| Move dict key to front/back | `OrderedDict` |
| Merge dicts with priority | `ChainMap` |
| Layered config system | `ChainMap` |
