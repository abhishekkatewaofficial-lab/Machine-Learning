```
import math
# Calculate the area and circumference of a circle
radius = 5
area = math.pi * (radius ** 2)
circumference = 2 * math.pi * radius
print(f"Area: {area:.2f}, Circumference: {circumference:.2f}")
```

DSA use: Find all prime factors up to `sqrt(n)`.


```
import math
def prime_factors(n):
    factors = []
    # Count the number of 2s that divide n
    while n % 2 == 0:
        factors.append(2)
        n //= 2
    # n must be odd at this point. Loop from 3 to sqrt(n)
    for i in range(3, math.isqrt(n) + 1, 2):
        while n % i == 0:
            factors.append(i)
            n //= i
    # This condition is to handle the case when n is a prime number greater than 2
    if n > 2:
        factors.append(n)
    return factors

print(f"Prime factors of 315: {prime_factors(315)}")
```

## Section 6: GCD and LCM (NUMBER THEORY — MOST IMPORTANT FOR DSA)
Greatest Common Divisor (GCD) is the largest positive integer that divides each of the integers. Least Common Multiple (LCM) is the smallest positive integer that is divisible by both integers. They are fundamental in number theory problems, simplifying fractions, and finding repeating patterns.


```
import math
# Greatest Common Divisor
print(f"math.gcd(48, 18): {math.gcd(48, 18)}")
```

`math.gcd(*list)` finds the GCD of multiple numbers (Available in Python 3.9+).


```
import math
print(f"math.gcd(48, 18, 30): {math.gcd(48, 18, 30)}")
```

`math.lcm(a, b)` finds the Least Common Multiple (Available in Python 3.9+).


```
import math
print(f"math.lcm(4, 6): {math.lcm(4, 6)}")
```

`math.lcm(*list)` finds the LCM of multiple numbers (Available in Python 3.9+).


```
import math
print(f"math.lcm(4, 6, 8): {math.lcm(4, 6, 8)}")
```

Relationship between GCD and LCM: `LCM(a, b) = a * b // GCD(a, b)`.


```
import math
a, b = 15, 20
lcm_manual = (a * b) // math.gcd(a, b)
print(f"LCM of {a} and {b} calculated manually: {lcm_manual}")
print(f"math.lcm({a}, {b}): {math.lcm(a, b)}")
```

DSA use: Simplify a fraction using GCD.


```
import math
numerator, denominator = 42, 56
gcd_val = math.gcd(numerator, denominator)
simplified = (numerator // gcd_val, denominator // gcd_val)
print(f"Simplified fraction of {numerator}/{denominator}: {simplified[0]}/{simplified[1]}")
```

DSA use: Minimum operations to make all elements equal using GCD.


```
import math
# Example problem: Given array, you can add/subtract 'd' to any element. Find max possible 'd' so all can be equal.
# The answer is the GCD of differences between elements and the minimum element.
arr = [10, 14, 22, 34]
min_val = min(arr)
diffs = [x - min_val for x in arr]
max_d = math.gcd(*diffs)
print(f"Max difference 'd' to make all equal: {max_d}")
```

DSA use: Water Jug Problem using GCD.


```
import math
# Problem: Can you measure exactly z liters using x and y liter jugs?
x, y, z = 3, 5, 4
can_measure = z % math.gcd(x, y) == 0 and z <= max(x, y)
print(f"Can measure {z}L with {x}L and {y}L jugs? {can_measure}")
```

## Section 13: DSA Problem Set with math

Prime checking using `math.isqrt` in O(√n) time.


```
import math
def is_prime(n):
    if n <= 1: return False
    for i in range(2, math.isqrt(n) + 1):
        if n % i == 0: return False
    return True

print(f"Is 29 prime? {is_prime(29)}")
```

Sieve of Eratosthenes setup using `math.isqrt`.


```
import math
def sieve(limit):
    primes = [True] * (limit + 1)
    primes[0] = primes[1] = False
    for p in range(2, math.isqrt(limit) + 1):
        if primes[p]:
            for i in range(p * p, limit + 1, p):
                primes[i] = False
    return [p for p in range(limit + 1) if primes[p]]

print(f"Primes up to 30: {sieve(30)}")
```

Find all divisors of `n` using `math.isqrt`.


```
import math
def find_divisors(n):
    divisors = set()
    for i in range(1, math.isqrt(n) + 1):
        if n % i == 0:
            divisors.add(i)
            divisors.add(n // i)
    return sorted(list(divisors))

print(f"Divisors of 36: {find_divisors(36)}")
```

## Section 14: math vs cmath (brief intro)

```
import math
try:
    math.sqrt(-1)
except ValueError as e:
    print(f"Error: {e}")
```

```
import cmath
complex_root = cmath.sqrt(-1)
print(f"cmath.sqrt(-1): {complex_root}")
```

Use `math` for real numbers and typical DSA problems. Use `cmath` when your domain includes complex numbers (e.g., engineering, advanced physics simulations).

**Exercise 4**: Find all prime factors of n using `math.isqrt` and a loop.


```
import math
n = 84
# your code...
```

**Solution to Exercise 4**:


```
import math
n = 84
factors = []
while n % 2 == 0:
    factors.append(2)
    n //= 2
for i in range(3, math.isqrt(n) + 1, 2):
    while n % i == 0:
        factors.append(i)
        n //= i
if n > 2:
    factors.append(n)
print(f"Prime factors: {factors}")
```

**Exercise 5**: Check if n is a power of 2 using `math.log2`, then find next power of 2 > n.


```
import math
n = 100
# your code...
```

**Solution to Exercise 5**:


```
import math
n = 100
is_pow2 = math.log2(n).is_integer()
print(f"Is {n} a power of 2? {is_pow2}")

next_pow2 = 2 ** math.ceil(math.log2(n)) if not is_pow2 else 2 ** (math.ceil(math.log2(n)) + 1)
# Simply: 2 ** (math.floor(math.log2(n)) + 1)
print(f"Next power of 2 > {n}: {2 ** (math.floor(math.log2(n)) + 1)}")
```
