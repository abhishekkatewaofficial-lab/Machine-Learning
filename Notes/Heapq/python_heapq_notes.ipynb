# Section 1: What is a Heap? 🌳
---

## The Core Idea

A **heap** is a special tree structure with one rule:
> **Every parent node is smaller than (or equal to) its children.**

This means: **the root (top) is ALWAYS the smallest element.**

```
           MIN-HEAP                      MAX-HEAP
              1                             9
           /     \                       /     \
          3       2                     7       8
         / \     / \                   / \     / \
        6   5   4   8                 3   4   5   1

  Parent always ≤ Children        Parent always ≥ Children
  Root = SMALLEST (1)             Root = LARGEST (9)
```

## Python's `heapq` is always a MIN-HEAP
- `heappop()` always gives you the **smallest** element
- To get the largest, we use a trick (Section 4)

## Why Not Just Sort?
| Operation | Sorting | Heap |
|-----------|---------|------|
| Find minimum | Sort → take first: **O(n log n)** | `heappop()`: **O(log n)** ⚡ |
| Add new item | Re-sort: **O(n log n)** | `heappush()`: **O(log n)** ⚡ |
| Top-N from huge data | Sort all: **O(n log n)** | `nsmallest(N)`: **O(n log N)** ⚡ |
| Memory | Full sorted list | Only O(N) for N items |

**Rule of thumb:**
- Finding top-N from huge data → use `heapq`
- Sorting a small list → use `sorted()` (simpler)

## How Python Stores a Heap — It's Just a List!

Python doesn't use a tree object. It stores the heap as a **regular list**
with a clever index rule:

```
heap = [1, 3, 2, 6, 5, 4, 8]

Index:   0   1   2   3   4   5   6
Value:   1   3   2   6   5   4   8

For item at index i:
  Left child  → index 2*i + 1
  Right child → index 2*i + 2
  Parent      → index (i-1) // 2
```
The heap property is maintained automatically by `heappush` and `heappop`.

# Section 2: Basic Operations 🟢
---

### `heapify(list)` — Convert any list into a heap IN-PLACE — O(n)

````python
import heapq

# heapify converts a regular list into a min-heap
numbers = [5, 3, 8, 1, 9, 2, 7, 4, 6]
print('Before heapify:', numbers)

heapq.heapify(numbers)   # modifies list IN-PLACE
print('After heapify :', numbers)
print('Smallest item (heap[0]):', numbers[0])  # Always the minimum!
````

````python
import heapq

# Verify the heap property: parent <= children
heap = [1, 3, 2, 6, 5, 4, 8]

print('Heap:', heap)
print()
print('Heap structure visualization:')
print(f'          {heap[0]}')
print(f'        /   \\')
print(f'      {heap[1]}       {heap[2]}')
print(f'     / \\     / \\')
print(f'   {heap[3]}   {heap[4]}   {heap[5]}   {heap[6]}')
````

### `heappush(heap, item)` — Add an item while maintaining heap property — O(log n)

````python
import heapq

heap = []
items = [5, 3, 8, 1, 9, 2]

print('Pushing items one by one:')
for item in items:
    heapq.heappush(heap, item)
    print(f'  Push {item} → heap: {heap}  | min = {heap[0]}')
````

````python
import heapq

# Stream of incoming numbers — always track current minimum
stream  = [42, 15, 73, 8, 56, 31, 99, 4, 22]
heap    = []

print('Tracking minimum in a live data stream:')
for num in stream:
    heapq.heappush(heap, num)
    print(f'  Received: {num:3d} | Current minimum: {heap[0]}')
````

### `heappop(heap)` — Remove and return the SMALLEST item — O(log n)

````python
import heapq

heap = [1, 3, 2, 6, 5, 4, 8]
heapq.heapify(heap)

print('Popping items in sorted order (heap sort!):')
while heap:
    item = heapq.heappop(heap)
    print(f'  Popped: {item} | Remaining heap: {heap}')
````

````python
import heapq

# Process customer tickets by urgency (lower number = more urgent)
tickets = [3, 1, 5, 2, 4]   # urgency levels
heapq.heapify(tickets)

print('Processing tickets by urgency (1=most urgent, 5=least):')
order = 1
while tickets:
    urgency = heapq.heappop(tickets)
    print(f'  Serving ticket #{order}: Urgency level {urgency}')
    order += 1
````

### `heappushpop(heap, item)` — Push then Pop in ONE step (faster than doing both)

````python
import heapq

# heappushpop: push item THEN pop smallest
# More efficient than heappush + heappop separately

heap = [1, 3, 5, 7, 9]
heapq.heapify(heap)

print('heap before         :', heap)

result = heapq.heappushpop(heap, 4)  # push 4, then pop smallest
print(f'heappushpop(heap, 4): popped {result}')
print('heap after          :', heap)
````

### `heapreplace(heap, item)` — Pop then Push in ONE step (faster, but heap must not be empty)

````python
import heapq

# heapreplace: pop smallest THEN push new item
# Use when you want to replace the minimum with a new value

heap = [1, 3, 5, 7, 9]
heapq.heapify(heap)

print('heap before          :', heap)

result = heapq.heapreplace(heap, 4)  # pop smallest, then push 4
print(f'heapreplace(heap, 4): popped {result}')
print('heap after           :', heap)
````

## The Two Functions You'll Use 90% of the Time

```python
heapq.nlargest(n, iterable, key=None)   # top n LARGEST items
heapq.nsmallest(n, iterable, key=None) # top n SMALLEST items
```

These are **much faster** than sorting when N is much smaller than the total data size.

### `nsmallest(n, iterable)` — Find N smallest items

````python
import heapq

# Find 3 smallest numbers from a list
numbers = [15, 3, 42, 8, 27, 1, 56, 12, 9, 33]

result = heapq.nsmallest(3, numbers)
print('Numbers :', numbers)
print('3 Smallest:', result)
````

````python
import heapq

# Cheapest 3 products from a catalogue
prices = [999, 299, 4999, 149, 799, 2499, 99, 1599]

cheapest = heapq.nsmallest(3, prices)
print('All prices (unsorted):', prices)
print('3 Cheapest products  :', cheapest)
````

### `nlargest(n, iterable)` — Find N largest items

````python
import heapq

# Top 3 scores in a game
scores = [45, 82, 91, 67, 55, 98, 73, 88, 61, 79]

top3 = heapq.nlargest(3, scores)
print('All scores:', scores)
print('Top 3     :', top3)
````

````python
import heapq

# Top 5 most common characters in a text
from collections import Counter

text    = 'the quick brown fox jumps over the lazy dog'
freq    = Counter(c for c in text if c != ' ')

top5 = heapq.nlargest(5, freq.items(), key=lambda x: x[1])
print('Top 5 most common characters:')
for char, count in top5:
    print(f'  {char!r}: {"█" * count} ({count})')
````

# Section 4: Max Heap — The Trick ⚡
---

## Python Only Has Min-Heap

By default, `heappop()` always gives the **smallest** value.
To get the **largest** value, we use a simple trick:

### The Trick: **Negate the values** (multiply by -1)
```
Original: [5, 3, 8, 1]   → Min-heap: pop gives 1  (smallest)
Negated:  [-5,-3,-8,-1]  → Min-heap: pop gives -8 (which is really 8 — largest!)
```

Push `-value`, pop and negate back to get the max.

````python
import heapq

# Max heap using negation trick
numbers = [5, 3, 8, 1, 9, 2, 7]
max_heap = []

# Push negated values
for num in numbers:
    heapq.heappush(max_heap, -num)  # store negative!

print('Numbers  :', numbers)
print('Max heap (negated internally):', max_heap)

# Pop — negate to get real value
print('\nPopping in LARGEST first order:')
while max_heap:
    largest = -heapq.heappop(max_heap)  # negate back!
    print(f'  Largest: {largest}')
````

````python
import heapq

# Track running maximum of student scores using max heap
scores  = [65, 90, 78, 55, 95, 88, 72]
max_heap = []

print('Tracking maximum score:')
for score in scores:
    heapq.heappush(max_heap, -score)
    current_max = -max_heap[0]  # peek at top without popping
    print(f'  Score: {score:3d} | Running max: {current_max}')
````

````python
import heapq

# Keep track of top 3 highest salaries using max heap trick
salaries  = [50000, 75000, 45000, 90000, 60000, 85000, 70000]
max_heap  = [-s for s in salaries]
heapq.heapify(max_heap)

top3 = []
for _ in range(3):
    top3.append(-heapq.heappop(max_heap))

print('All salaries:', salaries)
print('Top 3 highest:', top3)
````

# Section 5: Priority Queue with Tuples 🟡
---

## Using `(priority, item)` Tuples

When you push **tuples** into a heap, Python sorts by the **first element** first.
This is how you build a **Priority Queue** — a queue where items are processed by priority, not arrival order.

```python
heap = []
heapq.heappush(heap, (1, 'high priority task'))
heapq.heappush(heap, (3, 'low priority task'))
heapq.heappush(heap, (2, 'medium priority task'))
# heappop gives (1, 'high priority task') first!
```

**Lower number = Higher priority** (because it's a min-heap)

````python
import heapq

# Basic priority queue — tasks with priority levels
task_heap = []

tasks = [
    (3, 'Write documentation'),
    (1, 'Fix critical bug'),
    (2, 'Code review'),
    (1, 'Deploy hotfix'),
    (4, 'Team meeting'),
]

for priority, task in tasks:
    heapq.heappush(task_heap, (priority, task))

print('Processing tasks by priority (1=most urgent):')
while task_heap:
    priority, task = heapq.heappop(task_heap)
    print(f'  [P{priority}] {task}')
````

````python
import heapq

# Hospital emergency queue — lower number = more serious
emergency_queue = []

patients = [
    (3, 'Alice',   'Mild fever'),
    (1, 'Bob',     'Heart attack'),
    (2, 'Charlie', 'Broken arm'),
    (1, 'Diana',   'Stroke'),
    (4, 'Eve',     'Minor cut'),
]

for severity, name, condition in patients:
    heapq.heappush(emergency_queue, (severity, name, condition))

print('ER patients — Treating in order of severity:')
order = 1
while emergency_queue:
    severity, name, condition = heapq.heappop(emergency_queue)
    print(f'  #{order} Severity[{severity}] {name:8}: {condition}')
    order += 1
````

### `(priority, counter, item)` — Breaking Ties (Same Priority)

## The Tie-Breaking Problem

When two items have the **same priority**, Python tries to compare the next element.
If items are dicts or non-comparable objects, this causes a **TypeError**!

```python
# ❌ PROBLEM:
heapq.heappush(heap, (1, {'task': 'A'}))  # dicts can't be compared!
heapq.heappush(heap, (1, {'task': 'B'}))  # same priority → tries to compare dicts → TypeError!

# ✅ SOLUTION: Add a unique counter as the 2nd element
heapq.heappush(heap, (1, 0, {'task': 'A'}))  # counter breaks the tie
heapq.heappush(heap, (1, 1, {'task': 'B'}))  # counter is always unique
```

````python
import heapq, itertools

# Using a counter to avoid comparison issues with same-priority items
counter = itertools.count()   # auto-incrementing unique ID
task_heap = []

tasks = [
    (1, {'name': 'Fix bug A',   'team': 'Backend'}),
    (1, {'name': 'Fix bug B',   'team': 'Frontend'}),  # same priority as above
    (2, {'name': 'Write tests', 'team': 'QA'}),
    (1, {'name': 'Hotfix C',    'team': 'Backend'}),   # same priority again
]

for priority, task in tasks:
    heapq.heappush(task_heap, (priority, next(counter), task))  # counter breaks tie

print('Processing tasks (safe tie-breaking):')
while task_heap:
    priority, _, task = heapq.heappop(task_heap)
    print(f'  [P{priority}] {task["name"]:20} ({task["team"]})')
````

# Section 6: Deep Dive — Strings with `heapq` 🟢🟡
---

````python
import heapq

# Find K least frequent characters in a string
from collections import Counter

text    = 'programming in python is powerful and practical'
freq    = Counter(c for c in text if c != ' ')

least_3 = heapq.nsmallest(3, freq.items(), key=lambda x: x[1])
print('3 least common characters:')
for char, count in least_3:
    print(f'  {char!r}: {count} time(s)')
````

````python
import heapq

# Sort words by length using heapq (shortest first)
words = ['python', 'is', 'a', 'powerful', 'and', 'easy', 'programming', 'language']

word_heap = []
for word in words:
    heapq.heappush(word_heap, (len(word), word))  # (length, word)

print('Words sorted by length (shortest first):')
while word_heap:
    length, word = heapq.heappop(word_heap)
    print(f'  {word:15} (len={length})')
````

````python
import heapq

# Priority queue of messages — process most urgent first
messages = [
    (1, 'CRITICAL: Server is down!'),
    (3, 'INFO: Backup completed'),
    (2, 'WARNING: Disk 80% full'),
    (1, 'CRITICAL: Database timeout'),
    (4, 'DEBUG: User logged in'),
]

heap = []
for msg in messages:
    heapq.heappush(heap, msg)

print('Processing messages by priority:')
while heap:
    priority, message = heapq.heappop(heap)
    print(f'  [{priority}] {message}')
````

````python
import heapq

# K most frequent words in a text (word frequency + heapq)
from collections import Counter

text  = 'I love python python is great I love coding in python coding is fun'
freq  = Counter(text.split())

top4 = heapq.nlargest(4, freq.items(), key=lambda x: x[1])
print(f'Text: "{text}"')
print('\nTop 4 most frequent words:')
for word, count in top4:
    print(f'  {word!r:12}: {"█" * count} ({count})')
````

# Section 7: Deep Dive — Lists with `heapq` 🟢🟡
---

````python
import heapq

# nsmallest — top N from a large unsorted list
import random
random.seed(42)
big_list = [random.randint(1, 1000) for _ in range(50)]

smallest5 = heapq.nsmallest(5, big_list)
largest5  = heapq.nlargest(5, big_list)

print(f'List has {len(big_list)} items')
print('5 Smallest:', smallest5)
print('5 Largest :', largest5)
````

````python
import heapq

# K-th smallest element — find the 4th smallest number
numbers = [7, 10, 4, 3, 20, 15, 8, 2, 12, 9]
k       = 4

kth_smallest = heapq.nsmallest(k, numbers)[-1]  # last of top-k smallest
print(f'Numbers: {numbers}')
print(f'{k}th smallest element: {kth_smallest}')
````

````python
import heapq

# Heap Sort — sort a list using heapq
numbers    = [5, 3, 8, 1, 9, 2, 7, 4, 6]
heap       = numbers[:]
heapq.heapify(heap)

sorted_list = []
while heap:
    sorted_list.append(heapq.heappop(heap))

print('Original   :', numbers)
print('Heap Sorted:', sorted_list)
````

````python
import heapq

# heapq.merge() — merge multiple SORTED iterables into one sorted stream
# Does NOT require loading everything into memory!

list1 = [1, 4, 7, 10]
list2 = [2, 5, 8, 11]
list3 = [3, 6, 9, 12]

merged = list(heapq.merge(list1, list2, list3))
print('List 1:', list1)
print('List 2:', list2)
print('List 3:', list3)
print('Merged :', merged)
````

````python
import heapq

# Merge exam results from 3 different classrooms — all pre-sorted
class_a = [55, 68, 75, 88, 92]
class_b = [60, 72, 80, 85, 95]
class_c = [50, 65, 70, 78, 90]

all_scores = list(heapq.merge(class_a, class_b, class_c))
print('All scores merged (sorted):', all_scores)
print(f'Total students : {len(all_scores)}')
print(f'Top 3 scorers  : {heapq.nlargest(3, all_scores)}')
print(f'Class average  : {sum(all_scores)/len(all_scores):.1f}')
````

````python
import heapq

# Track running minimum using a heap — useful for data streams
data    = [42, 15, 73, 8, 56, 31, 99, 4, 22, 67]
heap    = []
running_mins = []

for val in data:
    heapq.heappush(heap, val)
    running_mins.append(heap[0])  # heap[0] is always the minimum

print('Data stream   :', data)
print('Running minimum:', running_mins)
````

# Section 8: Deep Dive — Dictionaries with `heapq` 🟡

````python
import heapq

# Top N values from a dictionary
stock = {'apple': 50, 'banana': 30, 'cherry': 80, 'mango': 20, 'grape': 60}

top3_items = heapq.nlargest(3, stock.items(), key=lambda x: x[1])
print('Stock quantities:')
for item, qty in stock.items():
    print(f'  {item:10}: {qty}')
print('\nTop 3 highest stock:')
for item, qty in top3_items:
    print(f'  {item:10}: {qty}')
````

````python
import heapq

# Find the 3 lowest-paid employees from a dict
salaries = {
    'Alice':   90000,
    'Bob':     65000,
    'Charlie': 75000,
    'Diana':   55000,
    'Eve':     85000,
    'Frank':   70000,
}

lowest3 = heapq.nsmallest(3, salaries.items(), key=lambda x: x[1])
print('3 Lowest Paid Employees:')
for name, salary in lowest3:
    print(f'  {name:10}: ₹{salary:,}')
````

````python
import heapq

# Task manager — dict-based tasks in a priority queue
tasks = {
    'task_1': {'name': 'Deploy hotfix',    'priority': 1, 'effort': 'low'},
    'task_2': {'name': 'Write blog post',  'priority': 4, 'effort': 'medium'},
    'task_3': {'name': 'Fix login bug',    'priority': 2, 'effort': 'high'},
    'task_4': {'name': 'Update homepage',  'priority': 3, 'effort': 'low'},
    'task_5': {'name': 'Patch security',   'priority': 1, 'effort': 'medium'},
}

task_heap = []
for tid, task in tasks.items():
    heapq.heappush(task_heap, (task['priority'], tid, task))

print('Tasks processed by priority:')
while task_heap:
    priority, tid, task = heapq.heappop(task_heap)
    print(f'  [P{priority}] {task["name"]:25} | Effort: {task["effort"]}')
````

````python
import heapq

# Top 3 products by rating AND find bottom 2 by rating
ratings = {
    'Laptop':   4.8,
    'Phone':    4.5,
    'Tablet':   3.9,
    'Watch':    4.2,
    'Earbuds':  4.7,
    'Charger':  3.5,
}

top3    = heapq.nlargest(3,  ratings.items(), key=lambda x: x[1])
bottom2 = heapq.nsmallest(2, ratings.items(), key=lambda x: x[1])

print('⭐ Top 3 rated:')
for name, rating in top3:
    print(f'  {name:10}: {"★" * int(rating)} ({rating})')

print('\n👎 Bottom 2 rated:')
for name, rating in bottom2:
    print(f'  {name:10}: {rating}')
````

````python
import heapq, itertools

# Priority queue of nested dict records (with tie-breaking counter)
counter = itertools.count()
pq      = []

orders = [
    {'order_id': 'ORD001', 'customer': 'Alice',   'priority': 2, 'amount': 5000},
    {'order_id': 'ORD002', 'customer': 'Bob',     'priority': 1, 'amount': 8000},
    {'order_id': 'ORD003', 'customer': 'Charlie', 'priority': 1, 'amount': 3000},
    {'order_id': 'ORD004', 'customer': 'Diana',   'priority': 3, 'amount': 1500},
    {'order_id': 'ORD005', 'customer': 'Eve',     'priority': 2, 'amount': 7000},
]

for order in orders:
    heapq.heappush(pq, (order['priority'], next(counter), order))

print('Processing orders by priority:')
while pq:
    priority, _, order = heapq.heappop(pq)
    print(f'  [P{priority}] {order["order_id"]} | {order["customer"]:10} | ₹{order["amount"]:,}')
````

````python
import heapq

# Top employee from each department (multi-level nested)
company = {
    'Engineering': [
        {'name': 'Alice',   'salary': 90000},
        {'name': 'Charlie', 'salary': 85000},
        {'name': 'Frank',   'salary': 95000},
    ],
    'Marketing': [
        {'name': 'Bob',   'salary': 75000},
        {'name': 'Eve',   'salary': 80000},
    ],
    'HR': [
        {'name': 'Diana', 'salary': 70000},
    ],
}

print('Highest paid employee per department:')
for dept, employees in company.items():
    top_emp = heapq.nlargest(1, employees, key=lambda e: e['salary'])[0]
    print(f'  {dept:15}: {top_emp["name"]:10} → ₹{top_emp["salary"]:,}')
````

# Section 10: Real-World Patterns 🔴

### Pattern 1: Dijkstra's Shortest Path (Simplified)

````python
import heapq

# Find shortest path from 'A' to all nodes
# Graph: {node: [(cost, neighbor), ...]}
graph = {
    'A': [(1, 'B'), (4, 'C')],
    'B': [(2, 'C'), (5, 'D')],
    'C': [(1, 'D')],
    'D': []
}

def dijkstra(graph, start):
    dist = {node: float('inf') for node in graph}
    dist[start] = 0
    heap = [(0, start)]  # (distance, node)

    while heap:
        cost, node = heapq.heappop(heap)
        if cost > dist[node]:
            continue  # skip outdated entries
        for edge_cost, neighbor in graph[node]:
            new_cost = cost + edge_cost
            if new_cost < dist[neighbor]:
                dist[neighbor] = new_cost
                heapq.heappush(heap, (new_cost, neighbor))
    return dist

distances = dijkstra(graph, 'A')
print('Shortest distances from A:')
for node, dist in distances.items():
    print(f'  A → {node}: {dist}')
````

### Pattern 3: Top K Items from a Data Stream (Memory Efficient)

````python
import heapq

# Keep top-3 scores using a min-heap of size K
# This only uses O(K) memory no matter how many items come in!

def top_k_stream(stream, k):
    """Find top K largest from a stream using only O(K) memory."""
    heap = []  # min-heap of size K
    for item in stream:
        heapq.heappush(heap, item)
        if len(heap) > k:
            heapq.heappop(heap)  # remove smallest — keeps only top K!
    return sorted(heap, reverse=True)

# Simulate a stream of 20 exam scores
import random
random.seed(10)
score_stream = [random.randint(40, 100) for _ in range(20)]

print('Score stream:', score_stream)
print('Top 5 scores:', top_k_stream(score_stream, 5))
````

# Section 11: `heapq` vs `sorted()` vs `Counter` — When to Use Which ⚡
---

## Decision Guide

| Situation | Best Tool | Why |
|-----------|-----------|-----|
| Find top-N from huge data (N << total) | `heapq.nlargest/nsmallest` | O(n log N) — very fast |
| Sort a small list | `sorted()` | Simpler, readable |
| Top-N from list of dicts by field | `heapq.nlargest(N, data, key=...)` | Built-in key support |
| Count frequency + top-N | `Counter.most_common(N)` | Counting + ranking in one |
| Process items by priority | `heapq` priority queue | O(log n) per operation |
| Merge sorted lists | `heapq.merge()` | Memory efficient |

### Rule of Thumb:
- If N is **close to total size** → use `sorted()`
- If N is **much smaller** than total → use `heapq`
- If you need **both counting and ranking** → use `Counter.most_common()`

````python
import heapq, timeit, random, matplotlib.pyplot as plt
from collections import Counter

random.seed(42)
data = [random.randint(1, 10000) for _ in range(100_000)]
N    = 10   # find top 10

def using_sorted():
    return sorted(data, reverse=True)[:N]

def using_heapq():
    return heapq.nlargest(N, data)

t_sorted = timeit.timeit(using_sorted, number=50) / 50 * 1000
t_heapq  = timeit.timeit(using_heapq,  number=50) / 50 * 1000

methods  = [f'sorted()\n(top {N} from 100k)', f'heapq.nlargest()\n(top {N} from 100k)']
times    = [t_sorted, t_heapq]
colors   = ['tomato', 'steelblue']

plt.figure(figsize=(8, 5))
bars = plt.bar(methods, times, color=colors, width=0.4)
for bar, t in zip(bars, times):
    plt.text(bar.get_x() + bar.get_width()/2, bar.get_height() + 0.05,
             f'{t:.2f} ms', ha='center', fontweight='bold', fontsize=12)
plt.title(f'Performance: sorted() vs heapq.nlargest() for top-{N} from 100,000 items',
          fontsize=11, fontweight='bold')
plt.ylabel('Time (milliseconds)')
plt.tight_layout()
plt.show()
print(f'sorted()         : {t_sorted:.2f} ms')
print(f'heapq.nlargest() : {t_heapq:.2f} ms')
print(f'heapq is {t_sorted/t_heapq:.1f}x faster for top-{N} from 100k items!')
````

# Section 12: ⚡ `heapq` Cheat Sheet
---

## All Functions

```python
import heapq

# === SETUP ===
heap = []
heapq.heapify(list)                   # convert list → min-heap in-place O(n)

# === ADD / REMOVE ===
heapq.heappush(heap, item)            # add item, maintain heap O(log n)
item = heapq.heappop(heap)            # remove & return smallest O(log n)
item = heap[0]                        # PEEK at smallest (no removal)
item = heapq.heappushpop(heap, item)  # push then pop (atomic, faster)
item = heapq.heapreplace(heap, item)  # pop then push (atomic, faster)

# === TOP-N ===
heapq.nlargest(n, iterable)           # n largest items
heapq.nlargest(n, iterable, key=f)   # n largest by custom key
heapq.nsmallest(n, iterable)          # n smallest items
heapq.nsmallest(n, iterable, key=f)  # n smallest by custom key

# === MERGE ===
heapq.merge(*sorted_iterables)        # merge sorted iterables
```

## Common Patterns

```python
# Min Heap
heap = []; heapq.heappush(heap, x)
smallest = heapq.heappop(heap)

# Max Heap (negate trick)
heapq.heappush(heap, -x)
largest = -heapq.heappop(heap)

# Priority Queue
heapq.heappush(heap, (priority, item))
priority, item = heapq.heappop(heap)

# Safe tie-breaking
import itertools
counter = itertools.count()
heapq.heappush(heap, (priority, next(counter), item))

# Top N from list of dicts
heapq.nlargest(3, data, key=lambda d: d['field'])

# K-th smallest
heapq.nsmallest(k, data)[-1]

# Top-K stream (O(K) memory)
for item in stream:
    heapq.heappush(heap, item)
    if len(heap) > k:
        heapq.heappop(heap)
```

## When to Use
| Use Case | Tool |
|----------|------|
| Top-N from big data | `nlargest` / `nsmallest` |
| Process by priority | `(priority, item)` heap |
| Streaming minimum | `heappush` + `heap[0]` |
| Merge sorted lists | `heapq.merge()` |
| Max heap | Negate values trick |
| Safe priority queue | `(priority, counter, item)` |
