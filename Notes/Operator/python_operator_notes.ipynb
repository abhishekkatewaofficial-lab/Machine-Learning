# Python `operator` Module — Complete Notes (DSA Edition)

## Section 1: Why operator module?

operator module gives you functions for all Python operators. Instead of writing `lambda x,y: x+y`, use `operator.add`. Functions are faster than lambdas and more readable.

Multiplication

```
print(operator.mul(10, 5))
```

## Section 5: itemgetter — MOST USED IN DSA SORTING

itemgetter(key) creates a function that extracts item at key/index. Much faster than lambda for sorting.

Extract first element of tuple/list

```
get_first = operator.itemgetter(0)
print(get_first([10, 20, 30]))
```

Extract second element

```
get_second = operator.itemgetter(1)
print(get_second([10, 20, 30]))
```

Extract dict value by key

```
get_name = operator.itemgetter('name')
print(get_name({'name': 'Alice', 'age': 25}))
```

Sort list of tuples by second element using itemgetter(1)

```
pairs = [(1, 3), (3, 2), (2, 1)]
print(sorted(pairs, key=operator.itemgetter(1)))
```

Sort list of tuples by multiple keys: itemgetter(1, 0)

```
pairs = [(1, 2), (3, 2), (2, 1)]
print(sorted(pairs, key=operator.itemgetter(1, 0)))
```

Sort list of dicts by key using itemgetter('score')

```
dicts = [{'name': 'A', 'score': 90}, {'name': 'B', 'score': 100}]
print(sorted(dicts, key=operator.itemgetter('score')))
```

Sort list of dicts by multiple keys: itemgetter('grade', 'name')

```
dicts = [{'name': 'B', 'grade': 'A'}, {'name': 'A', 'grade': 'A'}]
print(sorted(dicts, key=operator.itemgetter('grade', 'name')))
```

DSA — sort (score, name) pairs by score descending: key=itemgetter(0) with reverse=True

```
students = [(90, 'Alice'), (100, 'Bob')]
print(sorted(students, key=operator.itemgetter(0), reverse=True))
```

DSA — get minimum element from list of tuples by field

```
print(min(students, key=operator.itemgetter(0)))
```

Performance: lambda vs itemgetter — timeit benchmark + matplotlib chart

```
import timeit
setup = 'import operator; data = [(i, -i) for i in range(1000)]'
print('Lambda:', timeit.timeit('sorted(data, key=lambda x: x[1])', setup=setup, number=100))
print('Itemgetter:', timeit.timeit('sorted(data, key=operator.itemgetter(1))', setup=setup, number=100))
```

## Section 6: attrgetter — For Objects/Classes

Define a simple Student class with name, score, grade

```
class Student:
    def __init__(self, name, score, grade):
        self.name = name
        self.score = score
        self.grade = grade
    def __repr__(self):
        return f'{self.name}({self.score})'

students = [Student('Alice', 90, 'A'), Student('Bob', 80, 'B')]
```

attrgetter('score') — extract attribute from object

```
print(operator.attrgetter('score')(students[0]))
```

Sort list of objects by attribute using attrgetter

```
print(sorted(students, key=operator.attrgetter('score')))
```

Nested attribute access (dotted)

```
# e.g., operator.attrgetter('address.city')
```

## Section 7: methodcaller

Call method on object

```
print(operator.methodcaller('upper')('hello'))
```

Call method with arguments

```
print(operator.methodcaller('strip', '*')('**hello**'))
```

Use with map to apply method to all items in list

```
words = ['  a  ', 'b  ', ' c']
print(list(map(operator.methodcaller('strip'), words)))
```

DSA use — normalize all strings in a list using methodcaller

```
print(list(map(operator.methodcaller('lower'), ['A', 'B', 'C'])))
```
